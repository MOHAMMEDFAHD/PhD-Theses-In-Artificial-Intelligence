# Academic Summary: Learning Deep Generative Models

## Abstract
This doctoral dissertation investigates the principles, optimization algorithms, quantitative evaluation protocols, and practical applications of deep probabilistic generative models with many layers of latent variables[cite: 275, 279, 344]. The research demonstrates that deep hierarchical architectures containing millions of parameters can be learned efficiently from unlabeled data, overcoming the non-convex optimization plateaus and poor local optima that historically caused randomly initialized backpropagation to fail[cite: 275, 279, 280, 344]. The thesis is organized into two primary components: (1) **Deep Belief Networks (DBNs)**, examining greedy layer-wise Restricted Boltzmann Machine (RBM) pretraining, applications in kernel learning for Gaussian processes, deep autoencoder dimensionality reduction, fast document retrieval via Semantic Hashing, non-linear metric learning (Nonlinear NCA), and rigorous partition function estimation via Annealed Importance Sampling (AIS)[cite: 275, 280, 281, 282, 316, 344]; and (2) **Deep Boltzmann Machines (DBMs)**, introducing an undirected hierarchical Markov random field that incorporates top-down feedback alongside bottom-up signals to resolve ambiguous sensory inputs[cite: 275, 280, 282, 328, 345]. DBM training is achieved through an integration of mean-field variational inference, a Robbins-Monro Markov Chain Monte Carlo (MCMC) stochastic approximation procedure, and a doubled-weight greedy pretraining heuristic, establishing state-of-the-art generative and discriminative performance across benchmark datasets[cite: 281, 282, 328, 335, 336, 339, 345].

---

## Problems
- **Inadequacy of Shallow Architectures:** Classical machine learning systems (e.g., standard Support Vector Machines, kernel regressors, and single-hidden-layer networks) rely on shallow formulations whose internal representations cannot compactly extract high-order, non-linear combinatorial structures present in rich sensory data without requiring exponential numbers of units or extensive manual feature engineering[cite: 279].
- **Failure of Direct Gradient Descent in Deep Networks:** While multi-layer feedforward networks possess universal representation capabilities, optimizing them via backpropagation from small, random initializations fails in practice because gradient descent gets trapped in poor local optima or vast plateaus across non-convex loss surfaces[cite: 279].
- **Dependence on Scarcely Available Labeled Data:** Supervised models cannot leverage the vast supplies of unlabeled data; standard discriminative classifiers fail when labeled samples are limited[cite: 279, 280, 293].
- **Intractability of Normalized Probabilistic Inference:** Computing exact posterior distributions over latent variables in directed belief networks is intractable due to "explaining away"[cite: 305]. Conversely, in undirected Markov random fields, computing the partition function $\mathcal{Z}(\theta)$ involves summing over an exponential state space ($\mathcal{O}(2^{\min(D, F)})$), rendering exact maximum likelihood learning intractable and hindering model evaluation[cite: 284, 285, 316, 328, 330].
- **Unidirectional Information Flow in Deep Belief Networks:** Standard DBNs form directed belief networks in their lower layers, restricting approximate inference to a purely feedforward bottom-up pass and precluding top-down contextual signals from disambiguating low-level sensory evidence[cite: 275, 280, 328, 334].

---

## Proposed Solutions
- **Greedy Layer-by-Layer DBN Pretraining:** Trains successive layers of Restricted Boltzmann Machines (RBMs) sequentially on the aggregated posterior activations of the preceding layer, progressively improving a variational lower bound on data log-likelihood and initializing deep architectures in regions of parameter space favorable for local optimization[cite: 279, 280, 285, 286, 287].
- **Specialized Exponential Family RBMs:** Extends binary RBMs to continuous real-valued inputs via Gaussian RBMs and to variable-length text corpora via Replicated Softmax models with document-length-scaled biases[cite: 289, 291, 292].
- **Task-Specific DBN Feature Exploitation:**
  - *GP-DBN:* Employs deep representations as an adaptive covariance kernel for Gaussian process regression and classification[cite: 280, 281, 293, 294, 295].
  - *Deep Autoencoders:* Unrolls pretrained RBM stacks into symmetric encoder-decoder networks fine-tuned via backpropagation for non-linear dimensionality reduction[cite: 281, 300].
  - *Semantic Hashing:* Forces autoencoder bottleneck codes into compact binary addresses using deterministic Gaussian input noise, mapping documents into a semantic Hamming space searchable in $\mathcal{O}(1)$ time[cite: 281, 305, 306, 307].
  - *Nonlinear & Regularized NCA:* Fine-tunes deep encoders under a Neighborhood Component Analysis (NCA) objective combined with autoencoder reconstruction penalties to optimize $K$-nearest neighbor (KNN) classification in semi-supervised regimes[cite: 281, 310, 311, 313].
- **Annealed Importance Sampling (AIS) for Density Estimation:** Formulates a sequence of intermediate probability distributions over the bipartite graph of an RBM to derive unbiased estimators and variational lower bounds on DBN data log-probabilities[cite: 281, 282, 316, 317, 318, 320, 321].
- **Deep Boltzmann Machine (DBM) Framework:** Implements a fully undirected deep Markov random field supporting bidirectional message passing[cite: 275, 282, 328, 333]. Models are trained using a hybrid algorithm combining mean-field variational inference for data-dependent statistics and an MCMC stochastic approximation procedure (Robbins-Monro persistent chains) for model expectations, initialized via a modified layer-wise pretraining scheme with tied and doubled weights[cite: 281, 282, 328, 330, 332, 334, 335, 336].

---

## Purpose
The primary objective of this dissertation is to prove that deep generative architectures with millions of parameters and multiple latent layers can be trained efficiently in an unsupervised manner from high-dimensional, unlabeled sensory inputs[cite: 275, 279, 280, 344]. It seeks to demonstrate that the resulting internal feature hierarchies extract general, domain-invariant representations that transfer to discriminative, retrieval, and regression tasks, while formalizing rigorous statistical methodologies to evaluate deep networks as generative density models[cite: 275, 280, 282, 316, 344].

## Methodology

### 1. Mathematical Formulations

#### Restricted Boltzmann Machine (RBM)
The energy of state $\{v, h\}$ with visible units $v \in \{0, 1\}^D$ and hidden units $h \in \{0, 1\}^F$ is defined as[cite: 283]:
$$E(v, h; \theta) = -v^\top W h - b^\top v - a^\top h \quad[cite: 283]$$
Marginalizing out hidden variables yields the data distribution and exact conditional distributions[cite: 284]:
$$P(v; \theta) = \frac{1}{\mathcal{Z}(\theta)} \sum_h \exp(-E(v, h; \theta)) = \frac{1}{\mathcal{Z}(\theta)} \exp(b^\top v) \prod_{j=1}^F \left(1 + \exp\left(a_j + \sum_{i=1}^D W_{ij} v_i\right)\right) \quad[cite: 284]$$
$$p(h_j = 1 \mid v) = \sigma\left(\sum_{i=1}^D W_{ij} v_i + a_j\right), \quad p(v_i = 1 \mid h) = \sigma\left(\sum_{j=1}^F W_{ij} h_j + b_i\right) \quad[cite: 284]$$
where $\sigma(x) = \frac{1}{1 + \exp(-x)}$ is the logistic sigmoid[cite: 284].
Parameter updates follow Contrastive Divergence ($CD_T$)[cite: 285]:
$$\Delta W = \alpha \left( \mathbb{E}_{P_{\text{data}}}[v h^\top] - \mathbb{E}_{P_T}[v h^\top] \right) \quad[cite: 285]$$

#### Gaussian RBM and Replicated Softmax Extensions
- **Gaussian RBM (Continuous Real-Valued Data):**
  $$E(v, h; \theta) = \sum_{i=1}^D \frac{(v_i - b_i)^2}{2\sigma_i^2} - \sum_{i=1}^D \sum_{j=1}^F W_{ij} h_j \frac{v_i}{\sigma_i} - \sum_{j=1}^F a_j h_j \quad[cite: 289]$$
  $$p(v_i = x \mid h) = \mathcal{N}\left(b_i + \sigma_i \sum_{j=1}^F h_j W_{ij}, \, \sigma_i^2\right), \quad p(h_j = 1 \mid v) = \sigma\left(a_j + \sum_{i=1}^D W_{ij} \frac{v_i}{\sigma_i}\right) \quad[cite: 290]$$
- **Replicated Softmax (Word Count Vectors):**
  For document length $M$ and word count vector $\hat{v}$ over dictionary size $K$[cite: 291]:
  $$E(V, h) = -\sum_{j=1}^F \sum_{k=1}^K W_j^k h_j \hat{v}^k - \sum_{k=1}^K \hat{v}^k b^k - M \sum_{j=1}^F h_j a_j \quad[cite: 291]$$
  $$p(v_i^k = 1 \mid h) = \frac{\exp(b^k + \sum_j h_j W_j^k)}{\sum_{q=1}^K \exp(b^q + \sum_j h_j W_j^q)}, \quad p(h_j = 1 \mid V) = \sigma\left(M a_j + \sum_{k=1}^K \hat{v}^k W_j^k\right) \quad[cite: 291]$$

#### DBN Variational Lower Bound
A 2-layer DBN defines joint distribution $P(v, h^1, h^2; \theta) = P(v \mid h^1; W^1) P(h^1, h^2; W^2)$[cite: 285]. Given approximate posterior $Q(h^1 \mid v) = P(h^1 \mid v; W^1)$[cite: 286]:
$$\log P(v; \theta) \ge \sum_{h^1} Q(h^1 \mid v) \left[ \log P(h^1; W^2) + \log P(v \mid h^1; W^1) \right] + \mathcal{H}(Q(h^1 \mid v)) \quad[cite: 286]$$
When $W^2 = {W^1}^\top$, the bound is tight and equates to the single RBM likelihood; freezing $W^1$ and optimizing $W^2$ as an RBM maximizes $\sum_{h^1} Q(h^1 \mid v) \log P(h^1; W^2)$, guaranteeing non-decreasing lower bounds[cite: 285, 286, 287].

#### Partition Function Estimation via AIS
To calculate $\mathcal{Z}_B / \mathcal{Z}_A$ between an analytically solvable base-rate RBM $A$ ($W=0$) and target RBM $B$, AIS interpolates over $K$ intermediate distributions[cite: 317, 319, 320]:
$$P_k(v) \propto \exp\left((1 - \beta_k) v^\top b^A\right) \prod_{j=1}^{F_A} \left(1 + \exp\left((1 - \beta_k) a_j^A\right)\right) \exp\left(\beta_k v^\top b^B\right) \prod_{j=1}^{F_B} \left(1 + \exp\left(\beta_k\left(a_j^B + \sum_i W_{ij}^B v_i\right)\right)\right) \quad[cite: 320]$$
with schedule $0 = \beta_0 < \beta_1 < \dots < \beta_K = 1$[cite: 318, 320]. The unbiased partition function estimator is[cite: 318]:
$$\hat{\mathcal{Z}}_B = \frac{\mathcal{Z}_A}{M} \sum_{m=1}^M w_{\text{AIS}}^{(m)}, \quad w_{\text{AIS}}^{(m)} = \prod_{k=1}^K \frac{P_k^*(v_k^{(m)})}{P_{k-1}^*(v_k^{(m)})} \quad[cite: 318]$$
This yields a variational lower bound for test inputs $v^*$ under DBNs[cite: 322]:
$$\log P(v^*) \ge \frac{1}{M} \sum_{m=1}^M \log P^*(v^*, h^{1(m)}) + \mathcal{H}(Q(h^1 \mid v^*)) - \log \hat{\mathcal{Z}} \quad[cite: 322]$$

#### Deep Boltzmann Machine (DBM) Optimization
For a 2-hidden-layer DBM with energy $E(v, h^1, h^2; \theta) = -v^\top W^1 h^1 - {h^1}^\top W^2 h^2$, maximum likelihood updates follow[cite: 329, 333]:
$$\Delta W^1 = \alpha \left( \mathbb{E}_{P_{\text{data}}}[v {h^1}^\top] - \mathbb{E}_{P_{\text{model}}}[v {h^1}^\top] \right), \quad \Delta W^2 = \alpha \left( \mathbb{E}_{P_{\text{data}}}[h^1 {h^2}^\top] - \mathbb{E}_{P_{\text{model}}}[h^1 {h^2}^\top] \right) \quad[cite: 329, 333]$$
1. **Data-Dependent Statistics (Mean-Field Variational Inference):** Replaces $P(h^1, h^2 \mid v)$ with factorial distribution $Q(h^1, h^2; \mu, \nu)$ and iterates fixed-point equations until convergence[cite: 332, 333]:
   $$\mu_j = \sigma\left(\sum_{i} W_{ij}^1 v_i + \sum_{m} W_{jm}^2 \nu_m\right), \quad \nu_m = \sigma\left(\sum_{j} W_{jm}^2 \mu_j\right) \quad[cite: 332, 336]$$
2. **Model Expectations (Robbins-Monro Stochastic Approximation):** Maintains persistent Markov chains updated by a small number of Gibbs steps (e.g., 5 steps) between parameter updates[cite: 330, 331, 332, 339]:
   $$\theta^{t+1} = \theta^t + \alpha_t \left( \frac{1}{N} \sum_{n=1}^N \Phi(v^n, \mu^n, \nu^n) - \frac{1}{M} \sum_{m=1}^M \Phi(\tilde{v}^{t+1, m}, \tilde{h}^{1, t+1, m}, \tilde{h}^{2, t+1, m}) \right) \quad[cite: 331, 333]$$
   with decreasing schedule $\sum_t \alpha_t = \infty$ and $\sum_t \alpha_t^2 < \infty$[cite: 331].
3. **Greedy Pretraining Heuristic:** Avoids unit saturation by using tied weights with doubled bottom-up inputs ($2W^1$) for the bottom RBM and doubled hidden units for the top RBM, ensuring that composition exactly matches DBM conditional distributions[cite: 335, 336].

## Conclusions
- **Resolution of Optimization Pathologies in Deep Models:** The dissertation establishes that the longstanding difficulty in training deep multilayer neural networks is an artifact of poor initialization in non-convex parameter spaces[cite: 279, 302]. Greedy unsupervised pretraining places network parameters in basins of attraction from which gradient descent and backpropagation reliably discover superior functional optima[cite: 279, 280, 300, 302, 344].
- **Efficacy of Deep Non-linear Dimensionality Reduction:** Deep autoencoders consistently outperform classical linear baselines (PCA, SVD, LSA) by capturing non-linear manifolds in continuous image and discrete text spaces, reducing dimensionality while preserving semantic neighborhood topology[cite: 275, 281, 300, 301, 302, 304, 307, 344].
- **Discrepancy Between Heuristic CD and True Density Estimation:** Quantitative evaluation using Annealed Importance Sampling reveals that short-chain Contrastive Divergence ($CD_1$) is an inadequate surrogate for maximum likelihood, producing density models inferior by tens of nats relative to persistent chains or higher-order sampling ($CD_{25}$)[cite: 282, 324, 325]. Furthermore, adding hidden layers to a DBN only improves density when the underlying base model is poorly trained[cite: 326].
- **Bidirectional Representational Power of Deep Boltzmann Machines:** Deep Boltzmann Machines resolve the architectural limitations of feedforward DBN inference by incorporating top-down contextual priors alongside bottom-up sensory streams[cite: 275, 280, 282, 328, 345]. Combining mean-field variational inference with stochastic approximation enables stable learning across millions of undirected parameters, setting new standards in semi-supervised classification and visual recognition on permutation-invariant MNIST and NORB[cite: 281, 282, 328, 332, 339, 341, 342, 345].

## 1. Energy-Based Models and Restricted Boltzmann Machines (RBMs) (Chapter 2)

### Binary Restricted Boltzmann Machine (Section 2.1)
* **Energy Function:**
  For binary visible units $v \in \{0, 1\}^D$ and binary latent feature units $h \in \{0, 1\}^F$, the energy of state $\{v, h\}$ parameterized by $\theta = \{W, b, a\}$ is defined as[cite: 283]:
  $$E(v, h; \theta) = -v^\top W h - b^\top v - a^\top h = -\sum_{i=1}^D \sum_{j=1}^F W_{ij} v_i h_j - \sum_{i=1}^D b_i v_i - \sum_{j=1}^F a_j h_j \quad[cite: 283]$$
  where $W_{ij}$ represents the symmetric pairwise interaction weight, $b_i$ is the visible bias, and $a_j$ is the hidden bias[cite: 283].
* **Boltzmann Distribution and Partition Function:**
  The joint distribution over visible and hidden states is given by[cite: 283]:
  $$P(v, h; \theta) = \frac{1}{\mathcal{Z}(\theta)} \exp(-E(v, h; \theta)), \quad \mathcal{Z}(\theta) = \sum_{v \in \{0, 1\}^D} \sum_{h \in \{0, 1\}^F} \exp(-E(v, h; \theta)) \quad[cite: 283]$$
  The normalizing constant $\mathcal{Z}(\theta)$ is intractable to compute directly for large systems because it requires summing over $2^{D+F}$ discrete configurations[cite: 283, 285].
* **Marginal Likelihood via Factorization:**
  Due to the absence of within-layer connections (bipartite graph structure), hidden units are conditionally independent given $v$[cite: 284]. The hidden units can be marginalized out analytically[cite: 284]:
  $$P(v; \theta) = \frac{1}{\mathcal{Z}(\theta)} \exp(b^\top v) \prod_{j=1}^F \left(1 + \exp\left(a_j + \sum_{i=1}^D W_{ij} v_i\right)\right) \quad[cite: 284]$$
* **Factorial Conditional Distributions:**
  Conditioned on one layer, the activations of units in the other layer are mutually independent Bernoulli variables evaluated via the logistic sigmoid $\sigma(z) = \frac{1}{1 + \exp(-z)}$[cite: 284]:
  $$p(h_j = 1 \mid v) = \sigma\left(\sum_{i=1}^D W_{ij} v_i + a_j\right), \quad p(v_i = 1 \mid h) = \sigma\left(\sum_{j=1}^F W_{ij} h_j + b_i\right) \quad[cite: 284]$$
* **Log-Likelihood Gradient and Contrastive Divergence:**
  Differentiating the marginal log-likelihood with respect to parameters yields the difference between data-dependent and model expectations[cite: 284, 285]:
  $$\frac{\partial \log P(v; \theta)}{\partial W} = \mathbb{E}_{P_{\text{data}}}[v h^\top] - \mathbb{E}_{P_{\text{model}}}[v h^\top] \quad[cite: 284]$$
  Because computing the exact model expectation $\mathbb{E}_{P_{\text{model}}}[\cdot]$ requires sampling from the equilibrium distribution of the model, training relies on $T$-step Contrastive Divergence ($CD_T$)[cite: 285]:
  $$\Delta W = \alpha \left( \mathbb{E}_{P_{\text{data}}}[v h^\top] - \mathbb{E}_{P_T}[v h^\top] \right) \quad[cite: 285]$$
  where $P_T$ is the distribution obtained by initializing a Gibbs chain at the empirical data distribution and running alternating block Gibbs steps for $T$ full cycles[cite: 285].

### Exponential Family Generalizations (Section 2.3)
* **Gaussian-Binary RBM (Continuous Real-Valued Data):**
  Models continuous sensory observations $v \in \mathbb{R}^D$ paired with binary latents $h \in \{0, 1\}^F$[cite: 289]:
  $$E(v, h; \theta) = \sum_{i=1}^D \frac{(v_i - b_i)^2}{2\sigma_i^2} - \sum_{i=1}^D \sum_{j=1}^F W_{ij} h_j \frac{v_i}{\sigma_i} - \sum_{j=1}^F a_j h_j \quad[cite: 289]$$
  Conditioned on $h$, each visible unit follows an independent univariate Gaussian distribution whose conditional mean is shifted by top-down hidden signals[cite: 290]:
  $$p(v_i = x \mid h) = \frac{1}{\sqrt{2\pi}\sigma_i} \exp\left( -\frac{(x - b_i - \sigma_i \sum_j h_j W_{ij})^2}{2\sigma_i^2} \right), \quad p(h_j = 1 \mid v) = \sigma\left(a_j + \sum_{i=1}^D W_{ij} \frac{v_i}{\sigma_i}\right) \quad[cite: 290]$$
* **Replicated Softmax Model (Word-Count and Discrete Data):**
  Models bag-of-words document vectors of variable length $M$ over vocabulary size $K$[cite: 291]. By scaling the hidden bias terms proportionally by document length $M$, the energy function accounts for varying document sizes[cite: 291, 292]:
  $$E(V, h) = -\sum_{j=1}^F \sum_{k=1}^K W_j^k h_j \hat{v}^k - \sum_{k=1}^K \hat{v}^k b^k - M \sum_{j=1}^F a_j h_j \quad[cite: 291]$$
  where $\hat{v}^k = \sum_{i=1}^M v_i^k$ is the total count of word $k$ in the document[cite: 292]. Visible units are modeled via the softmax distribution, and hidden units act as Bernoulli latent topic detectors[cite: 291]:
  $$p(v_i^k = 1 \mid h) = \frac{\exp\left(b^k + \sum_j h_j W_j^k\right)}{\sum_{q=1}^K \exp\left(b^q + \sum_j h_j W_j^q\right)}, \quad p(h_j = 1 \mid V) = \sigma\left(M a_j + \sum_{k=1}^K \hat{v}^k W_j^k\right) \quad[cite: 291]$$

### Hybrid Generative Formulation (Section 2.2)
* **Joint Factorization:**
  A two-hidden-layer DBN specifies a probability distribution where the top two layers form an undirected bipartite RBM and the bottom layer forms a directed sigmoid belief network[cite: 285]:
  $$P(v, h^1, h^2; \theta) = P(v \mid h^1; W^1) P(h^1, h^2; W^2) \quad[cite: 285]$$
  $$P(v \mid h^1; W^1) = \prod_{i=1}^D p(v_i \mid h^1; W^1), \quad p(v_i = 1 \mid h^1) = \sigma\left(\sum_j W_{ij}^1 h_j^1\right) \quad[cite: 285]$$
  $$P(h^1, h^2; W^2) = \frac{1}{\mathcal{Z}(W^2)} \exp\left( {h^1}^\top W^2 h^2 \right) \quad[cite: 285]$$

### Variational Log-Likelihood Lower Bound (Section 2.2)
* **Derivation:**
  For any approximating conditional distribution $Q(h^1 \mid v)$, Jensen's inequality provides a variational lower bound on data log-likelihood[cite: 286]:
  $$\log P(v; \theta) \ge \sum_{h^1} Q(h^1 \mid v) \left[ \log P(h^1; W^2) + \log P(v \mid h^1; W^1) \right] + \mathcal{H}(Q(h^1 \mid v)) \quad[cite: 286]$$
  where $\mathcal{H}(Q) = -\sum_{h^1} Q(h^1 \mid v) \log Q(h^1 \mid v)$ is the Shannon entropy functional[cite: 286].
* **Proof of Greedy Layer-Wise Improvement:**
  1. Setting $Q(h^1 \mid v) = P(h^1 \mid v; W^1)$ from the base RBM, and tying the weights $W^2 = {W^1}^\top$, makes the variational lower bound tight[cite: 286]:
     $$\log P(v; W^1, {W^1}^\top) = \log P_{\text{RBM}}(v; W^1) \quad[cite: 286]$$
  2. Freezing $W^1$ and updating $W^2$ independently to fit the aggregated posterior $\frac{1}{N} \sum_{n=1}^N P(h^1 \mid v_n; W^1)$ maximizes $\sum_{h^1} Q(h^1 \mid v) \log P(h^1; W^2)$[cite: 286, 287].
  3. Consequently, any parameter update on $W^2$ that increases the second-layer RBM likelihood guarantees that the overall DBN variational lower bound increases or remains above the baseline single-layer model[cite: 287].

## 3. Statistical Metric Learning and Information Retrieval (Chapter 3)

### Gaussian Process Covariance Adaptation (GP-DBN) (Section 3.1)
* **Kernel Parameterization:**
  Given a deep non-linear mapping $F(x; W)$ parameterized by an unrolled DBN, the stationary squared-exponential covariance kernel is computed over latent features rather than raw pixel inputs[cite: 294, 295]:
  $$K(x_i, x_j) = \alpha \exp\left( -\frac{1}{2\beta} \| F(x_i; W) - F(x_j; W) \|_2^2 \right) \quad[cite: 295]$$
* **Marginal Likelihood Optimization:**
  For Gaussian process regression $y \sim \mathcal{N}(0, K + \sigma^2 I)$, parameters $\theta = \{\alpha, \beta, \sigma^2, W\}$ are fine-tuned by maximizing the marginal log-likelihood[cite: 294, 295]:
  $$\mathcal{L}(\theta) = -\frac{N}{2} \log 2\pi - \frac{1}{2} \log |K_y| - \frac{1}{2} y^\top K_y^{-1} y, \quad K_y = K + \sigma^2 I \quad[cite: 294, 296]$$
  Gradients with respect to internal network weights $W$ follow from matrix calculus and backpropagation[cite: 296]:
  $$\frac{\partial \mathcal{L}}{\partial W} = \frac{1}{2} \sum_{i, j} \left( K_y^{-1} y y^\top K_y^{-1} - K_y^{-1} \right)_{ij} \frac{\partial K_{ij}}{\partial W} \quad[cite: 296]$$

### Semantic Hashing (Section 3.3)
* **Deterministic Input Noise for Code Binarization:**
  To force deep autoencoder bottleneck activations $z$ toward discrete binary states $\{0, 1\}$ without gradient collapse, fixed zero-mean Gaussian noise $\epsilon_j \sim \mathcal{N}(0, \sigma_{\text{noise}}^2)$ is added to the pre-activation sum during conjugate-gradient fine-tuning[cite: 306]:
  $$z_j = \sigma\left( \sum_i W_{ij} h_i + a_j + \epsilon_j \right) \quad[cite: 306]$$
  Because the network must accurately reconstruct documents under noisy channel conditions, it pushes pre-activations toward extreme positive ($+\infty \implies z_j \to 1$) or extreme negative ($-\infty \implies z_j \to 0$) regimes[cite: 306]. Thresholding at $z_j > 0.1$ maps each document to a memory address where nearest neighbors within a small Hamming ball radius are retrievable in $\mathcal{O}(1)$ time[cite: 305, 306, 309].

### Non-linear Neighborhood Component Analysis (Nonlinear NCA) (Section 3.4)
* **Differentiable Stochastic Neighbor Selection:**
  For a labeled training set $\{(x_i, c_i)\}_{i=1}^N$, the probability that point $i$ selects point $j$ as its nearest neighbor in latent space $F(\cdot; W)$ is defined via a soft-max over pairwise Euclidean distances[cite: 311]:
  $$p_{ij} = \frac{\exp(-\|F(x_i; W) - F(x_j; W)\|^2)}{\sum_{z \ne i} \exp(-\|F(x_i; W) - F(x_z; W)\|^2)}, \quad p_{ii} = 0 \quad[cite: 311]$$
* **Classification Objective and Gradients:**
  The objective maximizes the expected number of correctly classified points under stochastic neighbor selection[cite: 311]:
  $$O_{\text{NCA}} = \sum_{i=1}^N p_i = \sum_{i=1}^N \sum_{j: c_j = c_i} p_{ij} \quad[cite: 311]$$
  Differentiating with respect to the extracted features $F_i = F(x_i; W)$ yields[cite: 312]:
  $$\frac{\partial O_{\text{NCA}}}{\partial F_i} = -2 \sum_{j: c_j = c_i} p_{ij} \left( (F_i - F_j) - \sum_k p_{ik} (F_i - F_k) \right) - 2 \sum_{j: c_j = c_i} p_{ji} \left( (F_i - F_j) - \sum_k p_{jk} (F_i - F_k) \right) \quad[cite: 312]$$
* **Regularized Semi-Supervised Extension:**
  When labeled data is scarce, the objective combines the supervised NCA metric term with an unsupervised autoencoder cross-entropy reconstruction loss $E$[cite: 313, 314]:
  $$C = \lambda O_{\text{NCA}} + (1 - \lambda) (-E) \quad[cite: 313]$$

### General Ratio Estimation Framework (Section 4.2)
* **Importance Weighting Identity:**
  The ratio of normalizing constants between a tractable base distribution $P_0(x) = P_0^*(x)/\mathcal{Z}_0$ and an intractable target distribution $P_K(x) = P_K^*(x)/\mathcal{Z}_K$ is derived via sequential importance sampling over an extended state space[cite: 317, 319]:
  $$\frac{\mathcal{Z}_K}{\mathcal{Z}_0} = \mathbb{E}_{\mathcal{Q}}\left[ \frac{\mathcal{P}^*(x_1, \dots, x_K)}{\mathcal{Q}^*(x_1, \dots, x_K)} \right] \quad[cite: 319]$$
  Using an annealing schedule $0 = \beta_0 < \beta_1 < \dots < \beta_K = 1$, the intermediate unnormalized distributions are defined as[cite: 318]:
  $$P_k^*(x) = \left(P_0^*(x)\right)^{1 - \beta_k} \left(P_K^*(x)\right)^{\beta_k} \quad[cite: 318]$$
  For each annealing trajectory, the cumulative importance weight is[cite: 318]:
  $$w_{\text{AIS}} = \frac{P_1^*(x_1)}{P_0^*(x_1)} \frac{P_2^*(x_2)}{P_1^*(x_2)} \cdots \frac{P_K^*(x_K)}{P_{K-1}^*(x_K)} = \prod_{k=1}^K \frac{P_k^*(x_k)}{P_{k-1}^*(x_k)} \quad[cite: 318]$$
  where $x_k \sim T_{k-1}(x_k \leftarrow x_{k-1})$ is generated using a Markov transition operator $T_{k-1}$ leaving $P_{k-1}$ invariant[cite: 317, 318]. The Monte Carlo average over $M$ independent runs yields an unbiased estimator of the partition function[cite: 318]:
  $$\hat{\mathcal{Z}}_K = \frac{\mathcal{Z}_0}{M} \sum_{m=1}^M w_{\text{AIS}}^{(m)}, \quad \text{Var}(\hat{\mathcal{Z}}_K) \approx \frac{\hat{s}^2}{M} \quad[cite: 318]$$

### Application to Bipartite RBMs (Section 4.2.2 & 4.2.3)
* **Exact Energy Interpolation:**
  Let Model $A$ be a base-rate RBM with zero weights ($W=0$, whose partition function $\mathcal{Z}_A = \prod_j (1 + e^{a_j}) \prod_i (1 + e^{b_i})$ is computed analytically)[cite: 321]. The interpolated energy function on state $\{v, h^A, h^B\}$ is[cite: 320]:
  $$E_k(v, h^A, h^B) = (1 - \beta_k) E(v, h^A; \theta_A) + \beta_k E(v, h^B; \theta_B) \quad[cite: 320]$$
  Summing out hidden layers analytically yields the exact unnormalized intermediate probability[cite: 320]:
  $$P_k^*(v) = e^{(1-\beta_k) \sum_i b_i^A v_i} \left[ \prod_{j=1}^{F_A} \left(1 + e^{(1-\beta_k) a_j^A}\right) \right] \times e^{\beta_k \sum_i b_i^B v_i} \left[ \prod_{j=1}^{F_B} \left(1 + e^{\beta_k \left( \sum_i W_{ij}^B v_i + a_j^B \right)}\right) \right] \quad[cite: 320]$$
  Alternating Gibbs transition operators sample $h^A, h^B \mid v$ and visible updates $v \mid h^A, h^B$ via logistic conditionals[cite: 320]:
  $$p(v_i = 1 \mid h^A, h^B) = \sigma\left( (1 - \beta_k) \left(\sum_j W_{ij}^A h_j^A + b_i^A\right) + \beta_k \left(\sum_j W_{ij}^B h_j^B + b_i^B\right) \right) \quad[cite: 320]$$

### Model Energy and Intractable Expectations (Section 5.2)
* **General Undirected Energy Formulation:**
  For a two-hidden-layer network $\{v, h^1, h^2\}$, the energy is defined as[cite: 333]:
  $$E(v, h^1, h^2; \theta) = -v^\top W^1 h^1 - {h^1}^\top W^2 h^2 \quad[cite: 333]$$
* **Parameter Gradients:**
  Exact log-likelihood gradients require evaluating expectations over both data-clamped and free model distributions[cite: 329]:
  $$\frac{\partial \log P(v; \theta)}{\partial W^1} = \mathbb{E}_{P_{\text{data}}}[v {h^1}^\top] - \mathbb{E}_{P_{\text{model}}}[v {h^1}^\top] \quad[cite: 329, 333]$$
  $$\frac{\partial \log P(v; \theta)}{\partial W^2} = \mathbb{E}_{P_{\text{data}}}[h^1 {h^2}^\top] - \mathbb{E}_{P_{\text{model}}}[h^1 {h^2}^\top] \quad[cite: 333]$$
  In DBMs, both expectations are intractable because within-layer conditioning does not fully factorize over the entire network when inputs or upper layers are fixed[cite: 330].

---

### Variational Mean-Field Inference for Data Expectations (Section 5.2.2)
* **Mean-Field Variational Lower Bound:**
  The true posterior $P(h^1, h^2 \mid v; \theta)$ is approximated by a fully factorized mean-field distribution $Q(h^1, h^2; \mu, \nu) = \prod_j q(h_j^1) \prod_m q(h_m^2)$, where $q(h_j^1 = 1) = \mu_j$ and $q(h_m^2 = 1) = \nu_m$[cite: 332]:
  $$\log P(v; \theta) \ge \sum_{i, j} W_{ij}^1 v_i \mu_j + \sum_{j, m} W_{jm}^2 \mu_j \nu_m - \log \mathcal{Z}(\theta) + \mathcal{H}(Q) \quad[cite: 332]$$
  where $\mathcal{H}(Q) = -\sum_j [\mu_j \log \mu_j + (1 - \mu_j) \log(1 - \mu_j)] - \sum_m [\nu_m \log \nu_m + (1 - \nu_m) \log(1 - \nu_m)]$[cite: 332].
* **Fixed-Point Iterative Equations:**
  Setting partial derivatives of the variational bound with respect to variational parameters $\mu_j$ and $\nu_m$ to zero yields coordinate ascent updates that integrate bottom-up inputs and top-down feedback[cite: 332]:
  $$\mu_j \leftarrow \sigma\left( \sum_{i=1}^D W_{ij}^1 v_i + \sum_{m=1}^F W_{jm}^2 \nu_m \right) \quad[cite: 332]$$
  $$\nu_m \leftarrow \sigma\left( \sum_{j=1}^{F_1} W_{jm}^2 \mu_j \right) \quad[cite: 332]$$

---

### Robbins-Monro Stochastic Approximation for Model Expectations (Section 5.2.1)
* **Asymptotic Convergence of Persistent MCMC Chains:**
  To evaluate the negative phase $\mathbb{E}_{P_{\text{model}}}[\Phi]$, the algorithm maintains $M$ persistent particles $(\tilde{v}, \tilde{h}^1, \tilde{h}^2)$ updated by a small number of Gibbs steps (e.g., $k=5$) between parameter updates[cite: 331, 333, 339]:
  $$\theta^{t+1} = \theta^t + \alpha_t \left[ \frac{1}{N} \sum_{n=1}^N \mu^n {\nu^n}^\top - \frac{1}{M} \sum_{m=1}^M \tilde{h}^{1, t+1, m} (\tilde{h}^{2, t+1, m})^\top \right] \quad[cite: 331, 333]$$
* **Robbins-Monro Conditions:**
  As long as the learning rate sequence satisfies $\sum_{t=0}^\infty \alpha_t = \infty$ and $\sum_{t=0}^\infty \alpha_t^2 < \infty$ (e.g., $\alpha_t = \frac{a}{b + t}$), the noise induced by persistent sampling vanishes asymptotically, guaranteeing convergence to a stationary point of the log-likelihood[cite: 331].

---

### Modified Greedy Pretraining with Weight Doubling (Section 5.3.1)
* **The Weight Doubling Problem:**
  In a joint DBM, intermediate layer $h^1$ receives input from both $v$ (via $W^1$) and $h^2$ (via $W^2$)[cite: 336]. Standard layer-by-layer pretraining would fail because stacking independently trained RBMs doubles the incoming input variance into $h^1$, saturating the sigmoid activations[cite: 335].
* **Mathematical Solution:**
  1. *Bottom RBM:* Train with visible-to-hidden weights doubled: $p(h_j^1 = 1 \mid v) = \sigma(2 \sum_i W_{ij}^1 v_i)$[cite: 335, 336].
  2. *Top RBM:* Train with double the number of hidden units or doubled incoming weights[cite: 336].
  3. *Composition:* When composing the models into an undirected DBM, the inputs are halved:
     $$p(h_j^1 = 1 \mid v, h^2) = \sigma\left( \sum_i W_{ij}^1 v_i + \sum_m W_{jm}^2 h_m^2 \right) \quad[cite: 336]$$
     This matches the conditional distributions of the target DBM without causing saturation[cite: 336].
