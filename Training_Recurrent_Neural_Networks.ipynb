# Academic Summary: Training Recurrent Neural Networks

## Abstract
This doctoral dissertation addresses the fundamental computational and optimization barriers that historically prevented the effective training of Recurrent Neural Networks (RNNs) on sequential tasks characterized by long-term temporal dependencies[cite: 528, 535]. The work introduces novel probabilistic sequence architectures, specialized second-order and first-order optimization algorithms, and rigorous empirical benchmarks across challenging sequence modeling domains[cite: 528]. Specifically, the dissertation establishes: (1) the Recurrent Temporal Restricted Boltzmann Machine (RTRBM), which resolves the intractable posterior inference of earlier temporal energy-based models[cite: 528, 558, 565, 567]; (2) a truncated-Newton Hessian-Free (HF) optimization framework augmented with "structural damping," demonstrating for the first time that standard RNNs can solve pathological temporal dependency problems spanning hundreds of time steps without specialized memory gating units[cite: 528, 572, 577, 581]; (3) the Multiplicative RNN (MRNN), leveraging tensor factorizations for character-level language modeling[cite: 528, 587, 589, 590]; (4) Augmented Hessian-Free (AHF) algorithms for non-linear optimal feedback motor control under delays and disturbances[cite: 528, 596, 597, 598]; and (5) a formal demonstration that stochastic gradient descent (SGD) paired with classical momentum or Nesterov's accelerated gradient (NAG) can train RNNs on long-term dependency benchmarks, provided the recurrent transition matrix is initialized near the boundary of chaos (spectral radius $\rho \approx 1.2$)[cite: 528, 607, 614, 615, 616].

---

## Problems
- **The Intractable Inference Problem in Sequence Energy Models:** The Temporal Restricted Boltzmann Machine (TRBM) possesses high expressive power via componential hidden states[cite: 558, 559], but computing the exact posterior distribution over latent variables conditioned on observations is intractable[cite: 558, 561, 566]. Filtering approximations introduce uncontrolled errors and high gradient variance during Contrastive Divergence (CD) training[cite: 535, 561, 562].
- **Vanishing and Exploding Gradients in Standard RNNs:** The iterative application of the hidden-to-hidden transition matrix $W_{hh}$ across $T$ steps causes gradient signals backpropagated through time to either decay exponentially to zero (if the spectrum of $W_{hh}$ is contractive) or grow exponentially large (if expansive)[cite: 544, 545]. This dynamic instability traditionally restricted gradient-based learning to short-term temporal dependencies[cite: 535, 545].
- **Curvature Pathology in Deep Sequence Objectives:** Standard first-order gradient descent fails in the presence of long-range dependencies because the loss surface contains extreme directional anisotropy (high condition numbers)[cite: 539, 540, 576, 577]. Large steps along low-curvature directions cause divergence or destabilization in orthogonal high-curvature directions[cite: 540, 576, 577].
- **Context-Character Interaction Bottlenecks in Language Modeling:** In standard additive RNNs, the current input vector $v_t$ contributes solely as an additive shift to the recurrent pre-activation[cite: 588]. This limits the model's capacity to modify transition dynamics based on the immediate lexical context, restricting character-level modeling of syntactic and long-range constraints (e.g., matching parentheses)[cite: 587, 588].
- **Sensorimotor Control under Disturbances and Delays:** Classical trajectory optimization algorithms (e.g., iterative Linear Quadratic Regulator, iLQR) compute open-loop or locally replanned command sequences[cite: 596, 597]. They cannot anticipate unobserved external physical disturbances or delayed feedback, leading to ringing, overshooting, and system instability[cite: 596, 599, 600].

---

## Proposed Solutions
- **The Recurrent Temporal Restricted Boltzmann Machine (RTRBM):** Eliminates intractable latent variable sampling by introducing a deterministic recurrent hidden state $r_t$ that parameterizes the conditional bias of an RBM at each time step[cite: 565, 567, 568]. This makes posterior inference deterministic and enables parameter updates via Backpropagation Through Time (BPTT) coupled with single-step Contrastive Divergence ($CD_k$)[cite: 567, 568, 569].
- **Hessian-Free Optimization with Structural Damping:** Employs an iterative truncated-Newton solver that leverages the linear conjugate gradient (CG) algorithm to invert the generalized Gauss-Newton curvature matrix $G_f$ without explicitly computing or storing the full Hessian[cite: 572, 573, 574]. Introduces "structural damping," which penalizes changes in the hidden-state sequence $h(\theta)$ via a matched divergence proxy, preventing destructive updates to recurrent dynamics[cite: 572, 576, 577, 578].
- **Multiplicative Recurrent Neural Network (MRNN):** Replaces the additive input integration with a factored 3-way tensor formulation[cite: 589, 590]. The input symbol gates and dynamically synthesizes the hidden-to-hidden transition matrix through rank-one factor matrices, allowing context-sensitive nonlinear transitions[cite: 589, 590].
- **Augmented Hessian-Free (AHF) Control Policies:** Trains an RNN-based closed-loop controller on simulated biomechanical plants using AHF, which greedily optimizes step sizes over an expanding conjugate basis of the Krylov subspace[cite: 597, 598]. This produces feedback control laws that counteract external force perturbations despite sensory feedback delays[cite: 596, 597, 604, 605].
- **Echo-State-Inspired Sparse Spectral Initialization:** Disproves the necessity of second-order methods for basic long-range dependency benchmarks by pairing SGD and momentum with a sparse random initialization where non-zero weights of $W_{hh}$ are scaled to have a spectral radius slightly above unity ($\rho(W_{hh}) \approx 1.2$)[cite: 528, 607, 614, 615].

---

## Purpose
The primary objective of this dissertation is to dismantle the theoretical and empirical consensus asserting that general recurrent neural networks cannot be trained on long-range dependencies with gradient-based methods[cite: 528, 535, 536]. The work establishes scalable optimization techniques and expressive architectures capable of exploiting extended temporal contexts in generative sequence modeling, complex continuous language modeling, and robust motor control[cite: 528, 535, 536, 618].

## Methodology

### 1. Mathematical Formulations

#### Recurrent Temporal RBM Formulation
At time step $t$, the distribution over visible vector $v_t$ and binary hidden variables $h'_t$ is conditioned on the previous deterministic representation $h_{t-1}$[cite: 566]:
$$\hat{P}(v_t, h'_t \mid h_{t-1}) = \frac{1}{Z(h_{t-1})} \exp\left( {h'_t}^T W v_t + v_t^T b_v + {h'_t}^T (b_h + W' h_{t-1}) \right) \quad[cite: 566]$$
The deterministic hidden state sequence updates via mean-field propagation[cite: 566, 567]:
$$h_t = \text{sigmoid}(W v_t + b_h + W' h_{t-1}) \quad[cite: 566, 567, 568]$$
Because $h_t$ is uniquely determined given $v_1^t$, the total sequence log-likelihood decouples[cite: 568]:
$$\log \hat{P}(v_1^T) = \sum_{t=1}^T \log \hat{P}(v_t \mid h(v)_{t-1}) \quad[cite: 568, 569]$$

#### Generalized Gauss-Newton Matrix with Structural Damping
Let $f(\theta)$ be the scalar sequence loss and $o$ be the pre-activation outputs[cite: 572, 574]. The generalized Gauss-Newton matrix $G_f$ is defined as[cite: 574]:
$$G_f \equiv J_{o,\theta}^T (L \circ g)'' J_{o,\theta} \quad[cite: 574]$$
where $J_{o,\theta} = \frac{\partial o}{\partial \theta}$ and $(L \circ g)''$ is the Hessian of the matching loss with respect to $o$[cite: 574, 575].
The structurally damped local quadratic sub-objective is[cite: 573, 577]:
$$q_{\theta_n}(\delta) = f(\theta_n) + \nabla f(\theta_n)^T \delta + \frac{1}{2} \delta^T \left( G_f + \lambda I + \lambda \mu G_S \right) \delta \quad[cite: 573, 577, 578]$$
where $G_S = J_{u,\theta}^T (D \circ e)'' J_{u,\theta}$ captures curvature with respect to recurrent hidden pre-activations $u$, and $\mu > 0$ balances state stability[cite: 574, 577, 578].

#### Multiplicative Recurrent Dynamics (MRNN)
To avoid the cubic parameter scaling of an unconstrained tensor $W_{hh} \in \mathbb{R}^{N_h \times N_h \times N_v}$[cite: 589], the transition tensor is factorized via three matrices $W_{fv}$, $W_{hf}$, and $W_{fh}$[cite: 589]:
$$f_t = \text{diag}(W_{fv} v_t) \cdot W_{fh} h_{t-1} \quad[cite: 590]$$
$$h_t = \tanh(W_{hf} f_t + W_{hv} v_t) \quad[cite: 590]$$
$$o_t = W_{oh} h_t + b_o, \quad P(v_{t+1} \mid v_{\le t}) = \text{softmax}(o_t) \quad[cite: 590, 591]$$

#### Nesterov's Accelerated Gradient (NAG)
The velocity vector $v_t$ and parameter vector $\theta_t$ update according to[cite: 609]:
$$v_t = \mu_{t-1} v_{t-1} - \epsilon_{t-1} \nabla f(\theta_{t-1} + \mu_{t-1} v_{t-1}) \quad[cite: 609]$$
$$\theta_t = \theta_{t-1} + v_t \quad[cite: 609]$$
where $\mu_t \in [0, 1)$ is the momentum decay factor and $\epsilon$ is the learning rate[cite: 539, 608]. Evaluating the gradient after applying the velocity step ($\theta_{t-1} + \mu_{t-1} v_{t-1}$) provides predictive directional damping[cite: 609, 610].

### 2. Experimental Setup
- **Pathological Synthetic Benchmarks:** Sequence lengths $T \in \{30, 50, 100, 200\}$[cite: 580]. Models evaluate:
  1. *Addition Problem:* Summing two marked continuous inputs separated by hundreds of uniform noise steps[cite: 580, 583].
  2. *Multiplication Problem:* Multiplying two marked scalar values[cite: 580, 583].
  3. *XOR Problem:* Binary parity over temporal gaps[cite: 580, 583].
  4. *Temporal Order & 3-Bit Temporal Order Tasks:* Discrete symbol sequence ordering[cite: 580, 583].
  5. *Random Permutation Task:* Requisite extraction of initial symbol despite terminal target dominance[cite: 580, 584].
  6. *5-Bit & 20-Bit Noiseless Memorization:* Storing and sequentially emitting bit strings across delay intervals[cite: 580, 581, 584].
- **Language Modeling Datasets:** Character-level text processing across 100MB corpora (86-character vocabulary)[cite: 591]:
  - English Wikipedia (cleaned markup)[cite: 591].
  - New York Times Annotated Corpus[cite: 591].
  - Machine Learning Research Papers Corpus (NIPS & JMLR)[cite: 591].
- **Motor Control Simulation:** A 2-joint planar revolute arm matching Macaca mulatta physiological mass/length distributions[cite: 599, 601]. Evaluated across:
  - *Center-Out Reaching:* 64 angular directions from a central origin[cite: 599, 602].
  - *Postural Resistance:* Maintaining position under 8 unobserved directional static force fields with 100 ms sensory feedback delay[cite: 599, 604].
  - *Delay-Disturbance-Noise (DDN):* 70 ms latency, continuous Gaussian joint torque disturbances, and signal-dependent multiplicative actuator noise[cite: 600, 604, 605].

## Conclusions
- **Resolution of the Fundamental Training Obstacle:** The long-standing belief that general recurrent neural networks are incapable of learning long-range temporal dependencies with gradient-based updates is refuted[cite: 528, 535, 618]. The difficulty was rooted in poor spectral initializations and optimization algorithms that were ill-equipped for non-convex, high-curvature loss surfaces[cite: 528, 607, 618].
- **Efficacy of Second-Order Curvature Alignment:** Truncated-Newton Hessian-Free optimization with structural damping eliminates catastrophic divergence in recurrent loops by constraining parameter proposals to respect state-space stability[cite: 572, 576, 577, 618]. This enables standard tanh-based RNNs to match or exceed specialized architectures like LSTMs across high-dimensional speech, music, and physical video data[cite: 581, 582].
- **Representational Superiority of Gated Tensor Transitions:** Factoring recurrent transitions conditioned on input tokens (MRNN) allows distributed hidden states to model complex contextual syntax[cite: 588, 589, 590], successfully tracking structured long-range dependencies (e.g., balanced quotes and nested parentheses) across dozens of characters[cite: 528, 535, 588, 618].
- **First-Order Viability via Spectral Initialization:** When the hidden-to-hidden transition matrix is initialized sparsely with a spectral radius calibrated slightly above unity ($\rho \approx 1.2$), standard first-order SGD equipped with Nesterov momentum can bridge temporal gaps of 50 to 200 time steps without requiring second-order solvers[cite: 607, 614, 615, 618].

## 1. Statistical Learning Theory and Generalization Bounds (Chapter 2)

### Bounded Overfitting and Uniform Convergence (Theorem 2.6.1)
* **Mathematical Formulation:**
  Let $\mathcal{F}$ be a finite class of hypotheses and let the loss function be bounded such that $L(z; y) \in [0, 1]$[cite: 546]. For a training sample set $S \sim D^{|S|}$ drawn i.i.d. from distribution $D$, the probability that the test error exceeds the empirical training error by more than a margin $t$ is bounded by[cite: 546]:
  $$\text{Pr}_{S \sim D^{|S|}}\left[ \exists f \in \mathcal{F} : \text{Test}_D(f) - \text{Train}_S(f) > \sqrt{\frac{\log |\mathcal{F}| + \log(1/\delta)}{|S|}} \right] \le \delta$$[cite: 546, 547]
* **Derivation via Hoeffding's Inequality and Union Bound:**
  Applying Hoeffding's inequality for a single fixed function $f$ gives[cite: 546]:
  $$\text{Pr}_{S \sim D^{|S|}}[\text{Test}_D(f) - \text{Train}_S(f) > t] \le \exp\left(-|S| t^2\right)$$[cite: 546]
  Taking the union bound across all $|\mathcal{F}|$ members[cite: 546]:
  $$\text{Pr}\left[\bigcup_{f \in \mathcal{F}} \{\text{Test}_D(f) - \text{Train}_S(f) > t\}\right] \le \sum_{f \in \mathcal{F}} \text{Pr}[\text{Test}_D(f) - \text{Train}_S(f) > t] \le |\mathcal{F}| \exp\left(-|S| t^2\right)$$[cite: 546]
  Setting $|\mathcal{F}| \exp\left(-|S| t^2\right) = \delta$ and solving for $t$ yields the uniform convergence threshold[cite: 546, 547]:
  $$t = \sqrt{\frac{\log(1/\delta) + \log |\mathcal{F}|}{|S|}}$$[cite: 547]
* **Role in the Paper:**
  This bound provides a theoretical justification for parameter counting in neural networks[cite: 547]. Sutskever points out that any 32-bit or 64-bit floating-point neural network parameterization with $|\theta|$ parameters belongs to a finite set $|\mathcal{F}| \le 2^{64|\theta|}$[cite: 547]. Hence, the number of required training samples $|S|$ scales proportionally to the number of parameters $|\theta|$[cite: 547].

---

### Equivalence Between Norm Regularization and Constraint Sets (Theorem 2.6.2)
* **Mathematical Formulation:**
  Let $\text{Train}_S(\theta)$ and regularizer $R(\theta)$ be smooth functions[cite: 547]. For every local minimum $\theta^*$ of the constrained problem:
  $$\min_\theta \text{Train}_S(\theta) \quad \text{subject to} \quad R(\theta) \le \lambda$$[cite: 547]
  there exists a Lagrange multiplier $\lambda' \ge 0$ such that $\theta^*$ is a local minimum of the unconstrained regularized objective[cite: 547]:
  $$\text{Train}_S(\theta) + \lambda' R(\theta)$$[cite: 547]
* **Role in the Paper:**
  Proves that adding weight penalties (such as $L_2$ weight decay $R(\theta) = \frac{1}{2}\|\theta\|^2$) is mathematically equivalent to restricting optimization to a bounded ball in hypothesis space, preserving generalization bounds[cite: 547, 548].

## 2. Convergence Rates of First-Order Optimization (Section 2.2)

### Standard Gradient Descent Convergence
* **Mathematical Formulation:**
  For a positive-definite quadratic objective with condition number $R = \lambda_{\max}/\lambda_{\min}$, gradient descent with optimal step size $\epsilon = 1/\lambda_{\max}$ converges at a linear rate[cite: 539]:
  $$F(\theta_t) - F(\theta^*) = \mathcal{O}\left( \left(1 - \frac{1}{R}\right)^t \right)$$[cite: 539]
  For general convex functions with $L$-Lipschitz gradients, the sublinear rate is[cite: 539]:
  $$F(\theta_t) - F(\theta^*) \le \frac{\|\theta_1 - \theta^*\|^2}{2 \epsilon t}$$[cite: 539]

### Strongly Convex Convergence Rate (Theorem 2.2.1)
* **Theorem & Proof:**
  If $F$ is $L$-Lipschitz and $\sigma$-strongly convex (i.e., $\frac{\sigma}{2}\|\theta - \theta^*\|^2 \le F(\theta) - F(\theta^*)$), then[cite: 539]:
  $$F(\theta_t) - F(\theta^*) \le \left(1 - \frac{\sigma}{6L}\right)^t (F(\theta_1) - F(\theta^*))$$[cite: 539]
  Substituting strong convexity into the sublinear bound gives $F(\theta_t) - F(\theta^*) \le \frac{2L}{t\sigma} (F(\theta_1) - F(\theta^*))$[cite: 539]. Halving the error every $t = 4L/\sigma$ iterations yields exponential contraction[cite: 539].

### Nesterov’s Accelerated Gradient (NAG) Convergence
* **Mathematical Formulation:**
  For convex functions with $L$-Lipschitz continuous gradients, NAG achieves the accelerated rate[cite: 540, 609]:
  $$F(\theta_t) - F(\theta^*) \le \frac{4L \|\theta_1 - \theta^*\|^2}{(t + 2)^2} = \mathcal{O}\left(\frac{1}{t^2}\right)$$[cite: 540, 609]
  When $F$ is $\sigma$-strongly convex and restarted periodically at intervals proportional to $\sqrt{L/\sigma}$, the convergence rate accelerates to[cite: 616, 617]:
  $$F(\theta_t) - F(\theta^*) \le L \left(1 - \sqrt{\frac{\sigma}{L}}\right)^t \|\theta_0 - \theta^*\|^2$$[cite: 616]
* **Role in the Paper:**
  Explains why momentum can emulate second-order optimization[cite: 540]: it accelerates motion along low-curvature directions while stabilizing high-curvature oscillations, providing the foundation for Chapter 7's results on RNN training[cite: 540, 607, 608].

## 3. Vanishing and Exploding Gradients in RNNs (Section 2.5.1)

### The Backpropagation Through Time (BPTT) Chain Rule
* **Mathematical Formulation:**
  Given hidden transitions $h_t = e(W_{hv} v_t + W_{hh} h_{t-1} + b_h)$ and output $z_T = g(W_{oh} h_T + b_o)$[cite: 544], the gradient of the loss at terminal time $T$ with respect to the recurrent weight matrix $W_{hh}$ is[cite: 544, 545]:
  $$\frac{\partial L(z_T; y_T)}{\partial W_{hh}} = \sum_{t=1}^T dz_t h_{t-1}^T$$[cite: 545]
  where the backpropagated error vector $dz_t$ satisfies[cite: 545]:
  $$dz_t = \left( \prod_{\tau=t+1}^T W_{hh}^T \text{diag}(e'(u_\tau)) \right) W_{oh}^T g'(o_T) \frac{\partial L(z_T; y_T)}{\partial z_T}$$[cite: 545]
* **Spectral Analysis of Vanishing Gradients:**
  The continuous product $\prod_{\tau=t+1}^T W_{hh}^T \text{diag}(e'(u_\tau))$ scales with the matrix power $(W_{hh}^T)^{T-t}$[cite: 545]. If the spectral radius $\rho(W_{hh}) = \max_i |\lambda_i| < 1$, the norm $\|dz_t\|_2 \to 0$ exponentially as temporal lag $T - t$ increases[cite: 545]. Conversely, if $\rho(W_{hh}) > 1$ and units do not saturate, $\|dz_t\|_2$ explodes[cite: 545]. This dynamic instability forms the core problem addressed throughout the thesis[cite: 528, 545].

## 4. Energy-Based Sequence Models: TRBM and RTRBM (Chapter 3)

### Temporal Restricted Boltzmann Machine (TRBM)
* **Conditional Distribution:**
  The TRBM defines a joint distribution over visible vector $v_t \in \{0, 1\}^{N_v}$ and hidden vector $h_t \in \{0, 1\}^{N_h}$ conditioned on history via dynamic biases[cite: 560]:
  $$P(v_t, h_t \mid v_{t-m}^{t-1}, h_{t-m}^{t-1}) = \frac{1}{Z(v_{t-m}^{t-1}, h_{t-m}^{t-1})} \exp\left( h_t^T W v_t + B_v(v_{t-m}^{t-1})^T v_t + B_h(v_{t-m}^{t-1}, h_{t-m}^{t-1})^T h_t \right)$$[cite: 560]
  where dynamic biases are linear autoregressive projections[cite: 560]:
  $$B_v(v_{t-m}^{t-1}) = \sum_{j=1}^m A_j v_{t-j} + b_v, \quad B_h(v_{t-m}^{t-1}, h_{t-m}^{t-1}) = \sum_{j=1}^m W'_j h_{t-j} + \sum_{j=1}^m W_j v_{t-j} + b_h$$[cite: 560]

### Recurrent Temporal RBM (RTRBM) Formulation
* **Deterministic Recurrent State Projection:**
  To resolve the intractable inference of the TRBM, the RTRBM replaces stochastic latent conditioning with a deterministic recurrent mean-field vector $h_t \in [0, 1]^{N_h}$[cite: 566, 567]:
  $$\hat{P}(v_t, h'_t \mid h_{t-1}) = \frac{1}{Z(h_{t-1})} \exp\left( {h'_t}^T W v_t + v_t^T b_v + {h'_t}^T (b_h + W' h_{t-1}) \right)$$[cite: 566]
  $$h_t = \text{sigmoid}(W v_t + b_h + W' h_{t-1})$$[cite: 566, 567, 568]
* **Exact Log-Likelihood Gradient via BPTT and Contrastive Divergence:**
  Because the posterior collapses to a Dirac point-mass $\hat{P}(h_1^T \mid v_1^T) = \delta_{h(v)_1^T}(h_1^T)$, the exact gradient of the sequence log-likelihood is[cite: 568, 569]:
  $$\frac{\partial \log \hat{P}(v_1^T)}{\partial W'} = \sum_{t=2}^T \left( r_t \odot (1 - r_t) \odot \frac{\partial L}{\partial r_t} \right) r_{t-1}^T$$[cite: 569]
  $$\frac{\partial \log \hat{P}(v_1^T)}{\partial W} = \sum_{t=1}^{T-1} \left( {W'}^T \left( r_{t+1} \odot (1 - r_{t+1}) \odot \frac{\partial L}{\partial r_{t+1}} \right) \right) v_t^T + \sum_{t=1}^T \frac{\partial \log \hat{P}(v_t \mid r_{t-1})}{\partial W}$$[cite: 569]
  where the rightmost term is computed via Contrastive Divergence ($CD_k$)[cite: 569].

## 5. Hessian-Free (HF) Optimization and Structural Damping (Chapter 4)


```

```
    HF Optimization with Structural Damping:

```

f(θ) Loss ---> Local Quadratic Approximation M_{θ_n}(δ) + λ R_{θ_n}(δ)
|
Curvature Matrix: B = G_f + λ I + λ μ G_S
|
[G_f]: Generalized Gauss-Newton Matrix (Curvature of output loss)
[λ I]: Tikhonov Regularization (Limits step norm ||δ||^2)
[G_S]: Structural Damping Matrix (Penalizes changes in hidden states h(θ))
|
Solves B δ = -∇f(θ_n) iteratively via Conjugate Gradient (Pearlmutter R-operator)

```

### Generalized Gauss-Newton Matrix
* **Mathematical Formulation:**
  For an objective $f(\theta) = L(g(o(\theta)); y)$, the generalized Gauss-Newton matrix $G_f$ is defined as[cite: 574]:
  $$G_f \equiv J_{o,\theta}^T (L \circ g)'' J_{o,\theta}$$[cite: 574]
  where $J_{o,\theta} = \frac{\partial o}{\partial \theta}$ is the network Jacobian and $(L \circ g)''$ is the Hessian of the matching loss with respect to pre-activations $o$[cite: 574, 575]. When $L$ and $g$ are matched (e.g., cross-entropy with softmax), $(L \circ g)'' \succeq 0$, guaranteeing that $G_f$ is positive semi-definite[cite: 575].

### Structural Damping Formulation
* **The State Divergence Penalty:**
  To prevent updates that disrupt recurrent dynamics across long temporal sequences, the damping function includes a hidden-state distance penalty[cite: 577]:
  $$R_{\theta_n}(\theta) = \frac{1}{2}\|\theta - \theta_n\|^2 + \mu S_{\theta_n}(\theta), \quad \text{where } S_{\theta_n}(\theta) \equiv D(h(\theta); h(\theta_n))$$[cite: 577]
* **Gauss-Newton Curvature of Structural Damping:**
  Linearizing the hidden state pre-activations $u$ leads to the positive semi-definite curvature matrix[cite: 578]:
  $$G_S \equiv J_{u,\theta}^T (D \circ e)'' J_{u,\theta}$$[cite: 578]
* **Total Curvature Matrix:**
  The linear Conjugate Gradient solver optimizes the quadratic using the matrix-vector product[cite: 573, 578]:
  $$B v = \left( G_f + \lambda I + \lambda \mu G_S \right) v$$[cite: 573, 578]
  evaluated in $\mathcal{O}(N_w)$ time using Pearlmutter's $\mathcal{R}\{\cdot\}$-operator forward-backward passes without forming explicit matrices[cite: 573, 575, 578, 586].

## 6. Multiplicative Tensor Recurrent Neural Networks (MRNN) (Chapter 5)

### Tensor Factorization of Recurrent Transitions
* **Full 3-Way Tensor Formulation:**
  In a Tensor RNN, the hidden-to-hidden transition matrix is gated by the current input vector $v_t$[cite: 589]:
  $$h_t = \tanh\left( W_{hv} v_t + W_{hh}^{(v_t)} h_{t-1} + b_h \right), \quad \text{where } W_{hh}^{(v_t)} = \sum_{m=1}^M v_t^{(m)} W_{hh}^{(m)}$$[cite: 589]
  Storing $M$ separate matrices requires $\mathcal{O}(M \cdot N_h^2)$ parameters, which is computationally prohibitive[cite: 589].
* **Low-Rank Factorized MRNN:**
  The tensor is factorized using three matrices $W_{fv} \in \mathbb{R}^{F \times N_v}$, $W_{fh} \in \mathbb{R}^{F \times N_h}$, and $W_{hf} \in \mathbb{R}^{N_h \times F}$[cite: 589, 590]:
  $$W_{hh}^{(v_t)} \equiv W_{hf} \cdot \text{diag}(W_{fv} v_t) \cdot W_{fh}$$[cite: 589]
  The recurrent execution decomposes into two sequential non-linear operations per time step[cite: 590]:
  $$f_t = \text{diag}(W_{fv} v_t) \cdot (W_{fh} h_{t-1})$$[cite: 590]
  $$h_t = \tanh(W_{hf} f_t + W_{hv} v_t)$$[cite: 590]
* **Role in the Paper:**
  This factorized formulation scales as $\mathcal{O}(F(N_v + 2N_h))$[cite: 589, 590]. It allows each input character to dynamically blend rank-one transition matrices, providing the representational capacity needed to track long-range syntax and balance parentheses over dozens of characters[cite: 588, 590].

## 7. Optimal Control and Augmented Hessian-Free (AHF) Optimization (Chapter 6)

### Stochastic Non-Linear Plant Dynamics
* **State Evolution Equation:**
  The discrete-time physics of a 2-joint planar arm under torque controls $u_t$, delayed sensory feedback $\tau$, continuous disturbances $d_t$, and multiplicative command noise $\xi_t^u \sim \mathcal{N}(0, \sigma_u^2 I)$ is modeled as[cite: 599, 600, 601]:
  $$x_{t+1} = f\left(x_t, u_t \odot (1 + \xi_t^u) + d_t, t\right) + \xi_t^f$$[cite: 601]
  The RNN controller observes delayed state feedback $v_{t-\tau} = [\theta, \phi, \dot{\theta}, \dot{\phi}]_{t-\tau}$ and minimizes tracking error and energy expenditure[cite: 600, 601]:
  $$L(\theta) = \mathbb{E}_{x^*, x_0} \left[ \frac{1}{2}\|x_T - x^*\|^2 + \frac{\alpha \Delta t}{2} \sum_{t=0}^T \|u_t\|^2 \right]$$[cite: 600, 601]

### Augmented Hessian-Free (AHF) Optimization
* **Conjugate Subspace Optimization:**
  Standard Conjugate Gradient (CG) minimizes the quadratic model $q(x) = \frac{1}{2} x^T A x + b^T x$ over the $i$-th Krylov subspace[cite: 597, 598]:
  $$\mathcal{K}_i(A, b) \equiv \text{span}\{b, Ab, A^2 b, \dots, A^{i-1} b\}$$[cite: 598]
  Because any conjugate basis $\{c_1, c_2, \dots, c_i\}$ satisfies $c_j^T A c_k = 0$ for $j \neq k$, the quadratic decomposes into independent 1D problems[cite: 598]:
  $$q\left(\sum_j \alpha_j c_j\right) = \sum_j q(\alpha_j c_j) = \sum_j \left( \frac{1}{2} \alpha_j^2 c_j^T A c_j + \alpha_j b^T c_j \right)$$[cite: 598]
* **Role in the Paper:**
  AHF performs a line search over the true non-linear loss along each conjugate direction generated by CG[cite: 598]. This eliminates the need for manual Tikhonov parameter ($\lambda$) tuning and prevents the optimizer from getting trapped in regions where the quadratic approximation fails[cite: 598].

## 8. Spectral Initialization and First-Order Momentum (Chapter 7)


```

```
    Nesterov Momentum vs. Classical Momentum:

```

Classical Momentum:  v_t = μ v_{t-1} - ε ∇f(θ_{t-1})
θ_t = θ_{t-1} + v_t

Nesterov Momentum:   v_t = μ v_{t-1} - ε ∇f(θ_{t-1} + μ v_{t-1})  <-- Evaluates gradient *after* velocity step
θ_t = θ_{t-1} + v_t

```

### Random Sparse Spectral Initialization (Section 7.4.1)
* **Mathematical Definition:**
  To allow information to circulate through recurrent units without exploding or vanishing during initial forward-backward passes, $W_{hh}$ is initialized as follows[cite: 614, 615]:
  1. Each hidden unit connects to exactly $k = 15$ incoming hidden units, with weights sampled from $\mathcal{N}(0, 1)$[cite: 614, 615].
  2. Compute the maximum absolute eigenvalue (spectral radius)[cite: 615]:
     $$\rho(W_{hh}) = \max_i |\lambda_i(W_{hh})|$$
  3. Rescale the matrix[cite: 615]:
     $$W_{hh} \leftarrow W_{hh} \cdot \frac{1.2}{\rho(W_{hh})}$$[cite: 615]
* **Mathematical Role:**
  A spectral radius of $\rho \approx 1.2$ places the non-linear $\tanh$ dynamical system near the "edge of chaos"[cite: 615]. The non-linear saturation of the $\tanh$ squashing function prevents runaway state explosion, while the slightly expansive linear regime prevents gradients from vanishing during the initial optimization steps[cite: 615, 617].

### Analytical Comparison: NAG vs. Classical Momentum (Section 7.2)
* **Classical Momentum:**
  $$v_t = \mu v_{t-1} - \epsilon \nabla f(\theta_{t-1}), \quad \theta_t = \theta_{t-1} + v_t$$[cite: 608]
* **Nesterov's Accelerated Gradient (Rewritten Form):**
  $$v_t = \mu v_{t-1} - \epsilon \nabla f(\theta_{t-1} + \mu v_{t-1}), \quad \theta_t = \theta_{t-1} + v_t$$[cite: 609]
* **Predictive Damping Mechanism:**
  Evaluating the gradient at the predicted position $\theta_{t-1} + \mu v_{t-1}$ provides a negative feedback correction[cite: 609, 610]:
  $$\nabla f(\theta_{t-1} + \mu v_{t-1}) \approx \nabla f(\theta_{t-1}) + \mu \nabla^2 f(\theta_{t-1}) v_{t-1}$$
  The term $\mu \nabla^2 f(\theta_{t-1}) v_{t-1}$ applies Hessian-based damping directly along high-curvature directions, preventing the violent cross-axis oscillations characteristic of standard momentum on anisotropic loss surfaces[cite: 610].
