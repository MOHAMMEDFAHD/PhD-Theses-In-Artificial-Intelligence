# Academic Summary: Learning to Generate Data by Estimating Gradients of the Data Distribution

## Abstract
This doctoral dissertation establishes the foundational theoretical, algorithmic, and statistical frameworks for **score-based generative modeling** (commonly unified with continuous-time **diffusion models**)[cite: 194, 219]. High-dimensional generative modeling has historically been limited by the intractability of normalizing probability distributions[cite: 194, 218]. By working with the **Stein score function**—defined as the gradient of the log probability density $\nabla_x \log p(x)$—generative models can bypass partition functions entirely while preserving unconstrained neural network expressivity[cite: 194, 219, 237, 238]. The dissertation develops scalable training objectives via random projections (Sliced Score Matching)[cite: 24, 241] and high-order denoising estimators[cite: 260], formulates generative sampling via Annealed Langevin Dynamics[cite: 63, 280] and continuous-time Stochastic Differential Equations (SDEs)[cite: 94, 311], introduces deterministic Probability Flow Ordinary Differential Equations (ODEs) for exact likelihood calculation[cite: 98, 103, 315, 320, 343], and solves ill-posed medical imaging inverse problems without paired supervision[cite: 108, 109, 325, 326].

---

## Problems
- **The Normalization Bottleneck:** Traditional probabilistic models require integrating over continuous high-dimensional spaces to ensure $\int p(x) dx = 1$[cite: 218, 229, 230]. Approximating the partition function (e.g., in Energy-Based Models via MCMC) introduces high variance, instability, or bias[cite: 219, 231, 232]; restricting model architectures (e.g., Autoregressive Models, Normalizing Flows, VAEs) constrains model capacity or enforces non-trivial structural transformations[cite: 219, 233, 234, 235]; and implicit formulations like GANs suffer from training instability, mode collapse, and an inability to compute likelihoods[cite: 219, 236].
- **Computational Intractability of Standard Score Matching:** Classical score matching requires evaluating the trace of the Hessian matrix $\text{tr}(\nabla_x^2 \log p(x))$[cite: 26, 243]. Naive automatic differentiation demands $D$ backward passes for $D$-dimensional data, rendering the objective intractable for high-dimensional images[cite: 26, 243].
- **The Manifold Hypothesis and Low-Density Regions:** Real-world data concentrates on low-dimensional manifolds embedded within high-dimensional ambient space[cite: 66, 283]. Ambient score functions are undefined outside the manifold[cite: 66, 283]. Moreover, in low data density regions, score matching lacks training signal[cite: 66, 67, 283, 284], which causes standard Langevin dynamics to fail when attempting to traverse modes or transition between disjoint supports[cite: 67, 68, 284, 285].
- **High-Dimensional Sampling Instability:** Discretized Langevin dynamics suffer from error accumulation in high dimensions, slow mixing, and sensitivity to noise scale hyperparameter schedules[cite: 79, 87, 296, 304].
- **Measurement Overfitting in Medical Inverse Problems:** Supervised deep learning approaches for CT/MRI reconstruction map measurements directly to targets[cite: 109, 326]. They require paired data and overfit to fixed physics models (e.g., a specific acceleration factor or projection count), destabilizing when acquisition settings change[cite: 109, 326].

---

## Proposed Solutions
- **Sliced Score Matching (SSM):** Replaces high-dimensional score matching by projecting score vectors onto random 1D directions $v$ using reverse-mode automatic differentiation (Hessian-vector products), reducing computational complexity from $\mathcal{O}(D)$ to $\mathcal{O}(1)$ without computing Hessian diagonals[cite: 24, 28, 241, 245].
- **Higher-Order Denoising Score Matching ($D_k\text{SM}$):** Extends Tweedie's formula to higher-order moments to learn second-order and higher scores via least-squares objectives without computing explicit Hessians[cite: 44, 46, 47, 49, 261, 263, 264, 266].
- **Noise Conditional Score Networks (NCSN) and Annealed Langevin Dynamics:** Perturbs data with a geometric progression of Gaussian noise levels $\{\sigma_i\}_{i=1}^L$[cite: 69, 286]. A single network is conditioned across all scales, and sampling executes via annealed Langevin dynamics from high noise to low noise, circumventing manifold degeneracies and accelerating multimodal mixing[cite: 69, 71, 286, 288].
- **Continuous-Time Score-Based SDE Framework:** Unifies NCSN/SMLD and Denoising Diffusion Probabilistic Models (DDPM) as continuous-time diffusion processes governed by Itô SDEs[cite: 94, 97, 100, 311, 314, 317]:
  - Variance Exploding (VE) SDE (continuous SMLD)[cite: 101, 318]
  - Variance Preserving (VP) SDE (continuous DDPM)[cite: 101, 318]
  - Sub-VP SDE[cite: 101, 318]
- **Predictor-Corrector (PC) Samplers & Probability Flow ODE:** Introduces a hybrid integration scheme combining numerical SDE solvers (Predictor) with score-based MCMC steps (Corrector)[cite: 102, 319]. Derives an exact deterministic Probability Flow ODE that shares identical marginal probability densities with the SDE, enabling exact likelihood computation, latent code manipulation, and fast numerical integration[cite: 103, 104, 320, 321].
- **Unsupervised Inverse Problem Solving via SDE Projections:** Solves linear inverse problems ($y = Ax + \epsilon$) in an unsupervised manner by combining unconditional score models with a closed-form proximal projection operator exploiting the structure $A = \mathcal{P}(\Lambda) \mathcal{T}$[cite: 110, 114, 115, 117, 327, 331, 332, 334].

---

## Purpose
The primary objective of this dissertation is to establish an unconstrained, stable, and theoretically rigorous probabilistic generative modeling paradigm[cite: 194, 219]. By defining and modeling distributions exclusively through their score functions, the work aims to resolve the fundamental trade-offs between architectural flexibility, sample generation quality, and exact likelihood computation across machine learning and scientific domains[cite: 194, 219].

## Methodology

### 1. Mathematical Formulations

#### Sliced Score Matching (SSM)
Given projection vector $v \sim p_v$ where $\mathbb{E}[v v^T] \succ 0$, the projected Fisher divergence objective is[cite: 28, 245]:
$$J(\theta; p_v) := \mathbb{E}_{p_v} \mathbb{E}_{p_d} \left[ v^T \nabla_x s_m(x; \theta) v + \frac{1}{2} (v^T s_m(x; \theta))^2 \right]$$
For multivariate standard normal or Rademacher distributions, the variance-reduced formulation (SSM-VR) simplifies to[cite: 29, 246]:
$$\hat{J}_{\text{vr}}(\theta; x_1^N, v_{11}^{NM}) := \frac{1}{NM} \sum_{i=1}^N \sum_{j=1}^M \left[ v_{ij}^T \nabla_x s_m(x_i; \theta) v_{ij} + \frac{1}{2} \|s_m(x_i; \theta)\|_2^2 \right]$$

#### Higher-Order Scores via Tweedie's Generalization
For perturbed distributions $q_\sigma(\tilde{x}) = \int p_{\text{data}}(x) \mathcal{N}(\tilde{x} \mid x, \sigma^2 I) dx$, posterior moments relate directly to scores[cite: 46, 48, 263, 265]:
$$\mathbb{E}[x \mid \tilde{x}] = \tilde{x} + \sigma^2 \tilde{s}_1(\tilde{x})$$
$$\mathbb{E}[x x^T \mid \tilde{x}] = \tilde{x}\tilde{x}^T + \sigma^2 \tilde{x}\tilde{s}_1(\tilde{x})^T + \sigma^2 \tilde{s}_1(\tilde{x})\tilde{x}^T + \sigma^4 \tilde{s}_2(\tilde{x}) + \sigma^4 \tilde{s}_1(\tilde{x})\tilde{s}_1(\tilde{x})^T + \sigma^2 I$$
The second-order denoising score matching ($D_2\text{SM}$) loss is minimized via[cite: 50, 267]:
$$\mathcal{L}_{D_2\text{SM}}(\theta) := \mathbb{E}_{p_{\text{data}}(x)} \mathbb{E}_{q_\sigma(\tilde{x} \mid x)} \left[ \left\| \tilde{s}_2(\tilde{x}; \theta) + \tilde{s}_1(\tilde{x}; \theta)\tilde{s}_1(\tilde{x}; \theta)^T + \frac{I - z z^T}{\sigma^2} \right\|_F^2 \right], \quad z = \frac{\tilde{x} - x}{\sigma}$$

#### Forward and Reverse-Time Diffusion SDEs
The continuous forward perturbation is modeled as an Itô SDE[cite: 98, 315]:
$$dx = f(x, t) dt + g(t) dw$$
The reverse-time trajectory running backwards from $t = T$ to $t = 0$ is governed by[cite: 99, 316]:
$$dx = \left[ f(x, t) - g(t)^2 \nabla_x \log p_t(x) \right] dt + g(t) d\bar{w}$$
Where the score network $s_\theta(x, t)$ is trained via the continuous denoising score matching objective[cite: 99, 316]:
$$\min_\theta \mathbb{E}_{t \sim \mathcal{U}(0, T)} \left[ \lambda(t) \mathbb{E}_{x(0)} \mathbb{E}_{x(t) \mid x(0)} \left[ \| s_\theta(x(t), t) - \nabla_{x(t)} \log p_{0t}(x(t) \mid x(0)) \|_2^2 \right] \right]$$

#### Probability Flow ODE
Every diffusion process admits a deterministic ODE whose trajectories share the identical marginal probability densities $p_t(x)$[cite: 103, 320]:
$$dx = \left[ f(x, t) - \frac{1}{2} g(t)^2 \nabla_x \log p_t(x) \right] dt$$

### 2. Algorithmic Implementations
- **Annealed Langevin Dynamics (NCSNv2):** Incorporates geometric noise schedules derived from maximal pairwise distances $\sigma_1 \approx \max \|x^{(i)} - x^{(j)}\|_2$[cite: 83, 84, 300, 301], noise conditioning parameterization $s_\theta(x, \sigma) = s_\theta(x) / \sigma$[cite: 86, 87, 303, 304], and exponential moving average (EMA) of model weights[cite: 88, 89, 305, 306].
- **Predictor-Corrector (PC) Sampling:** At each discretization step, a numerical ODE/SDE step predicts $x_{t-\Delta t}$, followed by multiple score-based Langevin MCMC steps that correct the sample distribution to align with $p_{t-\Delta t}$[cite: 102, 319].
- **Linear Inverse Problem Solver:** Decomposes measurement operators $A = \mathcal{P}(\Lambda) \mathcal{T}$ (where $\mathcal{T}$ is an invertible transform such as Radon or Fourier transforms, and $\mathcal{P}(\Lambda)$ represents subsampling)[cite: 115, 116, 332, 333]. Enforces conditional data fidelity at each backward step via closed-form proximal projection[cite: 117, 334]:
  $$\hat{x}'_t = \mathcal{T}^{-1} \left[ \lambda \Lambda \mathcal{P}^{-1}(\Lambda) \hat{y}_t + (1 - \lambda) \Lambda \mathcal{T} \hat{x}_t + (I - \Lambda) \mathcal{T} \hat{x}_t \right]$$

### Key Empirical Observations
- **Computational Scaling of SSM:** Evaluated on MNIST deep energy-based models (NICE)[cite: 37, 38], SSM-VR achieved equivalent or superior score matching losses while matching the per-iteration running time of Hessian-free alternatives, whereas exact score matching caused out-of-memory errors beyond 400 dimensions[cite: 36, 37].
- **High-Order Uncertainty Capture:** Second-order score models parameterizing low-rank Hessians provided up to $500\times$ speedups over automatic differentiation on MNIST and captured pixel-level denoising covariance[cite: 45, 53]. Ozaki discretization utilizing second-order curvature improved Langevin dynamics effective sample size (ESS) from 21.81 to 28.89 on synthetic targets[cite: 56, 57].
- **ODE Efficiency and Invertibility:** The Probability Flow ODE permitted adaptive Runge-Kutta numerical integration, reducing function evaluations by over 90% without perceptual quality degradation, while establishing uniquely identifiable, bijectively invertible latent codes[cite: 104, 321].

## Conclusions
- **Theoretical Unification:** Score matching, denoising autoencoders, ancestral sampling diffusion models, and Langevin dynamics represent discrete approximations of a continuous underlying diffusion framework governed by forward and reverse Itô SDEs[cite: 94, 97, 100, 311, 314, 317].
- **Overcoming Normalization Limits:** Modeling gradient fields directly circumvents the normalization constraints of probabilistic modeling, allowing arbitrary neural architectures to generate data with sample fidelity surpassing Generative Adversarial Networks[cite: 194, 219, 237, 238].
- **Inherent Determinism and Duality:** Every stochastic diffusion process is mathematically paired with an equivalent deterministic continuous-time process (Probability Flow ODE) that generates identical marginal densities and enables exact maximum-likelihood computation[cite: 103, 320, 343].
- **Universal Inverse Problem Solving:** Decoupling prior modeling from forward measurement processes allows a single unconditional score network to reconstruct degraded observations across disparate physical forward operators, resolving generalization and stability challenges in medical imaging inverse problems[cite: 108, 109, 325, 326].

## 1. Score Functions and Density Independence (Section 2.3)

### Definition of Stein Score Function
* **Mathematical Formulation:**
  For a continuously differentiable probability density function $p(x)$ over $\mathbb{R}^D$, the Stein score function $s(x)$ is defined as the gradient of the log-density with respect to the data vector $x$[cite: 236, 237]:
  $$s(x) := \nabla_x \log p(x)$$
* **Invariance to the Normalizing Constant (Partition Function):**
  If $\tilde{p}(x)$ represents an unnormalized density such that $p(x) = \frac{\tilde{p}(x)}{Z}$ with partition function $Z = \int \tilde{p}(x) \, dx$[cite: 230, 237]:
  $$s(x) = \nabla_x \log \left(\frac{\tilde{p}(x)}{Z}\right) = \nabla_x \log \tilde{p}(x) - \nabla_x \log Z = \nabla_x \log \tilde{p}(x)$$
* **Density Recovery via Line Integrals:**
  A density can be reconstructed from its score field up to an integration constant[cite: 237]:
  $$\log p(x) = \log p(x_0) + \int_0^1 s(x_0 + t(x - x_0))^T (x - x_0) \, dt$$
* **Simple Explanation & Role in the Paper:**
  The score function forms a vector field indicating the directions along which data density increases most rapidly[cite: 219, 279, 296]. Because differentiating with respect to $x$ eliminates the constant $Z$, deep neural networks can directly parameterize complex distributions without computing intractable high-dimensional integrals[cite: 218, 219, 230, 237].

## 2. Classical Score Matching and its Scalability Bottleneck (Section 3.2.1)

### Fisher Divergence Minimization
* **Mathematical Formulation:**
  Score matching minimizes the expected squared Euclidean distance (Fisher divergence) between model scores $s_m(x; \theta) = \nabla_x \log p_m(x; \theta)$ and true data scores $s_d(x) = \nabla_x \log p_d(x)$[cite: 242, 243]:
  $$L(\theta) := \frac{1}{2} \mathbb{E}_{p_d(x)} \left[ \| s_m(x; \theta) - s_d(x) \|_2^2 \right]$$
* **Integration by Parts (Hyvärinen's Objective):**
  Applying integration by parts removes the unknown $s_d(x)$, rewriting the objective up to a parameter-independent constant $C$[cite: 243]:
  $$J(\theta) := \mathbb{E}_{p_d(x)} \left[ \text{tr}(\nabla_x s_m(x; \theta)) + \frac{1}{2} \| s_m(x; \theta) \|_2^2 \right]$$
  where $\nabla_x s_m(x; \theta) = \nabla_x^2 \log \tilde{p}_m(x; \theta)$ denotes the Hessian matrix of the log-density[cite: 243].
* **The Computational Bottleneck:**
  Computing $\text{tr}(\nabla_x s_m(x; \theta)) = \sum_{i=1}^D \frac{\partial^2 \log \tilde{p}_m(x; \theta)}{\partial x_i^2}$ requires $D$ reverse-mode automatic differentiation passes[cite: 243]. When $D$ is in the thousands or millions, this becomes computationally prohibitive[cite: 243].

## 3. Sliced Score Matching (SSM) (Section 3.3 & Appendix A.1)

### Random 1D Projections
* **Mathematical Formulation:**
  Instead of comparing high-dimensional score vectors directly, SSM projects them onto random vectors $v \sim p_v$ (where $\mathbb{E}_{p_v}[v v^T] \succ 0$ and $\mathbb{E}_{p_v}[\|v\|_2^2] < \infty$)[cite: 245]:
  $$L(\theta; p_v) := \frac{1}{2} \mathbb{E}_{p_v} \mathbb{E}_{p_d(x)} \left[ \left( v^T s_m(x; \theta) - v^T s_d(x) \right)^2 \right]$$
  Using integration by parts yields the equivalent tractable objective[cite: 245]:
  $$J(\theta; p_v) = \mathbb{E}_{p_v} \mathbb{E}_{p_d(x)} \left[ v^T \nabla_x s_m(x; \theta) v + \frac{1}{2} (v^T s_m(x; \theta))^2 \right]$$
* **Hessian-Vector Product Efficiency:**
  The second-order term $v^T \nabla_x s_m(x; \theta) v = \nabla_x (v^T s_m(x; \theta))^T v$ can be evaluated using two forward/backward automatic differentiation operations regardless of dimension $D$[cite: 246].

### Variance-Reduced Sliced Score Matching (SSM-VR)
* **Formulation:**
  When $p_v$ has identity covariance ($\mathbb{E}_{p_v}[v v^T] = I$, e.g., standard Gaussian or Rademacher), $\mathbb{E}_{p_v}[(v^T s_m(x; \theta))^2] = \|s_m(x; \theta)\|_2^2$[cite: 246]. Replacing the sample average with its analytic expectation yields[cite: 246]:
  $$\hat{J}_{\text{vr}}(\theta; x_1^N, v_{11}^{NM}) := \frac{1}{NM} \sum_{i=1}^N \sum_{j=1}^M \left[ v_{ij}^T \nabla_x s_m(x_i; \theta) v_{ij} + \frac{1}{2} \| s_m(x_i; \theta) \|_2^2 \right]$$
* **Theoretical Guarantees:**
  - **Consistency (Theorem 3.2):** Under standard identifiability and compactness conditions, for any $M \ge 1$, $\hat{\theta}_{N,M} \xrightarrow{p} \theta^*$ as sample size $N \to \infty$[cite: 247, 248].
  - **Asymptotic Normality (Theorem 3.3):** As $N \to \infty$, $\sqrt{N}(\hat{\theta}_{N,M} - \theta^*) \xrightarrow{d} \mathcal{N}(0, \Sigma)$, where $\Sigma$ scales inversely with projection count $M$[cite: 248, 249].
* **Connection to Hutchinson's Trace Estimator:**
  SSM-VR mathematically equates to replacing the intractable trace $\text{tr}(A)$ in classical score matching with Hutchinson's unbiased stochastic estimator $\mathbb{E}_{p_v}[v^T A v]$[cite: 250].

## 4. Higher-Order Scores and Tweedie’s Generalization (Chapter 4 & Appendix A.2)

### Tensor Scores (Definition 4.1)
* **Mathematical Formulation:**
  The $k$-th order score is defined as the $k$-th order partial derivative tensor of the log-density[cite: 262]:
  $$[s_k(x)]_{i_1 i_2 \dots i_k} := \frac{\partial^k}{\partial x_{i_1} \partial x_{i_2} \dots \partial x_{i_k}} \log p(x)$$
  where $s_1(x) = \nabla_x \log p(x)$ is the gradient vector, and $s_2(x) = \nabla_x^2 \log p(x)$ is the Hessian matrix[cite: 262].

### Tweedie’s Formula & Posterior Moments
* **First-Order Moment (Classical Tweedie):**
  For Gaussian noise perturbation $q_\sigma(\tilde{x} \mid x) = \mathcal{N}(\tilde{x} \mid x, \sigma^2 I)$ and marginal noisy density $q_\sigma(\tilde{x}) = \int p_{\text{data}}(x) q_\sigma(\tilde{x} \mid x) \, dx$[cite: 263]:
  $$\mathbb{E}[x \mid \tilde{x}] = \tilde{x} + \sigma^2 \tilde{s}_1(\tilde{x}), \quad \text{where } \tilde{s}_1(\tilde{x}) = \nabla_{\tilde{x}} \log q_\sigma(\tilde{x})$$
* **Second-Order Moment (Theorem 4.1):**
  Extending to conditional second moments[cite: 265]:
  $$\mathbb{E}[x x^T \mid \tilde{x}] = \tilde{x}\tilde{x}^T + \sigma^2 \tilde{x}\tilde{s}_1(\tilde{x})^T + \sigma^2 \tilde{s}_1(\tilde{x})\tilde{x}^T + \sigma^4 \tilde{s}_2(\tilde{x}) + \sigma^4 \tilde{s}_1(\tilde{x})\tilde{s}_1(\tilde{x})^T + \sigma^2 I$$
* **Posterior Covariance (Uncertainty Quantification):**
  Subtracting $\mathbb{E}[x \mid \tilde{x}] \mathbb{E}[x \mid \tilde{x}]^T$ from $\mathbb{E}[x x^T \mid \tilde{x}]$ yields[cite: 270]:
  $$\text{Cov}[x \mid \tilde{x}] = \sigma^4 \tilde{s}_2(\tilde{x}) + \sigma^2 I$$
* **Second-Order Denoising Score Matching ($D_2\text{SM}$):**
  Rephrased as a least-squares problem predicting the second moment[cite: 265, 267]:
  $$\mathcal{L}_{D_2\text{SM}}(\theta) = \mathbb{E}_{p_{\text{data}}(x)} \mathbb{E}_{q_\sigma(\tilde{x} \mid x)} \left[ \left\| \tilde{s}_2(\tilde{x}; \theta) + \tilde{s}_1(\tilde{x}; \theta)\tilde{s}_1(\tilde{x}; \theta)^T + \frac{I - z z^T}{\sigma^2} \right\|_F^2 \right], \quad z = \frac{\tilde{x} - x}{\sigma}$$
* **Simple Explanation & Role in the Paper:**
  Rather than estimating curvature by computing derivatives through existing networks (which is computationally slow and magnifies numerical errors)[cite: 261], this formulation models the Hessian directly as a neural network and trains it through denoising regression[cite: 261, 262].

## 5. Noise Conditional Score Networks (NCSN) and Annealed Langevin Dynamics (Chapters 5 & 6)

### Discrete Multi-Scale Noise Perturbation
* **Geometric Noise Schedule:**
  A sequence $\{\sigma_i\}_{i=1}^L$ is configured such that $\frac{\sigma_1}{\sigma_2} = \dots = \frac{\sigma_{L-1}}{\sigma_L} = \gamma > 1$[cite: 286].
* **Unified Training Objective:**
  A single network $s_\theta(x, \sigma)$ is trained over all scales simultaneously[cite: 287]:
  $$\mathcal{L}(\theta; \{\sigma_i\}_{i=1}^L) = \frac{1}{L} \sum_{i=1}^L \frac{\sigma_i^2}{2} \mathbb{E}_{p_{\text{data}}(x)} \mathbb{E}_{\tilde{x} \sim \mathcal{N}(x, \sigma_i^2 I)} \left[ \left\| s_\theta(\tilde{x}, \sigma_i) + \frac{\tilde{x} - x}{\sigma_i^2} \right\|_2^2 \right]$$
  Weighting by $\sigma_i^2$ balances the loss magnitudes because $\|s_\theta(x, \sigma)\|_2 \propto \frac{1}{\sigma}$[cite: 287, 288].

### Theoretical Principles for Noise Configuration (Chapter 6)
* **Maximum Distance Heuristic (Technique 6.1):**
  For a Gaussian mixture $\hat{p}_{\sigma_1}(x) = \frac{1}{N} \sum_{i=1}^N \mathcal{N}(x \mid x^{(i)}, \sigma_1^2 I)$, the probability mass assigned by component $i$ to component $j$ is bounded by[cite: 300]:
  $$\mathbb{E}_{p^{(i)}(x)} [r^{(j)}(x)] \le \frac{1}{2} \exp\left( - \frac{\|x^{(i)} - x^{(j)}\|_2^2}{8 \sigma_1^2} \right)$$
  Setting $\sigma_1 \ge \max_{i,j} \|x^{(i)} - x^{(j)}\|_2$ prevents transitions between separated modes from collapsing to zero during early sampling steps[cite: 300, 301].
* **Geometric Ratio Derivation via Hyperspherical Shells (Technique 6.2):**
  In high dimensions ($D \to \infty$), the radial distribution $r = \|x\|_2$ of $x \sim \mathcal{N}(0, \sigma^2 I)$ concentrates within a thin spherical shell[cite: 302]:
  $$r \sim \mathcal{N}\left(\sqrt{D}\sigma, \frac{\sigma^2}{2}\right)$$
  To ensure adequate overlap between adjacent distributions $p_{\sigma_i}(r)$ and $p_{\sigma_{i-1}}(r)$, the ratio $\gamma = \sigma_{i-1}/\sigma_i$ is chosen such that the overlap probability satisfies[cite: 302]:
  $$\Phi(\sqrt{2D}(\gamma - 1) + 3\gamma) - \Phi(\sqrt{2D}(\gamma - 1) - 3\gamma) \approx 0.5$$
* **Variance Ratio Invariance for Annealed Langevin Dynamics (Proposition 6.3 & Technique 6.4):**
  When sampling with step size $\alpha_i = \epsilon \cdot \frac{\sigma_i^2}{\sigma_L^2}$, the variance $s_T^2$ after $T$ steps evolves according to[cite: 304]:
  $$\frac{s_T^2}{\sigma_i^2} = \left( 1 - \frac{\epsilon}{\sigma_L^2} \right)^{2T} \left( \gamma^2 - \frac{2\epsilon}{\sigma_L^2 - \sigma_L^2 (1 - \frac{\epsilon}{\sigma_L^2})^2} \right) + \frac{2\epsilon}{\sigma_L^2 - \sigma_L^2 (1 - \frac{\epsilon}{\sigma_L^2})^2}$$
  The parameters $\epsilon$ and $T$ are selected to bring the ratio $\frac{s_T^2}{\sigma_i^2}$ as close to 1 as possible[cite: 304, 305].

| :--- | :--- | :--- |
| **The Normalization Bottleneck in Generative Modeling**[cite: 194, 218] | Ensuring $\int p(x) dx = 1$ in high dimensions is intractable[cite: 218, 229, 230]. EBMs rely on expensive MCMC or contrastive divergence with high variance[cite: 219, 230, 231, 232]; Autoregressive models, Normalizing Flows, and VAEs restrict architectures or lower-bound likelihoods[cite: 219, 232, 233, 234, 235]; GANs bypass normalization entirely but exhibit unstable adversarial training and cannot evaluate likelihoods[cite: 219, 235, 236]. | **Score-Based Modeling via Stein Scores:** Parameterizes distributions via their score function $\nabla_x \log p(x)$, which is mathematically independent of the partition function[cite: 194, 219, 237]. This allows unconstrained, flexible deep neural architectures to be trained stably without adversarial objectives[cite: 194, 220, 238]. |
| **Computational Intractability of Standard Score Matching**[cite: 220, 221, 241] | Standard score matching requires computing the trace of the Hessian $\text{tr}(\nabla_x^2 \log p(x))$[cite: 243]. Evaluating this trace naively requires $D$ backpropagation passes for $D$-dimensional data, which is computationally prohibitive for high-dimensional inputs[cite: 243, 244]. Approximations like Curvature Propagation or approximate backprop suffer from large variance, bias, or lack of software support[cite: 241, 251]. | **Sliced Score Matching (SSM):** Projects score functions onto random 1D vectors $v \sim p_v$[cite: 241, 245]. This replaces the trace with Hessian-vector products $v^T \nabla_x s_m(x) v$, evaluable in $\mathcal{O}(1)$ reverse-mode automatic differentiation passes, while preserving estimator consistency and asymptotic normality[cite: 241, 246, 247, 248]. |
| **Degeneracy Under the Manifold Hypothesis and Low Data Density**[cite: 221, 283, 284] | When data concentrates on low-dimensional manifolds embedded in high-dimensional ambient space, ambient scores are undefined and score matching estimators become inconsistent[cite: 283]. Furthermore, data scarcity in low-density regions leaves scores poorly estimated, preventing Langevin dynamics from transitioning between isolated distribution modes[cite: 283, 284, 285]. | **Multi-Scale Gaussian Perturbation & Annealed Langevin Dynamics (NCSN):** Perturbs data with a geometric hierarchy of noise scales $\{\sigma_i\}_{i=1}^L$[cite: 286]. A single Noise Conditional Score Network learns scores across all scales, and annealed Langevin dynamics sample sequentially from high noise (filling low-density regions) to low noise (collapsing to clean data)[cite: 286, 288]. |
| **Heuristic Noise Configuration and Sampling Instability at High Resolutions**[cite: 295, 296] | Fixed heuristic noise schedules (e.g., $L=10$ scales on $32 \times 32$ images) fail to scale to larger resolutions ($>64 \times 64$), resulting in severe color artifacts, slow mixing, or divergence[cite: 296, 299, 305, 308]. | **NCSNv2 Principles & Exponential Moving Average (EMA):** Analytically sets $\sigma_1$ to the maximum pairwise data distance[cite: 301] and specifies geometric progression ratios $\gamma$ based on high-dimensional Gaussian hyperspherical shell overlap[cite: 302]. | **Theoretical Disconnect Between SMLD and DDPM Formulations**[cite: 311, 314] | Score Matching with Langevin Dynamics (SMLD) and Denoising Diffusion Probabilistic Models (DDPM) were viewed as separate discrete model paradigms, limiting cross-pollination of sampling procedures, architectural designs, and continuous-time theoretical analysis[cite: 311, 314, 315]. | **Continuous-Time Score SDE Framework:** Unifies both paradigms as continuous diffusion processes mapped to Itô SDEs (Variance Exploding SDE for SMLD; Variance Preserving SDE for DDPM)[cite: 311, 314, 317, 318]. Sample generation is unified under Anderson's reverse-time SDE[cite: 312, 316], introducing hybrid Predictor-Corrector (PC) solvers that outperform individual ancestral or MCMC samplers[cite: 312, 319, 320]. |
| **Inability to Compute Exact Likelihoods in Diffusion Models**[cite: 219, 224, 343] | Diffusion probabilistic models and score-based methods historically lacked exact likelihood evaluation, relying primarily on surrogate lower bounds (ELBO) or qualitative visual metrics[cite: 232, 234, 314, 343]. | **Probability Flow ODE:** Formulates a deterministic ODE sharing the identical marginal probability densities $p_t(x)$ with the reverse SDE[cite: 312, 320]. Enables exact log-likelihood calculation via continuous instantaneous change-of-variables[cite: 224, 343], uniquely identifiable deterministic latent encoding[cite: 320, 321], and adaptive numerical integration reducing function evaluations by $>90\%$[cite: 321]. |
| **Physics Overfitting in Supervised Medical Inverse Problems**[cite: 326] | Standard deep learning solvers for CT/MRI reconstruction train end-to-end models on paired images and synthetic measurements[cite: 326]. They overfit to fixed physical acquisition parameters (e.g., fixed projection angles or k-space acceleration factors) and severely degrade when measurement settings shift[cite: 326, 340]. | **Unsupervised Inverse Solving via Proximal Data Consistency:** Employs an unconditional score model trained solely on medical images as an expressive prior[cite: 326, 328, 331]. At test time, decomposes linear operators as $A = \mathcal{P}(\Lambda) \mathcal{T}$ and enforces measurement fidelity at each reverse diffusion step via a closed-form proximal optimization projection, adapting flexibly across arbitrary measurement ratios without retraining[cite: 331, 332, 334]. |
