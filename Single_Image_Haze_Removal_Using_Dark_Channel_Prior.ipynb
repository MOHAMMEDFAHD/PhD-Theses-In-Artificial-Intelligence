# Academic Summary: Single Image Haze Removal Using Dark Channel Prior

## Abstract
This doctoral dissertation addresses the fundamental ill-posed problem of single-image haze removal in computer vision and computational photography[cite: 382, 386]. Based on empirical and statistical analysis of outdoor haze-free imagery, the thesis establishes the **Dark Channel Prior**: in most local non-sky patches of haze-free images, at least one color channel contains pixels with near-zero intensity[cite: 382, 408]. Inverting the Koschmieder optical haze model with this prior yields an immediate, robust estimate of atmospheric transmission and airlight[cite: 382, 418, 419]. To eliminate halo artifacts caused by patch-wise spatial discontinuities, the dissertation develops two major edge-aware refinement algorithms: an unconventional large-kernel Conjugate Gradient linear solver leveraging an $\mathcal{O}(N)$ Matting Laplacian matrix-vector multiplier[cite: 382, 450, 456, 457], and the **Guided Filter**, a non-iterative $\mathcal{O}(N)$ explicit edge-preserving filter that enables real-time execution while avoiding the gradient reversal artifacts characteristic of bilateral filtering[cite: 382, 450, 465, 470, 475]. Finally, psychophysical illusion experiments provide evidence that human visual mechanisms perceive transparency and haze in a manner consistent with the dark channel prior[cite: 382, 488, 500].

---

## Problems
- **Inherent Mathematical Ambiguity of the Single-Image Haze Model:** Optical scattering degrades scenes via multiplicative direct attenuation and additive airlight according to Koschmieder's physical law:
  $$I(x) = J(x)t(x) + A(1 - t(x)) \quad[cite: 391]$$
  For an RGB image with $N$ pixels, there are $3N$ observed values ($I$) against $4N + 3$ unknowns ($3N$ scene radiance values $J$, $N$ transmission values $t$, and 3 atmospheric light components $A$)[cite: 396]. The problem is severely under-constrained and requires spatial priors to resolve[cite: 396, 402].
- **Limitations of Multi-Image Dehazing Approaches:** Methods utilizing multiple aligned photographs taken under differing weather conditions, distinct polarization angles, or externally provided 3D geometry require specialized acquisition hardware, complex geometric registration, or static scenes over long capture intervals, making them impractical for general photography and surveillance[cite: 398, 400, 401].
- **Artifacts and Invalidation of Prior Single-Image Methods:**
  - *Dark Object Subtraction:* Assumes spatially invariant transmission, which fails entirely across depth discontinuities[cite: 402].
  - *Visibility / Contrast Maximization (Tan, 2008):* Maximizes local gradient energy, but violates physical constraints by truncating out-of-gamut values, creating unnatural color over-saturation and severe halo artifacts[cite: 403, 404, 432].
  - *Independent Component Analysis (Fattal, 2008):* Assumes statistical independence between surface albedo and transmission[cite: 404, 405]. It requires high localized chromatic variance and fails in thick haze or low-contrast regions where local statistics degrade[cite: 405, 442].
- **Refinement Complexity and Edge-Preservation Pitfalls:** Combining pixel-wise transmission estimations with spatial smoothness via Markov Random Fields requires inverting large sparse linear systems (e.g., standard Matting Laplacian), scaling as $\mathcal{O}(N^{3/2})$[cite: 450]. Alternative local filters (such as the bilateral filter) are computationally expensive and suffer from gradient reversal artifacts during detail enhancement[cite: 465, 474, 476].

---

## Proposed Solutions
- **The Dark Channel Prior:** A statistical prior based on the observation that dark pixels (originating from shadows, dark objects, or saturated colorful surfaces) are ubiquitous in natural haze-free outdoor scenes[cite: 408, 409]. In hazy images, the elevated intensity of these dark pixels directly reflects additive atmospheric airlight[cite: 407, 417].
- **Direct Transmission and Atmospheric Light Estimation:** Direct mathematical evaluation of patch-level dark channels yields an explicit coarse transmission estimate without requiring iterative optimization[cite: 418, 419]. A robust atmospheric light estimator selects the top 0.1% brightest pixels in the hazy image's dark channel to identify the densest haze regions rather than relying on global image maxima[cite: 424, 427, 428].
- **Accelerated Large-Kernel Matting Laplacian Solver:** An exact reformalization of the Matting Laplacian matrix-vector product $q = L p$ using sliding-window box-filter box summations, driving per-iteration costs down from $\mathcal{O}(r^2 N)$ to $\mathcal{O}(N)$[cite: 456, 457, 458]. This enables larger filter radii $r$, which reduces the matrix condition number and overall solver time from $\mathcal{O}(r N^{3/2})$ to $\mathcal{O}(\frac{1}{r} N^{3/2})$[cite: 450, 456, 459].
- **The Guided Image Filter:** An explicit, non-iterative edge-preserving filtering algorithm derived from a local linear model $q_i = a_k I_i + b_k$[cite: 468, 469]. It runs in $\mathcal{O}(N)$ time independent of kernel radius $r$, avoids solving linear systems, preserves directional gradients, and is free of gradient reversal artifacts[cite: 450, 470, 474, 476].

---

## Purpose
The primary purpose of this dissertation is to formulate a physically grounded, mathematically robust, and computationally efficient framework for single-image haze removal and depth extraction[cite: 382, 386]. Additionally, it aims to establish a generic class of high-speed edge-aware filtering tools applicable across computer vision[cite: 382, 449], while demonstrating that the underlying prior reflects functional mechanisms in human visual perception[cite: 382, 488].

## Methodology

### 1. Mathematical Formulation of the Dark Channel Prior
For an arbitrary scene radiance image $J$, the dark channel $J^{\text{dark}}$ is formally defined as:
$$J^{\text{dark}}(x) = \min_{x' \in \Omega(x)} \left( \min_{c \in \{r, g, b\}} J_c(x') \right) \quad[cite: 411]$$
where $\Omega(x)$ is a local square patch centered at pixel $x$, and $J_c$ denotes the color channel intensity[cite: 411]. The Dark Channel Prior states that for natural, haze-free outdoor images (excluding sky regions)[cite: 413]:
$$J^{\text{dark}}(x) \to 0 \quad[cite: 413]$$

### 2. Inversion of the Optical Model for Transmission Estimation
Koschmieder's atmospheric scattering equation is normalized channel-wise by the estimated atmospheric light vector $A_c$[cite: 391, 418]:
$$\frac{I_c(x)}{A_c} = t(x) \frac{J_c(x)}{A_c} + 1 - t(x) \quad[cite: 418]$$
Assuming that the transmission is locally constant over a patch $\Omega(x)$, denoted as $\tilde{t}(x)$, the minimum operators are applied over spatial coordinates and color channels[cite: 418]:
$$\min_{x' \in \Omega(x)} \left( \min_c \frac{I_c(x')}{A_c} \right) = \tilde{t}(x) \min_{x' \in \Omega(x)} \left( \min_c \frac{J_c(x')}{A_c} \right) + 1 - \tilde{t}(x) \quad[cite: 418]$$
Applying the Dark Channel Prior ($\min_{x' \in \Omega(x)} \min_c \frac{J_c(x')}{A_c} \approx 0$) eliminates the scene radiance term, yielding the coarse transmission estimate[cite: 418, 419]:
$$\tilde{t}(x) = 1 - \min_{x' \in \Omega(x)} \left( \min_c \frac{I_c(x')}{A_c} \right) \quad[cite: 419]$$
In sky regions where depth $d \to \infty$ and $I_c \to A_c$, the normalized dark channel approaches 1, resulting in $\tilde{t}(x) \to 0$, which accurately models true zero transmission[cite: 419].

### 3. Atmospheric Light ($A$) Estimation
Rather than taking the brightest pixel of the entire image (which often corresponds to white cars or reflective surfaces), the algorithm:
1. Selects the top 0.1% brightest pixels in the hazy image dark channel $I^{\text{dark}}(x) = \min_{x' \in \Omega(x)} \min_c I_c(x')$[cite: 427].
2. Identifies the pixel within this subset that exhibits the highest original intensity in the input image $I$, setting its color vector as $A$[cite: 427].

### 4. Transmission Refinement Formulations

#### Soft Matting Formulation
The transmission map is regularized to respect boundaries by minimizing a constrained Markov Random Field cost function[cite: 420, 423]:
$$E(t) = \lambda \|t - \tilde{t}\|_2^2 + t^T L t \quad[cite: 423]$$
leading to the sparse symmetric linear system[cite: 423]:
$$(L + \lambda U) t = \lambda \tilde{t} \quad[cite: 423]$$
where $U$ is the identity matrix, and $L$ is the Matting Laplacian matrix with $(i,j)$-th entry defined as[cite: 423]:
$$L_{ij} = \sum_{k \mid (i,j) \in w_k} \left( \delta_{ij} - \frac{1}{|w_k|} \left( 1 + (I_i - \mu_k)^T \left( \Sigma_k + \frac{\epsilon}{|w_k|} U_3 \right)^{-1} (I_j - \mu_k) \right) \right) \quad[cite: 423]$$
Here $\mu_k$ and $\Sigma_k$ denote the color mean vector and $3 \times 3$ covariance matrix within local window $w_k$, while $\epsilon$ is a regularization factor[cite: 423].

#### Large-Kernel Linear Solver ($\mathcal{O}(N)$ Matrix-Vector Multiplication)
Instead of constructing $L$ explicitly, the matrix-vector product $q = L p$ for arbitrary vector $p$ is computed in $\mathcal{O}(N)$ time via three sequential box-filter passes[cite: 456, 457, 458]:
1. $a_k^* = \left( \Sigma_k + \frac{\epsilon}{|w_k|} U_3 \right)^{-1} \text{cov}_k(I, p)$[cite: 457, 471]
2. $b_k^* = \bar{p}_k - a_k^{*T} \mu_k$[cite: 457]
3. $q_i = |w| p_i - \left( \left( \sum_{k \in w_i} a_k^* \right)^T I_i + \sum_{k \in w_i} b_k^* \right)$[cite: 457]

#### Guided Image Filter Formulation
The guided filter explicitly computes the refined transmission $q_i$ as a local linear transform of guidance image $I$[cite: 468, 469]:
$$q_i = a_k I_i + b_k \quad \forall i \in w_k \quad[cite: 468, 469]$$
Minimizing the squared difference with input $p = \tilde{t}$ within local window $w_k$[cite: 469]:
$$E(a_k, b_k) = \sum_{i \in w_k} \left( (q_i - p_i)^2 + \epsilon a_k^2 \right) \quad[cite: 469]$$
yields the closed-form regression coefficients[cite: 469]:
$$a_k = \frac{\text{cov}_k(I, p)}{\sigma_k^2 + \epsilon}, \quad b_k = \bar{p}_k - a_k \mu_k \quad[cite: 469]$$
Averaging $(a_k, b_k)$ across all overlapping windows containing pixel $i$ produces the output in $\mathcal{O}(N)$ time[cite: 469, 470]:
$$q_i = \bar{a}_i I_i + \bar{b}_i \quad[cite: 470]$$

### 5. Scene Radiance Recovery and Aerial Perspective
Given refined transmission $t(x)$ and atmospheric light $A$, the true radiance is recovered by inverted direct attenuation, clipped by lower bound $t_0 = 0.1$ to avoid noise amplification in dense haze[cite: 428]:
$$J_c(x) = \frac{I_c(x) - A_c}{\max(t(x), t_0)} + A_c \quad[cite: 428]$$
To preserve natural aerial perspective and depth perception, transmission is adaptively modulated by a retention parameter $\kappa = 0.95$[cite: 430]:
$$t(x) := 1 - \kappa (1 - t(x)) \quad[cite: 430]$$

## Conclusions
- **Validity of the Dark Channel Statistical Invariant:** Natural outdoor image patches inherently contain low-reflectance pixels across shadows, geometry, and saturation[cite: 408, 409]. This prior simplifies single-image dehazing by providing an explicit, non-iterative constraint for scene transmission[cite: 382, 418, 419].
- **Algorithmic Duality Between Inversion and Filtering:** The Matting Laplacian matrix is algebraically equivalent to a high-pass form of the Guided Filter ($L \approx |w|(U - W)$)[cite: 476, 477]. Consequently, explicit linear guided filtering acts as an approximate one-step solver for the underlying diffusion system, bypassing iterative linear solvers without degrading edge fidelity[cite: 477, 478, 481].
- **Resolution of Bilateral Filtering Artifacts:** By enforcing a localized linear slope relationship between guidance and output ($\nabla q \approx \bar{a} \nabla I$), the Guided Filter guarantees positive directional consistency, providing a non-iterative, halo-free alternative for edge-preserving smoothing, detail enhancement, and HDR tone mapping[cite: 470, 474, 476, 482].
- **Biophysical Consistency with Human Vision:** Psychological illusion experiments reveal that the human visual system perceives brightness and transparency via a mechanism functionally analogous to the dark channel prior, suggesting that human haze perception relies on evolutionary adaptations to natural scene statistics[cite: 382, 488, 499, 500, 501].

## 1. Physical Optics, Atmospheric Scattering, and Problem Ambiguity (Chapter 2 & Appendix A)

### Koschmieder's Optical Haze Imaging Equation (Section 2.1 & Appendix A)
* **Mathematical Formulation:**
  The degradation of light through a scattering medium is described by the linear convex combination of direct attenuation and airlight[cite: 391, 392]:
  $$I(x) = J(x)t(x) + A(1 - t(x)) \quad \text{or component-wise} \quad I_c(x) = J_c(x)t(x) + A_c(1 - t(x)) \quad[cite: 391, 396]$$
  where $x = (x, y)$ denotes 2D pixel coordinates, $c \in \{r, g, b\}$ is the color channel index, $I(x)$ is the observed intensity vector, $J(x)$ is the true scene radiance (haze-free image), $A$ is the global atmospheric light vector, and $t(x) \in [0, 1]$ is the medium transmission[cite: 391, 392, 396].
* **Physical Derivation (Appendix A):**
  - *Direct Attenuation:* Light flux $E$ traveling through an infinitesimal atmospheric slab of thickness $dz$ is attenuated via total scattering coefficient $\beta(\lambda)$[cite: 505, 506]:
    $$dE(z, \lambda) = -\beta(z, \lambda)E(z, \lambda)dz \implies E(d, \lambda) = E(0, \lambda) e^{-\int_0^d \beta(z, \lambda)dz} \quad[cite: 506]$$
  - *Airlight Generation:* An infinitesimal atmospheric cylinder lit isotropically by atmospheric flux $\varepsilon(\lambda)$ scatters light toward the observer, which is attenuated over the remaining path $z$[cite: 507, 508]:
    $$dL(z, \lambda) = e^{-\int_0^z \beta(z', \lambda)dz'} \beta(z, \lambda)\varepsilon(\lambda)dz \implies L(d, \lambda) = \varepsilon(\lambda)\left(1 - e^{-\int_0^d \beta(z, \lambda)dz}\right) \quad[cite: 508]$$
  - *Wavelength Independence & Colorimetry:* Under particle radii typical of haze ($10^{-2}\,\mu\text{m}$ to $10^{1}\,\mu\text{m}$), $\beta$ is approximately independent of wavelength $\lambda$[cite: 509]. Integrating across camera spectral response $s_c(\lambda)$ produces the macroscopic imaging equation with transmission defined as[cite: 509, 510]:
    $$t(x) = \exp\left( -\int_0^{d(x)} \beta(z)dz \right) \quad[cite: 393, 510]$$
    When the atmosphere is homogeneous ($\beta = \text{const}$), transmission links depth $d(x)$ directly up to an unknown scale[cite: 393, 394]:
    $$t(x) = e^{-\beta d(x)} \iff d(x) = -\frac{\ln t(x)}{\beta} \quad[cite: 394]$$
* **Mathematical Ambiguity (Degrees of Freedom):**
  For an RGB image with $N$ pixels, there are $3N$ scalar equations against $4N + 3$ scalar unknowns ($3N$ for $J$, $N$ for $t$, and $3$ for $A$)[cite: 396]. The problem is under-constrained by $N + 3$ degrees of freedom, requiring at least one independent constraint per pixel[cite: 396, 402].

## 2. The Dark Channel Prior and Statistical Verification (Chapter 3)

### Dark Pixel and Dark Channel Formulation (Section 3.1.2)
* **Dark Pixel Condition:**
  A pixel $x$ is defined as a dark pixel if its minimum value across RGB channels is zero or bounded by an infinitesimal threshold $\delta \ll 1$[cite: 411]:
  $$\min_{c \in \{r, g, b\}} J_c(x) \le \delta \approx 0 \quad[cite: 411]$$
* **Dark Channel Operator:**
  For an image $J$, the dark channel $J^{\text{dark}}$ is the outcome of a channel-wise minimum followed by a spatial morphological minimum filter over a local patch $\Omega(x)$ centered at $x$[cite: 411, 413]:
  $$J^{\text{dark}}(x) \equiv \min_{x' \in \Omega(x)} \left( \min_{c \in \{r, g, b\}} J_c(x') \right) \quad[cite: 411]$$
* **The Dark Channel Prior:**
  For outdoor, haze-free images (excluding the sky), the dark channel satisfies[cite: 413]:
  $$J^{\text{dark}}(x) \approx 0 \quad[cite: 413]$$
* **Statistical Empirical Verification (Section 3.1.3):**
  Evaluated on 5,000 outdoor haze-free images (with sky cropped, using $15 \times 15$ patches)[cite: 414]:
  - $\approx 75\%$ of all pixels in the dark channels have an exact intensity of $0$[cite: 414].
  - $\approx 86\%$ of pixels fall in the range $[0, 16)$ (on a $0\text{--}255$ scale)[cite: 414].
  - $\approx 90\%$ of pixels have intensity $\le 25$ ($< 0.1$ in normalized gray scale)[cite: 414].

## 3. Transmission Estimation and Atmospheric Light Inversion (Section 3.2)

### Derivation of Patch Transmission
* **Mathematical Derivation:**
  Assuming atmospheric light $A_c > 0$ is known, divide Koschmieder's equation by $A_c$[cite: 418]:
  $$\frac{I_c(x')}{A_c} = t(x') \frac{J_c(x')}{A_c} + 1 - t(x') \quad[cite: 418]$$
  Assuming transmission is locally invariant inside patch $\Omega(x)$ (denoted as $\tilde{t}(x)$) and taking the minimum over all pixels $x' \in \Omega(x)$ and color channels $c$[cite: 418]:
  $$\min_{x' \in \Omega(x)} \left( \min_c \frac{I_c(x')}{A_c} \right) = \tilde{t}(x) \min_{x' \in \Omega(x)} \left( \min_c \frac{J_c(x')}{A_c} \right) + 1 - \tilde{t}(x) \quad[cite: 418]$$
  Because $A_c > 0$ and the dark channel prior dictates $\min_{x' \in \Omega(x)} \min_c J_c(x') \approx 0$, the first term on the right vanishes[cite: 418, 419]:
  $$\tilde{t}(x) = 1 - \min_{x' \in \Omega(x)} \left( \min_c \frac{I_c(x')}{A_c} \right) = 1 - \hat{I}^{\text{dark}}(x) \quad[cite: 419]$$
  where $\hat{I}$ represents the hazy image normalized channel-by-channel by $A_c$[cite: 419].
* **Handling Sky Regions:**
  In sky regions, the physical depth $d \to \infty \implies t(x) \to 0$[cite: 419]. Since sky color is dominated by pure airlight ($I_c \to A_c$), the normalized dark channel satisfies $\min_{x' \in \Omega(x)} \min_c \frac{I_c(x')}{A_c} \to 1$, which naturally yields $\tilde{t}(x) \to 0$ without requiring explicit sky segmentation[cite: 419].

### Atmospheric Light ($A$) Estimation via Extremal Order Statistics
* **Statistical Formulation:**
  Rather than choosing the global maximum $\max_x I(x)$ (which is susceptible to white cars, artificial lights, or bright surfaces)[cite: 424, 426], the method:
  1. Computes the hazy image dark channel: $I^{\text{dark}}(x) = \min_{x' \in \Omega(x)} \min_c I_c(x')$[cite: 427].
  2. Selects the top $0.1\%$ brightest pixels in $I^{\text{dark}}$: $\mathcal{S}_{\text{dense}} = \{ x \mid I^{\text{dark}}(x) \ge Q_{0.999}(I^{\text{dark}}) \}$[cite: 427].
  3. Estimates $A$ as the pixel in $\mathcal{S}_{\text{dense}}$ with the maximum original intensity in $I$[cite: 427]:
     $$A = \arg\max_{x \in \mathcal{S}_{\text{dense}}} \|I(x)\| \quad[cite: 427]$$

### Scene Radiance Inversion and Aerial Perspective
* **Bounded Inversion:**
  To prevent noise amplification where $t(x) \to 0$, transmission is bounded below by $t_0 \approx 0.1$[cite: 428]:
  $$J_c(x) = \frac{I_c(x) - A_c}{\max(t(x), t_0)} + A_c \quad[cite: 428]$$
* **Aerial Perspective Parameterization:**
  To retain human depth cues, transmission is linearly blended toward a small residual haze fraction using $\kappa = 0.95$[cite: 430]:
  $$t(x) \leftarrow 1 - \kappa(1 - t(x)) \quad[cite: 430]$$

## 4. Transmission Refinement: Soft Matting and the Large-Kernel Linear Solver (Chapter 4)

### Soft Matting Markov Random Field (MRF) Formulation (Section 3.2.2 & 4.1.2)
* **Optimization Objective:**
  Refining transmission balances fidelity to coarse estimate $\tilde{t}$ and local smoothness guided by image $I$[cite: 420, 452]:
  $$E(t) = \lambda \| t - \tilde{t} \|_2^2 + t^T L t \quad[cite: 423, 453]$$
  Minimizing $E(t)$ with respect to vector $t \in \mathbb{R}^N$ leads to the symmetric, positive-definite linear system[cite: 423]:
  $$(L + \lambda U) t = \lambda \tilde{t} \quad[cite: 423]$$
  where $U$ is the $N \times N$ identity matrix, $\lambda = 10^{-4}$ is the regularization weight, and $L$ is the Matting Laplacian matrix[cite: 420, 423].

### Matting Laplacian Matrix Derivation
* **Local Linear Model:**
  Assumes that within a window $w_k$, transmission is an affine transform of image intensity: $t_i = a_k I_i + b_k$, $\forall i \in w_k$[cite: 452]. Minimizing $\sum_{i \in w_k} (t_i - a_k I_i - b_k)^2 + \epsilon a_k^2$ yields the linear regression parameters[cite: 452, 453]:
  $$a_k = \frac{\text{cov}_k(I, t)}{\sigma_k^2 + \frac{\epsilon}{|w_k|}}, \quad b_k = \bar{t}_k - a_k \mu_k \quad[cite: 453]$$
  Eliminating $(a_k, b_k)$ yields the elements of the Matting Laplacian matrix $L \in \mathbb{R}^{N \times N}$[cite: 453, 454]:
  $$L_{ij} = \sum_{k \mid (i, j) \in w_k} \left( \delta_{ij} - \frac{1}{|w_k|} \left( 1 + \frac{(I_i - \mu_k)(I_j - \mu_k)}{\sigma_k^2 + \frac{\epsilon}{|w_k|}} \right) \right) \quad[cite: 454]$$
  For color images, the scalar variance term is replaced by vector inner products with covariance inverse $(\Sigma_k + \frac{\epsilon}{|w_k|} U_3)^{-1}$[cite: 423].

### Fast $\mathcal{O}(N)$ Matrix-Vector Multiplication for Large-Kernel CG (Section 4.1.3 & Appendix B.1)
* **Algorithm:**
  In Conjugate Gradient (CG), the bottleneck is evaluating $q = L p$[cite: 455, 456]. Rather than explicitly building $L$ (which has $(4r+1)^2$ non-zero entries per row)[cite: 454, 455], $q = L p$ is computed exactly in three steps using sliding box filters[cite: 456, 457]:
  1. $a_k^* = \frac{\text{cov}_k(I, p)}{\sigma_k^2 + \frac{\epsilon}{|w|}}$[cite: 457]
  2. $b_k^* = \bar{p}_k - a_k^* \mu_k$[cite: 457]
  3. $q_i = |w| p_i - \left( \left(\sum_{k \in w_i} a_k^*\right) I_i + \sum_{k \in w_i} b_k^* \right)$[cite: 457]
* **Proof of Equivalence (Appendix B.1):**
  Differentiating $q_i$ with respect to an arbitrary input element $p_j$ recovers the exact Laplacian entry[cite: 457, 511, 512]:
  $$\frac{\partial q_i}{\partial p_j} = |w|\delta_{ij} - \frac{1}{|w|} \sum_{k \in w_i \cap w_j} \left( 1 + \frac{(I_j - \mu_k)(I_i - \mu_k)}{\sigma_k^2 + \frac{\epsilon}{|w|}} \right) = L_{ij} \quad[cite: 512]$$
  Using 1D cumulative sums $C_I(x) = \sum_{x'=0}^x I(x')$ along separable dimensions, 2D box filters are evaluated in $\mathcal{O}(1)$ time per pixel[cite: 512, 513]. Thus, $q = L p$ executes in $\mathcal{O}(N)$ time independent of filter radius $r$[cite: 458, 514].

### Spectral Analysis and Condition Number Proof (Section 4.1.3 & Appendix B.3)
* **Eigenvalue Derivations:**
  In flat image regions ($\sigma_k^2 \to 0$), the Laplacian matrix reduces to $L_{ij} = \sum_{k \mid (i, j) \in w_k} (\delta_{ij} - \frac{1}{|w|})$[cite: 458, 459, 514]. Its matrix-vector product is[cite: 514]:
  $$L p = |w| p - \frac{1}{|w|} B_y B_x B_y B_x p \quad[cite: 514]$$
  where $B_x, B_y$ are 1D box filters with Discrete-Time Fourier Transform (DTFT) eigenvalues[cite: 515]:
  $$\xi_{k_x} = \frac{\sin\left( (2r + 1) \frac{k_x \pi}{2 N_x} \right)}{\sin\left( \frac{k_x \pi}{2 N_x} \right)} \quad[cite: 515]$$
  The eigenvalues of $L$ are $\xi^L = |w| - \frac{1}{|w|} \xi_{k_x}^2 \xi_{k_y}^2$[cite: 515]:
  $$\xi_{\max}^L = |w| = (2r + 1)^2 \quad[cite: 459, 515]$$
  $$\xi_{\min}^L \approx \frac{1}{3} \left[ \left(\frac{\pi}{2 N_x}\right)^2 + \left(\frac{\pi}{2 N_y}\right)^2 \right] |w|^2 \approx \frac{\pi^2}{6} \frac{|w|^2}{N} \quad[cite: 459, 515]$$
* **Condition Number and Convergence Rate:**
  The condition number of the system matrix $A = L + \lambda U$ satisfies[cite: 458, 459]:
  $$\kappa = \frac{\xi_{\max}^A}{\xi_{\min}^A} \approx \frac{\xi_{\max}^L}{\xi_{\min}^L} = \mathcal{O}\left( \frac{N}{|w|} \right) = \mathcal{O}\left( \frac{N}{r^2} \right) \quad[cite: 459]$$
  Because CG iteration complexity is $n_{\text{iter}} \approx \frac{1}{2}\sqrt{\kappa} \ln(2/\epsilon_{\text{tol}})$[cite: 458], the number of iterations scales as[cite: 459]:
  $$n_{\text{iter}} = \mathcal{O}\left( \frac{\sqrt{N}}{r} \right) \quad[cite: 459]$$
  Combined with the $\mathcal{O}(N)$ multiplication step, the total time complexity drops from $\mathcal{O}(r N^{3/2})$ to $\mathcal{O}(\frac{1}{r} N^{3/2})$[cite: 456, 459].

## 5. The Guided Image Filter (Section 4.2)

### Filter Definition and Closed-Form Solution
* **Local Linear Model:**
  The output $q$ is assumed to be an affine function of guidance image $I$ in window $w_k$[cite: 468, 469]:
  $$q_i = a_k I_i + b_k \quad \forall i \in w_k \quad[cite: 468, 469]$$
* **Least-Squares Ridge Regression Objective:**
  Minimizing the squared error between $q$ and input $p$ (where $p = \tilde{t}$ in dehazing) with an $L_2$ penalty on slope $a_k$[cite: 469]:
  $$E(a_k, b_k) = \sum_{i \in w_k} \left( (a_k I_i + b_k - p_i)^2 + \epsilon a_k^2 \right) \quad[cite: 469]$$
  Setting partial derivatives to zero yields the regression coefficients[cite: 469]:
  $$a_k = \frac{\frac{1}{|w|} \sum_{i \in w_k} I_i p_i - \mu_k \bar{p}_k}{\sigma_k^2 + \epsilon} = \frac{\text{cov}_k(I, p)}{\sigma_k^2 + \epsilon}, \quad b_k = \bar{p}_k - a_k \mu_k \quad[cite: 469]$$
* **Symmetric Window Averaging:**
  Because a pixel $i$ is covered by multiple overlapping windows $w_k$, the final output averages all local predictions[cite: 469, 470]:
  $$q_i = \frac{1}{|w|} \sum_{k: i \in w_k} (a_k I_i + b_k) = \bar{a}_i I_i + \bar{b}_i \quad[cite: 470]$$
  where $\bar{a}_i = \frac{1}{|w|} \sum_{k \in w_i} a_k$ and $\bar{b}_i = \frac{1}{|w|} \sum_{k \in w_i} b_k$[cite: 470].

### Explicit Kernel Representation
* **Kernel Weight Equation:**
  Substituting $(a_k, b_k)$ into $q_i$ reveals that the guided filter is an explicit linear translation-variant filter: $q_i = \sum_j W_{ij}(I) p_j$ with kernel[cite: 468, 470, 471]:
  $$W_{ij}(I) = \frac{1}{|w|^2} \sum_{k: (i, j) \in w_k} \left( 1 + \frac{(I_i - \mu_k)(I_j - \mu_k)}{\sigma_k^2 + \epsilon} \right) \quad[cite: 471]$$
  The kernel naturally sums to unity ($\sum_j W_{ij}(I) = 1$) without requiring post-hoc normalization[cite: 471].

### Mathematical Duality with the Matting Laplacian
* **High-Pass / Low-Pass Duality:**
  Direct comparison between $W_{ij}(I)$ and $L_{ij}$ reveals the matrix identity[cite: 476, 477]:
  $$L_{ij} = |w|(\delta_{ij} - W_{ij}) \iff L = |w|(U - W) \quad[cite: 476, 477]$$
  Thus, the Matting Laplacian matrix is an exact high-pass filter: $L p \propto (p - W p)$[cite: 477].
* **One-Step Jacobi Solver Interpretation:**
  Rewriting the soft matting linear system $(L + \Lambda) t = \Lambda \tilde{t}$ using $L = |w|(U - W)$ and setting $\Lambda_{ii} = |w| W_{ii}$ yields the approximation[cite: 477, 478]:
  $$t \approx W \tilde{t} \quad[cite: 478]$$
  Hence, running the non-iterative guided filter once directly approximates solving the diffusion-based MRF system[cite: 478].

## 6. Mathematical Models in Human Visual Perception (Chapter 5)

### Perceptual Illumination Inversion
* **Visual Haze Reduction Model:**
  Psychophysical experiments testing lightness illusions across altered dark channels indicate that the human visual system (HVS) mitigates perceived haze via an internal inversion of Koschmieder's model[cite: 494, 495, 498, 500]:
  $$\check{I}_c(x) = \frac{I_c(x) - A_c}{t_{\text{perceived}}(x)} + A_c \quad[cite: 498]$$
* **Non-Linear Thresholding Operator:**
  Illusion experiments parameterize context changes via non-linear lower-bounding: $\check{I}_c(x) = \max(I_c(x), \theta)$[cite: 495, 499]. This non-linear clipping selectively lifts the image's dark channel without altering the target patch[cite: 495, 499]:
  $$\check{I}^{\text{dark}}(x) = \min_{x' \in \Omega} \min_c \max(I_c(x'), \theta) \ge \theta \quad[cite: 495]$$
  The corresponding shift in human brightness perception supports the hypothesis that the human visual system uses the dark channel to infer environmental transparency[cite: 495, 496, 500].
