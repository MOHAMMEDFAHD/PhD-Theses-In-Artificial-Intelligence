# Academic Summary: Scalable Feature Learning

## Abstract
This doctoral dissertation addresses the fundamental bottlenecks of manual feature engineering and computational scalability in machine learning and artificial intelligence[cite: 633, 634]. By framing unsupervised feature learning as a scalable alternative to human-engineered representations[cite: 633, 634], the research presents three primary innovations: (1) **Reconstruction Independent Component Analysis (RICA)**, an unconstrained optimization framework that replaces rigid orthonormality constraints with a soft reconstruction penalty to enable scalable, overcomplete feature learning robust to incomplete whitening[cite: 634, 667, 669, 670]; (2) **Tiled Convolutional Neural Networks (Tiled CNNs)**, an architecture that partially unties local receptive field weights to learn complex, non-translational invariances (such as scale and rotation) while retaining the parameter efficiency of convolutional designs[cite: 634, 635, 698, 699, 700]; and (3) a massively distributed deep learning framework leveraging **model parallelism** and **asynchronous stochastic gradient descent (Downpour SGD)** within the DistBelief infrastructure[cite: 635, 712, 717, 718]. Evaluated on a 9-layer network comprising 1.15 billion parameters trained on 10 million unlabeled YouTube video frames[cite: 650, 712, 715], the system demonstrated the emergent development of high-level category detectors (e.g., human faces, human bodies, and cat faces) from completely unlabeled data[cite: 651, 712, 720, 723, 725], achieving state-of-the-art performance on ImageNet classification (a 70% relative improvement on 22,000 categories)[cite: 651, 726, 727, 728].

---

## Problems
- **Reliance on Manual Feature Engineering:** Traditional machine learning algorithms depend on human-crafted domain-specific descriptors (e.g., SIFT, HOG, SURF for vision; MFCC for speech; BM25, TF-IDF for NLP)[cite: 633, 649, 650]. Designing such representations is time-consuming, expensive, brittle, and difficult to transfer across modalities[cite: 633, 650, 678].
- **Degeneracy and Scalability Bottlenecks in Independent Component Analysis (ICA):** Standard ICA formulations require a hard orthonormality constraint ($W W^T = I$) to prevent degenerate basis representations[cite: 634, 651, 668, 669]. When scaling to overcomplete regimes ($k > n$), this constraint cannot be satisfied algebraically[cite: 667, 669]. Approximate projection routines or score-matching alternatives are slow, complex, and computationally prohibitive[cite: 667, 669].
- **Sensitivity of Classical ICA to Whitening:** Traditional ICA requires strict data sphering/whitening via full-covariance eigenvalue decompositions (e.g., PCA)[cite: 634, 651, 667, 669]. Computing exact whitening on high-dimensional data (e.g., $200 \times 200$ images yielding $40,000 \times 40,000$ matrices) is computationally intractable[cite: 668, 674]; training ICA on raw or approximately whitened data yields noisy, degraded filters[cite: 668, 674].
- **Architectural Rigidity of Convolutional Neural Networks (CNNs):** Standard CNNs enforce complete spatial weight-tying to guarantee translation invariance and reduce parameters[cite: 634, 635, 652, 697, 698]. However, pooling over identical shifted weights precludes the model from learning more complex affine, rotational, scaling, or out-of-plane transformations from data[cite: 635, 652, 653, 698, 699].
- **Computational Limits of Deep Learning on Massive Unlabeled Datasets:** Standard deep neural networks are fully connected, causing parameter counts to scale quadratically with input size[cite: 634, 697]. Furthermore, synchronous distributed optimization bottlenecks training on clusters because parameter updates are bounded by the slowest computing node[cite: 635, 718].

---

## Proposed Solutions
- **Reconstruction Independent Component Analysis (RICA):** Swaps the hard constraint $W W^T = I$ with a smooth, linear reconstruction penalty $\frac{\lambda}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2$[cite: 634, 667, 669, 670]. This removes the requirement for specialized constrained optimizers, enables standard quasi-Newton solvers (e.g., L-BFGS, CG) to optimize overcomplete bases, and establishes robustness against imperfect whitening[cite: 634, 651, 667, 669, 670, 674].
- **$L_2$-Norm Ball Projection for Overcomplete Features:** Formulates an unconstrained projected gradient step where gradients are analytically backpropagated through an $L_2$-sphere normalization step ($\hat{W}_i = W_i / \|W_i\|_2$), preventing inactive basis vectors from collapsing to zero[cite: 673, 735, 736].
- **Reconstruction Independent Subspace Analysis (RISA):** Extends RICA to spatial and spatio-temporal video blocks by grouping simple quadratic cell outputs into disjoint subspaces via an $L_2$ square-root pooling layer, yielding phase- and translation-invariant complex cells that retain frequency and velocity tuning[cite: 634, 678, 679, 682, 683, 684].
- **Tiled Convolutional Neural Networks (Tiled CNNs):** Introduces a tiling hyperparameter $k$, enforcing weight-tying only between units spaced $k$ steps apart[cite: 634, 635, 652, 653, 698, 700]. This bridges fully tied CNNs ($k=1$) and untied local receptive field networks ($k = n - s + 1$), enabling pooling units to combine heterogeneous basis functions and learn non-translational transformations[cite: 698, 700, 703].
- **Model-Parallel and Asynchronous Data-Parallel Distributed Training:** Implements local receptive fields without parameter sharing to divide the model across 169 machines (model parallelism) within the DistBelief framework[cite: 635, 712, 716, 717]. Combines this with asynchronous stochastic gradient descent (Downpour SGD) across replicated model instances communicating with distributed parameter servers to scale to 1,000 machines (16,000 cores)[cite: 635, 650, 712, 717, 718].

---

## Purpose
The primary purpose of this research is to automate the extraction of feature representations across diverse modalities (static images, video, MRI)[cite: 634] by scaling unsupervised learning algorithms from small, synthetic patches to massive real-world data[cite: 633, 634, 650, 711]. Specifically, the dissertation explores whether high-level, semantic, class-selective representations (akin to cortical "grandmother cells") can emerge purely from large-scale unlabeled inputs without explicit human supervision or category labels[cite: 653, 710, 711, 712].

## Methodology

### 1. Mathematical Formulations

#### Reconstruction ICA (RICA) Objective
For an unlabeled dataset $\{x^{(i)}\}_{i=1}^m \subset \mathbb{R}^n$ with zero mean, RICA solves the unconstrained objective[cite: 668, 670, 671]:
$$\min_{W} \frac{\lambda}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 + \sum_{i=1}^m \sum_{j=1}^k g(W_j x^{(i)}) \quad[cite: 670]$$
where $W \in \mathbb{R}^{k \times n}$ is the weight matrix, $W_j$ is the $j$-th row feature[cite: 668], $\lambda > 0$ balances reconstruction fidelity against sparsity[cite: 670, 716], and $g(z) = \log(\cosh(z))$ or $\sqrt{z^2 + \epsilon}$ is a smooth convex $L_1$ penalty[cite: 668, 670, 716].

#### Equivalence Proofs: Orthonormality, Reconstruction, and Autoencoders
- **Lemma 3.3.1 (Whitening Equivalence):** Under whitened data $\mathbb{E}[x x^T] = \frac{1}{m} \sum_{i=1}^m x^{(i)} (x^{(i)})^T = I$, the empirical reconstruction error is algebraically equivalent to column orthonormality[cite: 732]:
  $$\frac{1}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 = \text{tr}\left( (W^T W - I)^T (W^T W - I) \frac{1}{m} \sum_{i=1}^m x^{(i)} (x^{(i)})^T \right) = \| W^T W - I \|_{\mathcal{F}}^2 \quad[cite: 732]$$
  For unwhitened data with covariance $C_x = E D E^T$, the error equates to weighted linear regression[cite: 732]:
  $$\frac{1}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 = \| (W^T W - I) E D^{1/2} \|_{\mathcal{F}}^2 \quad[cite: 732]$$
- **Lemma 3.3.2 (Row vs. Column Orthonormality):** Via properties of the trace operator, column and row penalties differ solely by a dimensional constant[cite: 733]:
  $$\| W W^T - I_k \|_{\mathcal{F}}^2 = \| W^T W - I_n \|_{\mathcal{F}}^2 + k - n \quad[cite: 733]$$
- **Lemma 3.3.3 (Denoising Equivalence):** Adding isotropic Gaussian noise $\epsilon^{(i)} \sim \mathcal{N}(0, \delta^2 I)$ to inputs in RICA is mathematically equivalent to optimizing uncorrupted RICA with an $L_2$ weight-decay penalty $2 \delta^2 \sum_{j=1}^k \| W_j \|_2^2$[cite: 733, 734].

#### Reconstruction Independent Subspace Analysis (RISA)
Extends RICA to group simple features into multi-dimensional invariant subspaces[cite: 678, 682, 683]:
$$p_i(x^t; W, V) = \sqrt{\sum_{k=1}^m V_{ik} \left( \sum_{j=1}^n W_{kj} x_j^t \right)^2} \quad[cite: 682]$$
$$\min_{W} \sum_{t=1}^T \sum_{i=1}^m p_i(x^t; W, V) + \lambda \| W^T W x^t - x^t \|_2^2 \quad[cite: 682]$$
where $V \in \mathbb{R}^{m \times k}$ is a fixed block-diagonal pooling matrix summing over non-overlapping adjacent simple-cell pairs[cite: 682, 683].

#### Tiled Convolutional Layer Dynamics
Within a 1D/2D feature map, units separated by $k$ spatial steps share parameters[cite: 698, 700]:
$$W^{(x)} = W^{(x + k)} \quad[cite: 700]$$
The model employs local receptive fields of size $s$, tiled with step size $1$, followed by $L_2$ spatial pooling of size $p$ across adjacent untied simple units[cite: 698, 700, 703, 704]. Global reconstruction over the complete sparse matrix prevents redundant filter copying across adjacent receptive fields[cite: 667, 702, 703].

### 2. Large-Scale System Architecture and Distributed Infrastructure
- **Dataset Generation:** 10 million random $200 \times 200$ pixel color frames extracted from 10 million distinct YouTube videos[cite: 651, 653, 712]. Images undergo local contrast normalization (LCN) to simulate retinal on-center/off-surround receptive fields[cite: 712, 713].
- **Deep Hierarchical Architecture:** A 9-layer feedforward network constructed by cascading three identical stages of[cite: 714, 715]:
  1. *Simple Filtering Layer:* $18 \times 18$ local receptive fields, 8 feature maps, stride 1, untied weights[cite: 714, 715].
  2. *Complex Pooling Layer:* $5 \times 5$ overlapping $L_2$ pooling neighborhoods[cite: 714, 715].
  3. *Local Contrast Normalization (LCN):* $5 \times 5$ Gaussian subtractive and divisive normalization filters[cite: 714, 715].
- **Distributed Computing Scheme (DistBelief):**
  - *Model Parallelism:* The 1.15-billion parameter network is partitioned across 169 machines (each housing 16 CPU cores)[cite: 650, 717]. Sparse local receptive fields restrict inter-machine network communication to boundary activations[cite: 635, 712, 717].
  - *Data Parallelism:* 5 model replicas process distinct data shards in parallel using minibatches of 100 images[cite: 717, 718].
  - *Asynchronous Downpour SGD:* Replicas asynchronously pull current weights $W$ from 256 partitioned parameter servers every $P$ steps and push parameter gradients $\Delta W$ every $G$ steps without synchronization locks[cite: 717, 718]. Training executed continuously over 3 days across 1,000 machines (16,000 CPU cores)[cite: 650, 653, 712, 718].

### Key Experimental Insights
- **Invariance Properties of Emergent Neurons:** Numerical optimization over input space ($\arg\max_x f(x)$ subject to $\|x\|_2=1$) and empirical activations reveal that the learned face neuron remains active under horizontal/vertical spatial shifts ($\pm 10$ pixels), 2D scale changes (0.8 to $1.2\times$), and 3D out-of-plane head rotations (up to 30 degrees)[cite: 720, 721, 722, 723].
- **Ablation on Face Stimuli Presence:** Training the identical architecture on YouTube frames filtered to exclude human faces dropped face detection accuracy from 81.7% down to 72.5%, demonstrating that feature selectivity is directly driven by data distribution statistics rather than structural bias alone[cite: 723].
- **Scaling Law Dynamics:** Classification accuracy on ImageNet categories scaled monotonically as the total parameter count increased from 0.5 billion to 2.0 billion connections, confirming that capacity scaling is essential for complex conceptual extraction[cite: 727, 728].

## Conclusions
- **Unification of Unsupervised Learning Principles:** Soft reconstruction penalties bridge the gap between classical Independent Component Analysis, sparse coding, and autoencoders[cite: 634, 651, 669, 670, 671]. Replacing hard matrix inversions with smooth reconstruction yields massive computational speedups ($30\times$ to $50\times$) and robust convergence on unwhitened, overcomplete inputs[cite: 634, 651, 667, 675, 676].
- **Superiority of Learned Invariances:** Tiled CNNs demonstrate that hardcoding translation invariance via uniform weight sharing unnecessarily restricts model capacity[cite: 634, 635, 698, 699, 700]. Partial weight untying allows neural architectures to autonomously discover rotational, scaling, and out-of-plane invariances while controlling sample complexity[cite: 635, 652, 653, 698, 708, 709].
- **Unsupervised Emergence of High-Level Concepts:** Extreme-scale deep learning over abundant unlabeled video frames successfully produces specialized, category-selective neurons for complex natural concepts (faces, bodies, animals) without requiring human supervision, labels, or aligned bounding boxes[cite: 650, 651, 653, 710, 712, 720, 728].
- **Distributed Computing Paradigm:** Model parallelism enabled by local receptive fields combined with asynchronous, lock-free parameter updates (Downpour SGD) establishes a viable technical path for scaling neural networks to billions of parameters across commodity CPU clusters[cite: 635, 712, 716, 717, 718, 729].

## 1. Classical Independent Component Analysis (ICA) vs. Reconstruction ICA (RICA)

### Classical Constrained ICA Formulation (Section 3.2)
* **Mathematical Formulation:**
  Given centered observations $\{x^{(i)}\}_{i=1}^m \subset \mathbb{R}^n$, classical ICA estimates a linear transformation $W \in \mathbb{R}^{k \times n}$ by minimizing a non-linear contrast function subject to strict row orthonormality:
  $$\min_W \sum_{i=1}^m \sum_{j=1}^k g(W_j x^{(i)}) \quad \text{subject to} \quad W W^T = I_k \quad$$
  where $W_j$ denotes the $j$-th row of $W$, and $g(\cdot)$ is a smooth, convex surrogate for the $L_1$ norm (e.g., $g(z) = \log(\cosh(z))$ or $\sqrt{z^2 + \epsilon}$).
* **Mathematical Failure in Overcomplete Settings:**
  When learning an overcomplete representation where the number of features exceeds the data dimensionality ($k > n$), the matrix $W \in \mathbb{R}^{k \times n}$ has rank at most $\text{rank}(W) \le \min(k, n) = n < k$. Consequently, the symmetric matrix $W W^T \in \mathbb{R}^{k \times k}$ has rank at most $n$ and cannot equal the full-rank identity matrix $I_k$. Thus, standard symmetric orthogonalization $W \leftarrow (W W^T)^{-1/2} W$ breaks down.

---

### Reconstruction ICA (RICA) Unconstrained Objective (Section 3.2)
* **Mathematical Formulation:**
  RICA replaces the hard algebraic constraint with a continuous quadratic reconstruction penalty:
  $$\min_W \frac{\lambda}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 + \sum_{i=1}^m \sum_{j=1}^k g(W_j x^{(i)}) \quad$$
  where $\lambda > 0$ controls the tradeoff between data reconstruction and latent sparsity.
* **Explanation & Role in the Thesis:**
  $W x^{(i)}$ acts as a linear encoder and $W^T (W x^{(i)})$ acts as a linear decoder with tied weights. Because the objective function is unconstrained and continuously differentiable, it can be minimized using unconstrained quasi-Newton solvers (such as L-BFGS or conjugate gradient), yielding a $30\times$ to $50\times$ training speedup over overcomplete score-matching ICA.

### Lemma 3.3.1: Equivalence of Reconstruction Cost and Column Orthonormality
* **Statement & Proof (Appendix A.1):**
  Let the data covariance be $C_x = \frac{1}{m} \sum_{i=1}^m x^{(i)} (x^{(i)})^T$. Expanding the squared Frobenius norm via the matrix trace operator:
  $$\frac{1}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 = \frac{1}{m} \sum_{i=1}^m \text{tr}\left( (W^T W x^{(i)} - x^{(i)})^T (W^T W x^{(i)} - x^{(i)}) \right) \quad$$
  Using the cyclic permutation property of the trace ($\text{tr}(u^T v) = \text{tr}(v u^T)$):
  $$= \text{tr}\left( (W^T W - I)^T (W^T W - I) \left( \frac{1}{m} \sum_{i=1}^m x^{(i)} (x^{(i)})^T \right) \right) = \text{tr}\left( (W^T W - I)^T (W^T W - I) C_x \right) \quad$$
  When the data is strictly whitened ($C_x = I_n$), this simplifies directly to:
  $$\text{tr}\left( (W^T W - I)^T (W^T W - I) \right) = \| W^T W - I_n \|_{\mathcal{F}}^2 \quad$$

* **Generalization to Unwhitened Data:**
  If the input data is unwhitened with spectral decomposition $C_x = E D E^T$ (where $E$ is the orthogonal eigenvector matrix and $D$ is the diagonal eigenvalue matrix):
  $$\frac{1}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 = \text{tr}\left( (W^T W - I) E D E^T (W^T W - I)^T \right) = \| (W^T W - I) E D^{1/2} \|_{\mathcal{F}}^2 \quad$$
* **Mathematical Insight:**
  For unwhitened data, the reconstruction cost acts as a weighted regression penalized along principal axes by $D^{1/2}$. Because PCA whitening applies $D^{-1/2} E^T$, the optimization encourages $W$ to internally cancel $E D^{1/2}$, explaining why RICA learns clean, non-degenerate Gabor filters even without explicit data whitening.

---

### Lemma 3.3.2: Equivalence of Row and Column Orthonormality Under the Frobenius Norm
* **Statement & Proof (Appendix A.1):**
  Expanding the squared Frobenius norm of the row penalty for $W \in \mathbb{R}^{k \times n}$:
  $$\| W W^T - I_k \|_{\mathcal{F}}^2 = \text{tr}\left( (W W^T - I_k)^T (W W^T - I_k) \right) = \text{tr}(W W^T W W^T) - 2\,\text{tr}(W W^T) + \text{tr}(I_k) \quad$$
  Using the cyclic property $\text{tr}(A B) = \text{tr}(B A)$ for rectangular factors ($A = W$, $B = W^T W W^T$), and noting that $\text{tr}(I_k) = k$ and $\text{tr}(I_n) = n$:
  $$\text{tr}(W W^T W W^T) = \text{tr}(W^T W W^T W), \quad \text{tr}(W W^T) = \text{tr}(W^T W) \quad$$
  Substituting these identities into the expansion:
  $$= \text{tr}(W^T W W^T W) - 2\,\text{tr}(W^T W) + \text{tr}(I_n) + (k - n) = \| W^T W - I_n \|_{\mathcal{F}}^2 + (k - n) \quad$$
* **Mathematical Insight:**
  Minimizing the column orthonormality penalty $\|W^T W - I_n\|_{\mathcal{F}}^2$ is algebraically equivalent to minimizing the row orthonormality penalty $\|W W^T - I_k\|_{\mathcal{F}}^2$ up to an additive constant $(k - n)$. As $\lambda \to \infty$ for complete/undercomplete data, RICA's soft reconstruction penalty approaches the hard constraint of classical ICA ($W W^T = I_k$).

---

### Lemma 3.3.3: Denoising Autoencoders Equate to RICA with $L_2$ Weight Decay
* **Statement & Proof (Appendix A.1):**
  Corrupt the input vector with additive zero-mean isotropic noise $\epsilon^{(i)} \sim \mathcal{N}(0, \delta^2 I_n)$, independent of $x^{(i)}$ (such that $\mathbb{E}[x \epsilon^T] = 0$). The expected denoising reconstruction loss is:
  $$\frac{1}{m} \sum_{i=1}^m \| W^T W (x^{(i)} + \epsilon^{(i)}) - x^{(i)} \|_2^2 = \frac{1}{m} \sum_{i=1}^m \| (W^T W - I) x^{(i)} + W^T W \epsilon^{(i)} \|_2^2 \quad$$
  Expanding the cross-terms and taking the expectation over $\epsilon$:
  $$= \text{tr}\left( (W^T W - I)^T (W^T W - I) C_x \right) + \text{tr}\left( W^T W W^T W (\delta^2 I_n) \right) \quad$$
  Rewriting $W^T W W^T W = (W^T W - I)^T (W^T W - I) + 2 W^T W - I_n$:
  $$= \text{tr}\left( (W^T W - I)^T (W^T W - I) (C_x + \delta^2 I_n) \right) + 2 \delta^2 \text{tr}(W^T W) - \delta^2 \text{tr}(I_n) \quad$$
  Using $C_x = E D E^T$ and expressing the trace of $W^T W$ as the sum of squared row norms $\text{tr}(W^T W) = \sum_{j=1}^k \|W_j\|_2^2$:
  $$= \| (W^T W - I) E (D + \delta^2 I)^{1/2} \|_{\mathcal{F}}^2 + 2 \delta^2 \sum_{j=1}^k \| W_j \|_2^2 + \text{constant} \quad$$
* **Mathematical Insight:**
  When inputs are whitened ($E = I$, $D = I$), training a denoising autoencoder is mathematically equivalent to training RICA with an explicit Frobenius weight decay penalty $2 \delta^2 \sum_{j=1}^k \|W_j\|_2^2$.

## 3. Optimization via Analytic Backpropagation on the $L_2$-Norm Sphere (Appendix A.2)

### Degeneracy in Overcomplete RICA
When $k \gg n$, the reconstruction error $\sum_{i=1}^m \|W^T W x^{(i)} - x^{(i)}\|_2^2$ can reach zero using only a complete subset of $n$ rows, leaving the remaining $k - n$ rows at zero to trivialize the sparsity loss $\sum g(W_j x)$. To prevent this degeneracy, every row must lie on the unit $L_2$-sphere:
$$\| W_i \|_2^2 = 1, \quad \forall i \in \{1, \dots, k\} \quad$$

### Reparameterization and Projected Gradient Derivations
Rather than relying on constrained optimization, RICA evaluates the loss on normalized weights $\hat{W}$, allowing unconstrained gradient descent:

1. **Forward Normalization Projection:**
   $$\hat{W}_{ij} = \frac{W_{ij}}{\sqrt{\epsilon + \sum_{l=1}^n W_{il}^2}} \quad$$
2. **Intermediate Loss Gradient:**
   Compute the gradient with respect to the projected weight: $G_{ij} = \frac{\partial \mathcal{L}}{\partial \hat{W}_{ij}}$.
3. **Analytic Inverse Projection Gradient (Chain Rule):**
   By differentiating $\hat{W}_{ij}$ with respect to the unconstrained parameter $W_{ij}$:
   $$\frac{\partial \hat{W}_{il}}{\partial W_{ij}} = \frac{\delta_{lj}}{\sqrt{\epsilon + \sum_p W_{ip}^2}} - \frac{W_{il} W_{ij}}{\left( \epsilon + \sum_p W_{ip}^2 \right)^{3/2}}$$
   Applying the chain rule $\frac{\partial \mathcal{L}}{\partial W_{ij}} = \sum_{l=1}^n \frac{\partial \mathcal{L}}{\partial \hat{W}_{il}} \frac{\partial \hat{W}_{il}}{\partial W_{ij}}$ yields the modified update gradient:
   $$g_{ij} = \frac{G_{ij}}{\sqrt{\epsilon + \sum_{l=1}^n W_{il}^2}} - \hat{W}_{ij} \frac{\sum_{l=1}^n G_{il} W_{il}}{\epsilon + \sum_{l=1}^n W_{il}^2} \quad$$
* **Mathematical Insight:**
  Subtracting the inner product term $\hat{W}_{ij} (\sum_l G_{il} W_{il})$ projects the gradient onto the tangent plane of the unit sphere, preventing the unconstrained optimizer from diverging radially.

## 4. Subspace Invariance: Reconstruction Independent Subspace Analysis (RISA) (Chapter 4)

### Two-Layer Subspace Transformation
RISA models visual cortex complex cells by pairing quadratic simple units with square-root pooling units:
* **Simple Unit Quadratic Response:**
  $$s_k(x) = \left( \sum_{j=1}^n W_{kj} x_j \right)^2 \quad$$
* **Complex Unit $L_2$ Pooling Response:**
  $$p_i(x; W, V) = \sqrt{\sum_{k=1}^m V_{ik} s_k(x)} = \sqrt{\sum_{k=1}^m V_{ik} \left( \sum_{j=1}^n W_{kj} x_j \right)^2} \quad$$
  where $V \in \{0, 1\}^{m \times k}$ is a binary block-diagonal operator pooling adjacent linear features.

### Optimization Objective
$$\min_W \sum_{t=1}^T \sum_{i=1}^m p_i(x^t; W, V) + \lambda \| W W^T x^t - x^t \|_2^2 \quad$$
* **Mathematical Role:**
  Minimizing the sum of pooled square roots enforces joint sparsity across the subspace defined by $V_i$. As a result, the learned simple units within each subspace span an invariant manifold (orthogonal phase variations of the same spatial frequency and orientation). This makes $p_i(x)$ robust to phase/translation while retaining sharp selectivity for frequency, orientation, and motion velocity.

## 5. Tiled Convolutional Neural Networks (Tiled CNNs) (Chapter 5)


```

```
   CNN vs. Tiled CNN Parameter Tying Schemes (Tile Size k)

```

Standard CNN (k=1):     [W_1]  [W_1]  [W_1]  [W_1]  [W_1]   (Identical filters)
Tiled CNN (e.g., k=2):  [W_1]  [W_2]  [W_1]  [W_2]  [W_1]   (Filters cycle mod k)
Fully Untied (k=p):     [W_1]  [W_2]  [W_3]  [W_4]  [W_5]   (Heterogeneous filters)

```

### Algebraic Tiling Parameterization
Let $x$ index the spatial location of a local receptive field of size $s$. The weights $W^{(x)}$ are constrained to repeat with period $k$:
$$W^{(x)} = W^{(x + k)} \quad$$
* **Case $k = 1$:** Fully tied standard convolutional network ($W^{(x)} = W^{(1)}$ for all $x$).
* **Case $k = p$ (where $p$ is the pooling size):** Units pooled together into the same complex cell have untied parameters, while the pattern repeats every $k$ units across the visual field.
* **Case $k = n - s + 1$:** Fully untied local receptive field network.

### Global vs. Local Orthogonality in Tiled Topographic ICA
The unsupervised pre-training objective minimizes spatial pooling sparsity over $l$ feature maps:
$$\min_W \sum_{i=1}^m \sum_{j=1}^K \sqrt{\epsilon + H_j (W x^{(i)})^2} \quad$$
* **Failure of Local Orthogonalization:**
  Restricting the constraint $W_{\text{loc}} W_{\text{loc}}^T = I$ to units at the same spatial location allows adjacent unconstrained units to learn identical features, leading to redundant filters.
* **Global Reconstruction Solution:**
  Replacing local constraints with a global reconstruction loss over the entire sparse tiled matrix $W$:
  $$\mathcal{L}_{\text{recon}} = \frac{\lambda}{m} \sum_{i=1}^m \| W^T W x^{(i)} - x^{(i)} \|_2^2 \quad$$
  forces receptive fields across different spatial locations and maps to reconstruct the global image jointly, preventing duplicate features and producing an even distribution of orientations and frequencies.

## 6. Mathematical Operations of the 9-Layer Distributed Model (Chapter 6)

### Subtractive and Divisive Local Contrast Normalization (LCN)
Each stage in the deep network applies a three-step transformation: linear filtering, $L_2$ pooling, and local contrast normalization. Let $h_{i,j,k}$ denote the activation at map $i$ and spatial coordinates $(j, k)$:

1. **Subtractive Normalization (Local Mean Zeroing):**
   $$g_{i,j,k} = h_{i,j,k} - \sum_{u=-P}^P \sum_{v=-P}^P G_{uv} \, h_{i,\, j+u,\, k+v} \quad$$
   where $G_{uv}$ is a discrete 2D normalized Gaussian kernel of size $5 \times 5$ ($\sum_{uv} G_{uv} = 1$).
2. **Divisive Normalization (Local Variance Scaling):**
   $$\sigma_{i,j,k} = \sqrt{\sum_{u=-P}^P \sum_{v=-P}^P G_{uv} \, g_{i,\, j+u,\, k+v}^2} \quad$$
   $$y_{i,j,k} = \frac{g_{i,j,k}}{\max(c, \, \sigma_{i,j,k})} \quad$$
   where $c = 0.01$ is a small regularizer preventing division by zero in low-contrast, uniform regions.

## 7. Numerical Optimization for Feature Visualization (Section 6.4.4)

### Bounded Optimal Stimulus Reconstruction
To inspect the visual pattern that maximally excites a trained high-level neuron without relying on empirical sample retrieval, the thesis solves an optimization problem over input image space:
$$x^* = \arg\max_x f(x; W, H) \quad \text{subject to} \quad \| x \|_2^2 = 1 \quad$$
where $f(x; W, H)$ denotes the feedforward activation of the targeted neuron given input $x$ and fixed network weights $W, H$.

* **Optimization Procedure:**
  Solved via projected gradient ascent with backtracking line search:
  $$x^{(t+1)} = \mathcal{P}_{\mathbb{S}^{n-1}} \left( x^{(t)} + \alpha_t \nabla_x f(x^{(t)}; W, H) \right) = \frac{x^{(t)} + \alpha_t \nabla_x f(x^{(t)}; W, H)}{\| x^{(t)} + \alpha_t \nabla_x f(x^{(t)}; W, H) \|_2} \quad$$
  Using multiple random initializations on the unit sphere avoids local extrema. Applying this procedure to the top-layer neuron produced a smooth human face reconstruction, demonstrating that category selectivity emerged in the network without supervision.

## 8. Asynchronous Distributed Stochastic Optimization (Section 6.3.3)

### Downpour Asynchronous Stochastic Gradient Descent
The thesis scales training across 16,000 CPU cores using an asynchronous parameter-server architecture:

Let $W \in \mathbb{R}^D$ be the global parameter vector partitioned across $S = 256$ independent parameter servers. Replicas operate in parallel without synchronization barriers:
* **Pull Step:** Model replica $r$ fetches parameters $W$ from the parameter servers every $P$ iterations:
  $$W_{\text{local}}^{(r)} \leftarrow W \quad$$
* **Gradient Calculation:** Replica $r$ evaluates the gradient over a local minibatch $B$ of size $|B| = 100$:
  $$\Delta W^{(r)} = \frac{1}{|B|} \sum_{x \in B} \nabla_W \mathcal{L}(x; W_{\text{local}}^{(r)}) \quad$$
* **Push Step:** Replica $r$ sends updates $\Delta W^{(r)}$ back every $G$ iterations. Parameter servers update the master weights using their local learning rate $\alpha$:
  $$W \leftarrow W - \alpha \Delta W^{(r)} \quad$$

* **Statistical Properties:**
  Because parameter reads and writes occur asynchronously without global locks, gradients $\Delta W^{(r)}$ are computed over stale parameter snapshots:
  $$\Delta W^{(r)} = \nabla_W \mathcal{L}(W_{t - \tau})$$
  where the delay $\tau \ge 0$ reflects system communication latency. Despite this staleness, asynchronous updates act as an implicit regularizer, making the distributed system robust to straggler nodes and communication bottlenecks on large computer clusters.
