# Academic Summary: Data as Foundation: Designing Systematic Curation for an Evolving Foundation Model Landscape

## Abstract
Modern foundation models rely heavily on large-scale web data, yet standard data curation practices remain largely heuristic, non-transparent, and heavily coupled with specific model setups. This dissertation formalizes data curation as a principled, systematic scientific discipline for foundation model pretraining across vision-language and language modalities. Spanning four core contributions, the work investigates:
1. The dynamics of combining diverse web corpora, demonstrating that naive aggregation dilutes model robustness.
2. The systemic English bias in standard curation pipelines, proving that translating and re-filtering non-English web crawls yields substantial gains across standard benchmarks and geographically diverse domains.
3. The use of generative image captioning to salvage discarded image-text pairs, overcoming the diversity loss inherent to filtering.
4. The recycling of low-quality web texts via guided language model rewriting to alleviate the impending pretraining "data wall."

## Problems
- **The Quantity Fallacy and Lack of Systematic Pretraining Data Science:** Pretraining practices have prioritized scale over composition. The empirical and theoretical effects of mixing heterogeneous web-scale datasets on out-of-distribution (OOD) robustness and generalization remain under-characterized.
- **Exclusionary Filtering and English-Centric Bias:** Standard filtering heuristics (e.g., scoring text-image alignment using English-centric vision-language models like CLIP) disproportionately reject non-English and culturally diverse samples. This practice impoverishes geographic representation and limits downstream generalization.
- **The Noise-Diversity Trade-off in Filtering:** Conventional quality filters discard up to 60%–90% of web image-text pairs due to weak alt-texts. While noise is reduced, critical visual and conceptual diversity is discarded.
- **Impending Pretraining "Data Wall":** In natural language pretraining, compute resources outpace the generation of high-quality human text. Common filtering pipelines discard up to 99% of raw web data, exhausting the usable web text pool and leading to severe token scarcity.

## Proposed Solutions
- **Systematic Testbeds and Theoretical Foundations of Data Mixing:** Establishment of a 6-dataset pretraining benchmark (YFCC, LAION, Conceptual Captions, WIT, RedCaps, Shutterstock) to isolate dataset effects, paired with an analytical framework modeling data mixing and filtering via high-dimensional linear models and probit-transformed accuracies.
- **Multilingual Data Inclusion via Translation and Re-Filtering:** A pipeline translating raw, unfiltered multilingual web image alt-texts into English using an advanced machine translation model (NLLB) prior to score-based filtering, thereby surfacing culturally and visually diverse images.
- **Generative Recaptioning:** Using vision-language models (BLIP-2, OpenCLIP-CoCa) to generate synthetic captions for web images that would otherwise be discarded, subsequently filtering and mixing raw and synthetic captions.
- **ReWire (REcycling the Web with guIded REwrite):** A framework prompting a large language model (Llama-3.3-70B-Instruct) with metacognitive chain-of-thought instructions to identify the core purpose of discarded, moderate-quality web documents and synthesize high-quality, comprehensive pretraining documents, followed by custom fastText quality filtering.

## Purpose
The primary purpose is to transition data curation from an ad-hoc engineering step into an independent, rigorous, and inclusive scientific discipline. The thesis demonstrates that data diversity—spanning linguistic, geographic, and conceptual dimensions—is a primary determinant of pretraining quality, and that the pretraining scaling boundary can be extended by recovering, synthesizing, and recycling discarded web data rather than merely pruning it.

## Methodology

### 1. Data Mixing and Robustness Analysis
- **Empirical Setup:** CLIP models (ResNet-50 backbones) were pretrained from scratch on subsets (up to 15M samples) of six public web corpora (YFCC, LAION, CC-12M, RedCaps, Shutterstock, WIT).
- **Evaluations:** Effective robustness was measured across natural distribution shifts derived from ImageNet: ImageNet-V2, ImageNet-R, ImageNet-Sketch, and ObjectNet.
- **Output vs. Input Mixing:** Evaluated whether ensembling outputs of models trained on individual sources could predict the linear accuracy trends obtained from input mixing.
- **Theoretical Modeling:** Formalized under high-dimensional binary classification settings where data from source $k$ satisfies:
  $$x_i = y_i \theta_k + \left(\frac{\|\theta_k\|}{\rho_k}\right) z_i, \quad z_i \sim \mathcal{N}(0, I_d)$$
  Accuracies across test distributions $P_{\theta_1, \rho_1}$ and $P_{\theta_2, \rho_2}$ were evaluated using the probit transform $\Phi^{-1}(\text{Acc})$.

### 2. Overcoming Exclusion via Translation
- **Candidate Pool:** DataComp medium pool (128M raw image-text pairs).
- **Translation:** Captions were translated into English via the No Language Left Behind (NLLB-200, 600M) model.
- **Filtering & Training:** Embeddings were extracted using OpenAI CLIP-ViT-L/14 and Data Filtering Networks (DFN). CLIP models (ViT-B/32) were trained under fixed compute budgets (128M and 1.28B sample passes).
- **Evaluations:** Evaluated on the 38 DataComp tasks, ImageNet shifts, Flickr30K/MS-COCO retrieval, and GeoDE (geographically diverse object recognition).

### 3. Multimodal Synthetic Recaptioning
- **Models & Prompts:** Evaluated BLIP, BLIP-2, and OpenCLIP-CoCa (base vs. MS-COCO fine-tuned) with top-$k$ sampling ($k=50$) at varying temperatures ($T=0.75$).
- **Pipelines:** Retained high-scoring raw image-caption pairs and replaced low-scoring pairs with filtered synthetic captions.
- **Scales:** Evaluated across DataComp small (12.8M), medium (128M), and large (1.28B) pools.

### 4. ReWire Text Recycling Pipeline
- **Corpus:** DCLM-RefinedWeb (heuristic-filtered Common Crawl).
- **Pipeline:** Moderate-quality documents (normally rejected by DCLM fastText filtering) were processed by Llama-3.3-70B-Instruct with guided rewriting.
- **Filtering & Training:** Rewritten texts were filtered using a custom fastText classifier (trained on OpenHermes 2.5/ELI5 vs. random rewrites). Llama-2 architectures (1B, 3B, 7B parameters) were trained via the Lingua framework up to Chinchilla-optimal token budgets and multi-epoch repetition schedules.

### Key Theoretical Insights
- **Universality of Accuracy on the Line:** Under high-dimensional linear models, as $n, d \to \infty$ with $n/d \to \alpha^2$:
  $$\Phi^{-1}(\text{Acc}_{\theta_2, \rho_2}) = \frac{\cos(\theta_2, \theta) \rho_2}{\cos(\theta_1, \theta) \rho_1} \Phi^{-1}(\text{Acc}_{\theta_1, \rho_1}) + \mathcal{O}\left(\frac{e^{cn/d}}{\sqrt{n}}\right)$$
  The slope depends strictly on the data geometry and noise levels, remaining invariant to sample scale $n$ and algorithmic variations $\xi$.
- **Intermediate Slope from Mixing:**
  $$\text{Slope}_1 \le \text{Slope}\left(\mathcal{D}_1 \cup \mathcal{D}_2\right) \le \text{Slope}_2$$
  Training across mixed distributions mathematically bounds model robustness between the individual sources, explaining why naive aggregation fails to preserve peak robustness.
- **Model Specialization vs. Pretraining Utility:** Maximizing traditional human-alignment metrics (such as CIDEr via fine-tuning on MS-COCO) severely reduces synthetic caption vocabulary diversity ($n$-gram entropy), harming general zero-shot classification while boosting narrow in-domain retrieval.

## Conclusions
- **Quality Dominates Volume:** Indiscriminately expanding pretraining datasets via unweighted web scrapes dilutes out-of-distribution robustness. High-performing dataset design requires principled selection rather than passive scaling.
- **Diversity as an Essential Component of Quality:** Geographic, linguistic, and cultural diversities are not auxiliary fairness constraints; they provide mutually non-redundant training signals that boost performance across standard, Western-centric evaluation benchmarks.
- **Transition from Filtering to Transformation:** Purely subtractive curation pipelines discard invaluable visual and factual information. Generative recaptioning and guided LLM text rewriting effectively recycle discarded web artifacts, providing a sustainable, compute-efficient method to overcome the multimodal and language data wall.

## 1. High-Dimensional Linear Classification Framework (Section 2.7 & Appendix A.6)

### Data Distribution Model (Assumption 1)
* **Equation & Formulation:**
  A training dataset $\mathcal{D}_{n,\theta,\rho} = \{(x_i \in \mathbb{R}^d, y_i \in \{-1, +1\})\}_{i=1}^n$ consists of $n$ independent and identically distributed (i.i.d.) pairs where[cite: 55]:
  $$y_i \in \{-1, +1\} \quad \text{uniformly at random}$$
  $$x_i = y_i \theta + \left(\frac{\|\theta\|}{\rho}\right) z_i$$
  where $\theta \in \mathbb{R}^d$ is the ground-truth linear classifier[cite: 55], $z_i \in \mathbb{R}^d$ is zero-mean noise with independent entries of unit variance ($\mathbb{E}[z_i] = 0, \mathrm{Var}(z_{i,j}) = 1$)[cite: 55], $\|\theta\|^2$ is signal power[cite: 55], and $\rho^2 \in \mathbb{R}_+$ denotes the Signal-to-Noise Ratio ($\mathrm{SNR}$)[cite: 55].
* **Explanation:**
  This models binary classification as identifying a true direction $\theta$ embedded in Gaussian-like isotropic noise[cite: 55]. The parameter $\rho$ indicates dataset cleanliness or signal strength relative to noise[cite: 55].

---

### Trained Model Distribution (Assumption 2)
* **Equation & Formulation:**
  A learned linear model $\hat{\theta}_{n,\xi} \in \mathbb{R}^d$ predicts binary labels via $\mathrm{sign}(\langle x, \hat{\theta}_{n,\xi} \rangle)$[cite: 55]. It is assumed to be unbiased ($\mathbb{E}[\hat{\theta}_{n,\xi}] = \theta$) and takes the form[cite: 55]:
  $$\hat{\theta}_{n,\xi} = \theta + \left(\frac{\xi \|\theta\|}{\rho \sqrt{n}}\right) z$$
  where $z \in \mathbb{R}^d$ has zero mean and unit variance entries[cite: 55], and $\xi \in \mathbb{R}_+$ represents the algorithmic variability/randomness relative to the standard sample average estimator $\hat{\theta} = \frac{1}{n} \sum_{i=1}^n y_i x_i$ (where $\xi = 1$)[cite: 55, 56].
* **Explanation:**
  The variance of the learned parameter contracts at the standard rate of $1/\sqrt{n}$ with the sample size $n$[cite: 55], while $\xi$ encapsulates stochasticity introduced by architectural or algorithmic choices (such as optimization schedules or initialization)[cite: 55, 56].

---

### Accuracy on the Line & The Probit Transform (Theorem 1)
* **Equation & Formulation:**
  For test distributions $P_{\theta_1, \rho_1}$ and $P_{\theta_2, \rho_2}$, the accuracy is:
  $$\mathrm{Acc}_{\theta_k, \rho_k} := \mathbb{P}_{(X,Y) \sim P_{\theta_k, \rho_k}}\left(\mathrm{sign}(\langle X, \hat{\theta}_{n,\xi} \rangle) = Y\right)$$
  Using the standard normal cumulative distribution function (CDF) $\Phi(t) = \mathbb{P}(Z \le t)$ where $Z \sim \mathcal{N}(0,1)$, and its inverse (the probit function) $\Phi^{-1}$[cite: 56]:
  $$\lim_{d \to \infty} \mathrm{Acc}_{\theta_1, \rho_1} = \Phi\left(\cos(\theta_1, \theta) \frac{\rho_1 \rho \alpha}{\xi}\right), \quad \lim_{d \to \infty} \mathrm{Acc}_{\theta_2, \rho_2} = \Phi\left(\cos(\theta_2, \theta) \frac{\rho_2 \rho \alpha}{\xi}\right)$$
  where $\lim_{d \to \infty} \frac{n}{d} = \alpha^2$[cite: 56]. Under bounded third moments (Berry-Esseen theorem)[cite: 169], for universal constant $c > 0$[cite: 56]:
  $$\Phi^{-1}(\mathrm{Acc}_{\theta_2, \rho_2}) = \left(\frac{\cos(\theta_2, \theta) \rho_2}{\cos(\theta_1, \theta) \rho_1}\right) \Phi^{-1}(\mathrm{Acc}_{\theta_1, \rho_1}) + \mathcal{O}\left(\frac{e^{cn/d}}{\sqrt{n}}\right)$$
* **Explanation:**
  The inner product $\langle X, \hat{\theta} \rangle$ sums over $d$ features, yielding asymptotically normal error tails by the Central Limit Theorem[cite: 57]. The probit map linearizes accuracy into a signal-to-noise ratio[cite: 57]. The slope of the resulting linear relationship depends strictly on geometric alignment ($\cos(\theta_k, \theta)$) and test-set difficulties ($\rho_k$), rendering it invariant to sample size $n$ and algorithmic variations $\xi$[cite: 56, 57].

---

### Input and Output Mixing (Theorem 2 & Assumption 3)
* **Equation & Formulation:**
  When training on a combined dataset $\mathcal{D}_{n_1', \tilde{\theta}_1, \tilde{\rho}_1} \cup \mathcal{D}_{n_2', \tilde{\theta}_2, \tilde{\rho}_2}$, the effective parameters are[cite: 57]:
  $$\bar{\theta} = \frac{n_1' \tilde{\theta}_1 + n_2' \tilde{\theta}_2}{n_1' + n_2'}, \quad \bar{\rho} = \frac{\|\bar{\theta}\|}{\sqrt{\frac{n_1' \|\tilde{\theta}_1\|^2}{\tilde{\rho}_1^2} + \frac{n_2' \|\tilde{\theta}_2\|^2}{\tilde{\rho}_2^2}}}$$
  The resulting linear trend slope is[cite: 58]:
  $$\mathrm{Slope}\left(\hat{\theta}_{n_1' + n_2'}(\mathcal{D}_1 \cup \mathcal{D}_2)\right) = \frac{\cos(\theta_2, \bar{\theta}) \rho_2}{\cos(\theta_1, \bar{\theta}) \rho_1}$$
  Theorem 2 proves monotonicity[cite: 58]:
  $$\mathrm{Slope}_1 \le \mathrm{Slope}\left(\hat{\theta}_{n_1' + n_2'}(\mathcal{D}_1 \cup \mathcal{D}_2)\right) \le \mathrm{Slope}_2$$
* **Explanation:**
  The model's parameter becomes a convex combination of the parameters of the individual data sources[cite: 170]. Consequently, the effective robustness slope of a model trained on mixed data is bounded between the slopes of models trained on each source individually, explaining why naive data mixing dilutes peak out-of-distribution robustness[cite: 43, 58].

---

### Data Filtering via Pretrained Representations (Theorem 3 & Assumption 4)
* **Equation & Formulation:**
  Given a pretrained model parameter $\hat{\theta}_{\mathrm{pretrained}}$ satisfying[cite: 58]:
  $$\mathrm{Slope}(\hat{\theta}_{\mathrm{unfiltered}}) < \mathrm{Slope}(\hat{\theta}_{\mathrm{pretrained}}) \le 1$$
  Data points are retained with probability proportional to a monotonic score function $h(\langle x_i y_i, \hat{\theta}_{\mathrm{pretrained}} \rangle)$[cite: 59]. Because spherical Gaussians are rotationally invariant orthogonal to $\hat{\theta}_{\mathrm{pretrained}}$, the expectation of the filtered estimator is[cite: 170]:
  $$\mathbb{E}[\hat{\theta}_{\mathrm{filtered}}] = \theta_{\mathrm{train}} + c \hat{\theta}_{\mathrm{pretrained}} \quad (c > 0)$$
  Hence[cite: 59]:
  $$\mathrm{Slope}(\hat{\theta}_{\mathrm{unfiltered}}) < \mathrm{Slope}(\mathbb{E}[\hat{\theta}_{\mathrm{filtered}}]) \le \mathrm{Slope}(\hat{\theta}_{\mathrm{pretrained}})$$
* **Explanation:**
  Filtering with a model of superior robustness projects the training signal toward the pretrained reference direction, strictly improving the expected robustness slope of the retrained model over the unfiltered baseline[cite: 59, 170].

## 2. Statistical Metrics and Divergence Measures

### MAUVE Divergence (Section 3.6.2 & Table 3.2)
* **Concept:**
  A statistical divergence metric designed to measure the discrepancy between two probability distributions of text ($P$ and $Q$) by calculating soft Kullback-Leibler (KL) divergences in a quantized, low-dimensional embedding space[cite: 78]:
  $$\mathcal{D}_{\mathrm{MAUVE}}(P, Q) \in [0, 1]$$
* **Role in Paper:**
  Quantifies distributional divergence between native English captions and non-English captions translated into English[cite: 78]. Even after translation to English, a significant distributional divergence remained ($\mathrm{MAUVE} \approx 0.616$), proving that non-English web data contains fundamentally different semantic content[cite: 78].

---

### Spearman Rank-Order Correlation (Section 5.5.1 & Figure 5.3)
* **Concept:**
  Non-parametric measure of statistical dependence between the rankings of two variables:
  $$r_s = 1 - \frac{6 \sum d_i^2}{n(n^2 - 1)}$$
* **Role in Paper:**
  Assesses whether the quality of raw web text correlates with the quality of its LLM-rewritten counterpart. The correlation was found to be minimal ($r_s = 0.179$, $p = 6.52 \times 10^{-73}$), proving that low-quality web documents can be transformed into high-quality training tokens without being bottlenecked by the initial document score.

---

### N-Gram Diversity and Vocabulary Scaling (Sections 4.7.1, 5.5.3)
* **Concept:**
  The count of unique $n$-grams (specifically trigrams $N_3$ and bigrams $N_2$) as a function of the total token count or document sample size:
  $$\text{Diversity}(S) = |\mathrm{Unique}(n\text{-grams in } S)|$$
* **Role in Paper:**
  Serves as a proxy for linguistic and conceptual diversity[cite: 92]. It demonstrates that while synthetic captions (e.g., BLIP-2) have higher individual alignment scores, their global vocabulary diversity saturates faster than raw web crawl distributions[cite: 92, 96].

---

### Normalized Centered Accuracy (Section 5.3.3)
* **Concept:**
  Linear scaling of classification accuracy to normalize for chance:
  $$\mathrm{Acc}_{\mathrm{centered}} = \frac{\mathrm{Acc} - \mathrm{Acc}_{\mathrm{random}}}{1 - \mathrm{Acc}_{\mathrm{random}}}$$
* **Role in Paper:**
  Used for DCLM's CORE benchmark across 22 NLP tasks to create a low-variance composite benchmark score where $0$ represents random guessing and $1$ represents perfect accuracy[cite: 104].

## 3. Geometric and Multimodal Metrics

### Embedding Cosine Similarity
* **Equation:**
  $$\mathrm{sim}(u, v) = \frac{\langle u, v \rangle}{\|u\|_2 \|v\|_2} = \cos(\angle(u, v))$$
* **Role in Paper:**
  Measures image-text alignment in multimodal embedding spaces (CLIP, DFN) for threshold filtering[cite: 48, 70], SimCSE semantic sentence preservation during text rewriting, and translation semantic consistency in back-translation rounds.

### Probit-Space Regression Fit (Effective Robustness)
* **Concept:**
  Fitting an empirical line $y = m x + b$ where $x = \Phi^{-1}(\mathrm{Acc}_{\mathrm{ID}})$ and $y = \Phi^{-1}(\mathrm{Acc}_{\mathrm{OOD}})$[cite: 47, 56].
* **Role in Paper:**
  Quantifies distribution shift resilience independent of baseline architecture capacity or dataset size[cite: 48]. Movement toward the diagonal line $y = x$ denotes improved effective robustness[cite: 47].
