# Visual Object Recognition Using Generative Models of Images

**Author:** Vinod Nair (PhD Thesis, University of Toronto, 2010; supervised by Geoffrey Hinton)

---

## Abstract

The thesis examines whether generative and reconstructive models of images, which learn the latent structure explaining how an image was produced, can improve visual object recognition, a task traditionally solved with discriminative models of $$P(l \mid v)$$. Three classes of models are considered: supervised generative models of the joint distribution $$P(v, l)$$, unsupervised generative models of $$P(v)$$, and unsupervised reconstructive models (e.g., autoencoders) that reconstruct images without defining a proper distribution. Four modes of use are demonstrated empirically: (1) injecting complex domain knowledge by inverting a synthesis model, (2) using latent representations of generative models as features, (3) optimizing a hybrid generative-discriminative objective, and (4) generating synthetic training data for discriminative classifiers. The central conclusion is that such models are not only useful for recognition but can outperform purely discriminative models on difficult tasks, achieving near or at state-of-the-art results on MNIST and NORB.

---

## Problems

1. **Visual invariance.** Recognition must be invariant to lighting, viewpoint, deformation, and other nuisance variations while remaining sensitive to object identity.
2. **Apparent redundancy of generative modeling.** Classification requires only $$P(l \mid v)$$; fitting $$P(v, l)$$ or $$P(v)$$ in high-dimensional image space is harder and often intractable, so the value of generative models for recognition is not obvious.
3. **Limited domain knowledge in discriminative models.** Discriminative models admit only restricted priors (e.g., convolutional weight sharing, tangent-distance invariances), while rich knowledge naturally expressed as a graphics or physics program cannot be exploited directly.
4. **Inverting black-box synthesis models.** Training images lack ground-truth codes, the synthesizer's gradient is unavailable, and valid codes occupy a tiny region of code space, making naive sampling wasteful.
5. **Intractable mixtures of RBMs.** RBM partition functions are intractable, preventing their use as components in standard EM-trained mixture models.
6. **Poor mixing in DBN top-level training.** Contrastive Divergence (CD) learning of $$P(v, l)$$ tends to keep the label variable fixed, degrading the learned joint model.

## Purpose

The thesis aims to demonstrate empirically that generative and reconstructive models, together with unsupervised learning, have a substantive role in object recognition. A secondary aim is to provide a practical toolkit of techniques for building more accurate recognition systems, regardless of one's theoretical commitment to generative modeling.

---

## Methodology

### 1. Breeder Learning (Chapter 3)

The analysis model is a one-hidden-layer sigmoid network mapping a data vector to a code vector. Inputs are a black-box synthesizer, a hand-specified prototype code near the mode of the code distribution, and unlabeled images. Output biases are initialized so that the network initially outputs the prototype. Each iteration:

1. Infer codes for training images using the current analysis network.
2. Perturb codes with zero-mean spherical Gaussian noise.
3. Synthesize images from the perturbed codes with the black box.
4. Train the network to regress the perturbed code from the synthesized image (squared error); discard the pairs after one update.

The region of code space that the network handles correctly expands gradually outward from the prototype. After training, a synthesis network is fitted to emulate the black box using the analysis network's inferred codes, allowing closed-loop inference by gradient descent on pixel reconstruction error:

$$
\mathbf{c}^{*} = \arg\min_{\mathbf{c}} \; \lVert \mathbf{v} - \hat{g}(\mathbf{c}) \rVert^2, \quad \mathbf{c}_0 = f_{\text{analysis}}(\mathbf{v})
$$

**Applications:** a 2D eye-image model (Moriyama et al.; 8-dimensional code, 32 x 64 images) and an Active Appearance Model of faces (60-dimensional code, 30 x 30 images). Baseline: random-code learning, which samples codes around the prototype at a fixed variance.

### 2. Physics-Based Synthesis of Handwritten Digits (Chapter 4)

A pen of mass $$m$$ is driven by four antagonistic springs with time-varying stiffnesses $$k_i[n]$$. Along the x-axis:

$$
a_x[n] = \frac{k_1[n](\bar{x}_1 - x[n-1]) + k_2[n](\bar{x}_2 - x[n-1])}{m}
$$

$$
v_x[n] = (1 - \eta)\, v_x[n-1] + a_x[n], \qquad x[n] = x[n-1] + v_x[n]
$$

Ink of a given thickness and brightness is then applied along the trajectory. A motor program has dimension $$4n + 2$$ (70 for most classes, with $$n = 17$$). Class-specific extensions handle pen lifts (4s and 5s) and an extra stroke (dashed 1s and 7s). Twelve analysis networks (784-400-output) are trained on MNIST with breeder learning, using prototypes created through an interactive GUI.

The motor-program representation is evaluated for classification in three ways:

- **Energy-based:** 30 class-specific energies (patch-shifted reconstruction error, RBM free energy of the trajectory, PCA residual energy) fed into a 310-parameter logistic regression.
- **Synthetic data:** perturbing inferred motor programs produces 1.2 million "motor distortions", optionally combined with 0.5 million affine distortions, to train kNN, fully-connected networks, and convolutional networks.
- **Feature pre-training:** the 4,000 hidden features of the analysis networks initialize the first layer of a fully-connected network; 25 selected 5 x 5 kernels initialize a convolutional network.

### 3. Implicit Mixture of RBMs (Chapter 5)

The energy function introduces a 1-of-K cluster variable $$l$$ that gates the RBM weights:

$$
E(\mathbf{v}, \mathbf{h}, \mathbf{l}) = -\sum_{i,j,k} W^{I}_{ijk}\, v_i\, h_j\, l_k
$$

Mixing proportions are defined implicitly by the energy rather than parameterized explicitly. Component responsibilities are computed exactly from the free energies:

$$
P(l_k = 1 \mid \mathbf{v}) = \frac{\exp(-F(\mathbf{v}, l_k = 1))}{\sum_{k'} \exp(-F(\mathbf{v}, l_{k'} = 1))}, \qquad F(\mathbf{v}, l_k = 1) = -\sum_j \log\Big(1 + \exp\big(\textstyle\sum_i W^{I}_{ijk} v_i\big)\Big)
$$

Training uses CD, with component selection by sampling the responsibilities in both positive and negative phases. Unnormalized responsibilities serve as inputs to a logistic regression classifier on MNIST and MNORB (lighting-normalized, 32 x 32, monocular NORB).

### 4. DBNs with a Third-Order Top Level and Hybrid Learning (Chapter 6)

The implicit mixture model is reused as the top-level model of a DBN, with the label observed and $$K$$ equal to the number of classes. The hybrid objective is:

$$
\mathcal{L}(W) = \mathcal{L}_{CG}(W) + \lambda\, \mathcal{L}_{D}(W)
$$

$$
\mathcal{L}_{CG}(W) = -\sum_{n=1}^{N} \log P(\mathbf{v}_n \mid l_n, W), \qquad \mathcal{L}_{D}(W) = -\sum_{n=1}^{N} \log P(l_n \mid \mathbf{v}_n, W)
$$

The generative term is approximated by CD on the class-specific slice (avoiding label sampling and thus slow mixing); the discriminative gradient is exact. The best setting is $$\lambda = 5$$. Unlabeled cases contribute generative updates using sampled labels. A sparsity penalty targets an average activation of $$p = 0.1$$.

**Benchmark:** normalized-uniform NORB (5 classes, 24,300 training stereo pairs, disjoint object instances for testing), reduced to 8,976 dimensions with a foveal representation. For semi-supervised experiments, unlabeled data is generated by small global translations (25 times the training set).

---

## Conclusions

- Generative and reconstructive models contribute to object recognition in at least four complementary ways, and can match or surpass purely discriminative models on MNIST and NORB.
- Breeder learning provides a general, gradient-free method for inverting nonlinear black-box synthesis models, allowing rich domain knowledge such as physics and graphics to shape recognition. The resulting motor programs yield compact classifiers, highly effective synthetic data, and useful pre-trained features.
- Making mixing proportions implicit renders mixtures of RBMs tractable under CD, and their responsibilities capture class structure better than pixels or Mixtures of Factor Analyzers.
- A third-order top-level model with hybrid generative-discriminative training outperforms both purely generative and purely discriminative top levels in DBNs. With semi-supervised pre-training, it achieved the best published result on normalized-uniform NORB at the time (5.2%, or 5.0% with discriminative fine-tuning at the top level).
- **Limitations identified:** generic generative models approximate the true image distribution only crudely; generative models require segmented inputs because they attempt to explain all image structure, including background; and intractable likelihood evaluation, such as RBM partition functions, hinders principled model comparison.
- **Outlook:** larger unlabeled datasets, increased computational power, and better deep generative models with improved training and inference are expected to strengthen the role of unsupervised learning in recognition.

## 1. Generative vs. Discriminative Classification

### 1.1 Bayes' Rule for Generative Classifiers (Eq. 1.1)

$$
P(l \mid v, \theta_g) = \frac{P(v, l \mid \theta_g)}{\sum_{l'} P(v, l' \mid \theta_g)}
$$

- $$v$$: image; $$l$$: object label; $$\theta_g$$: parameters of a generative model.
- **Meaning:** A model of the joint distribution of images and labels can classify an image by normalizing the joint probability over all labels.
- **Role:** This is the basic justification for using generative models in recognition. A discriminative model learns $$P(l \mid v, \theta_d)$$ directly, while a generative model learns $$P(v, l)$$ and obtains $$P(l \mid v)$$ through this formula.

### 1.2 Maximum Likelihood Estimation

Both model types are fitted by choosing parameters that make the observed training data most probable. For generative models of images this is usually intractable, so the thesis relies on approximations such as Contrastive Divergence.

## 2. Background Mathematics of Invariance (Chapter 2)

### 2.1 First-Order Taylor Approximation of Transformations (Eq. 2.1)

$$
s(I, \alpha) \approx I + J\alpha, \qquad J = \left.\frac{\partial s(I, \alpha)}{\partial \alpha}\right|_{\alpha = 0}
$$

- $$s(I, \alpha)$$: the image $$I$$ transformed by parameters $$\alpha$$ (e.g., shift, rotation).
- $$J$$: the Jacobian, whose columns are tangent vectors.
- **Meaning:** Small transformations move an image along a flat plane (the tangent plane) in pixel space.
- **Role:** This is the basis of tangent distance (comparing images by the distance between tangent planes) and tangent propagation.

### 2.2 Tangent Propagation Regularizer

$$
E_r = \sum_{i=1}^{N} \left\lVert \left.\frac{\partial G_w(s(I_i, \alpha))}{\partial \alpha}\right|_{\alpha=0} \right\rVert^2
$$

- **Meaning:** A penalty that pushes the classifier output $$G_w$$ to remain unchanged when the input is slightly transformed.
- **Role:** A differentiable way of building known invariances into a classifier, presented as an alternative to adding transformed training images.

### 2.3 Lie Group Model with Matrix Exponential (Eqs. 2.2 and 2.3)

$$
s(I, \alpha) = \sum_{j=0}^{\infty} \frac{1}{j!} \left( \sum_{i=1}^{m} \alpha_i G_i \right)^{j} I = \exp\!\left( \sum_{i=1}^{m} \alpha_i G_i \right) I
$$

- $$G_i$$: generator matrices of the transformations.
- **Meaning:** If the tangent vectors depend linearly on the image, the full (non-truncated) Taylor series sums to a matrix exponential.
- **Role:** Defines a generative model in which an observed image is a "normalized" image transformed by $$\alpha$$, allowing large transformations.

### 2.4 Invariant Linear SVM (Eqs. 2.4 to 2.9)

Classifier with a linear preprocessor $$B$$:

$$
f(x) = \operatorname{sgn}\!\left( \sum_{i=1}^{l} \alpha_i y_i (Bx \cdot Bx_i) + b \right)
$$

Dual objective, maximized subject to $$\alpha_i \ge 0$$ and $$\sum_i \alpha_i y_i = 0$$:

$$
\sum_{i=1}^{l} \alpha_i - \frac{1}{2} \sum_{i,k=1}^{l} \alpha_i y_i \alpha_k y_k (Bx_i \cdot Bx_k)
$$

Invariance regularizer and tangent covariance matrix:

$$
\frac{1}{l} \sum_{i=1}^{l} \left( \left.\frac{\partial g(L_t x_i)}{\partial t}\right|_{t=0} \right)^2, \qquad C = \frac{1}{l} \sum_{i=1}^{l} \left( \left.\frac{\partial L_t x_i}{\partial t}\right|_{t=0} \right) \left( \left.\frac{\partial L_t x_i}{\partial t}\right|_{t=0} \right)^{T}
$$

- **Meaning:** $$C$$ is the covariance of tangent vectors. Choosing $$B$$ based on $$C$$ suppresses the directions in which transformations move the data, making the SVM approximately invariant.
- **Role:** A reviewed example of hand-designed invariance in discriminative models.

### 2.5 Third-Order Boltzmann Machine for Image Transformations (Eqs. 2.10 and 2.11)

$$
p(y, h \mid x) = \frac{\exp(-E(y, h; x))}{\sum_{y, h} \exp(-E(y, h; x))}, \qquad E(y, h; x) = -\sum_{i,j,k} W_{ijk}\, x_i\, y_j\, h_k
$$

- **Meaning:** Energy-based model in which three groups of units interact multiplicatively through a 3D weight tensor. When $$x$$ is fixed, the model becomes an ordinary RBM whose weights depend on $$x$$.
- **Role:** Precursor to the third-order models of Chapters 5 and 6.

### 2.6 Bilinear Style-Content Models (Eqs. 2.12 to 2.18)

$$
y = \sum_{i,j} w_{ij}\, a^{s}_i\, b^{c}_j \quad \text{(symmetric)}, \qquad y = \sum_{j} w^{s}_j\, b^{c}_j \quad \text{(asymmetric)}, \qquad w^{s}_j = \sum_i w_{ij} a^{s}_i
$$

Fitted with the Singular Value Decomposition:

$$
\bar{Y} = WB, \qquad \bar{Y} = USV^{T}, \qquad W = US,\; B = V^{T}
$$

- **Meaning:** Data are explained by two interacting factors (style and content). SVD gives the least-squares solution, and discarding small singular values reduces dimensionality. The symmetric model uses alternating SVDs with "vector transposes" and converges to a local minimum of squared error.
- **Role:** Background on multiplicative (higher-order) latent factor models.

## 3. Breeder Learning (Chapter 3)

### 3.1 Analysis Network Unit

$$
y = \frac{1}{1 + \exp\!\left(-b - \sum_{i=1}^{n} w_i x_i\right)}
$$

- **Meaning:** Logistic sigmoid neuron with output in $$[0, 1]$$.
- **Role:** Building block of the single-hidden-layer network that maps an image to its code vector.

### 3.2 Learning Procedure as a Statistical Process

1. Infer the code: $$\mathbf{c} = f(\mathbf{v})$$.
2. Perturb: $$\tilde{\mathbf{c}} = \mathbf{c} + \boldsymbol{\epsilon}$$, where $$\boldsymbol{\epsilon} \sim \mathcal{N}(\mathbf{0}, \sigma^2 I)$$. For sigmoid outputs the noise is added to the pre-activation, so codes stay in $$[0, 1]$$.
3. Synthesize: $$\tilde{\mathbf{v}} = g(\tilde{\mathbf{c}})$$ using the black box.
4. Minimize the squared regression error:

$$
\mathcal{E} = \lVert \tilde{\mathbf{c}} - f(\tilde{\mathbf{v}}) \rVert^2
$$

- **Assumption:** The data distribution is produced by a unimodal code distribution passed through a deterministic nonlinear synthesizer.
- **Initialization:** Output biases are set to the logit (inverse sigmoid) of the prototype code, so the network initially outputs the prototype.
- **Meaning:** The network creates its own labeled training pairs, which gradually spread out from the prototype in code space.
- **Role:** Learns to invert a synthesizer without its gradient and without ground-truth codes.

### 3.3 Evaluation Metric: Root Mean Squared Error

$$
\text{RMSE} = \sqrt{\frac{1}{N} \sum_{n=1}^{N} \lVert \mathbf{v}_n - g(f(\mathbf{v}_n)) \rVert^2}
$$

- **Role:** Measures reconstruction quality and is used for early stopping on a validation set and for comparing breeder learning with random-code learning.

### 3.4 Closed-Loop Iterative Inference

$$
\mathbf{c}_{t+1} = \mathbf{c}_t - \eta \, \frac{\partial}{\partial \mathbf{c}} \lVert \mathbf{v} - \hat{g}(\mathbf{c}_t) \rVert^2, \qquad \mathbf{c}_0 = f(\mathbf{v})
$$

- $$\hat{g}$$: a neural network trained to emulate the black box, which makes the gradient available through backpropagation.
- **Meaning:** Gradient descent on the code to reduce pixel reconstruction error, starting from the analysis network's estimate.
- **Role:** Improved test squared pixel error on faces by 5.41%.

## 4. Physics-Based Digit Synthesis (Chapter 4)

### 4.1 Hooke's Law (Eq. 4.1)

$$
F(t) = k(t)\,(x(t) - \bar{x})
$$

- **Meaning:** Spring force is proportional to stretch, scaled by a time-varying stiffness $$k(t)$$ representing muscle activation.

### 4.2 Initial Equilibrium Position (Eq. 4.2)

$$
x[1] = \frac{\bar{x}_1 k_1[1] + \bar{x}_2 k_2[1]}{k_1[1] + k_2[1]}
$$

- **Meaning:** The pen starts where opposing spring forces balance, which is a stiffness-weighted average of the rest positions.

### 4.3 Discrete-Time Dynamics (Eqs. 4.3 to 4.5)

$$
a_x[n] = \frac{k_1[n](\bar{x}_1 - x[n-1]) + k_2[n](\bar{x}_2 - x[n-1])}{m}
$$

$$
v_x[n] = (1 - \eta)\, v_x[n-1] + a_x[n], \qquad x[n] = x[n-1] + v_x[n]
$$

- $$m$$: pen mass; $$\eta$$: viscosity (damping).
- **Meaning:** A simple Euler integration of Newton's second law with friction. The y-axis is handled identically.
- **Role:** Converts the motor program (stiffness sequence) into a pen trajectory.

### 4.4 Motor Program Dimensionality

$$
\dim(\mathbf{c}) = 4n + 2
$$

- Four stiffnesses per time step plus ink thickness and brightness. With $$n = 17$$, the dimension is 70.

### 4.5 Inking by Convolution

The trajectory trace is convolved with a $$3 \times 3$$ kernel whose entries depend on thickness and brightness parameters, and the result is clipped to $$[0, 1]$$. This produces sharper stroke edges than Gaussian blurring.

### 4.6 Gaussian-Binary RBM Energy (Eq. 4.6)

$$
E(\mathbf{v}, \mathbf{h}) = \sum_i \frac{(v_i - b_i)^2}{2} - \sum_j b_j h_j - \sum_{i,j} v_i h_j W_{ij}
$$

- **Meaning:** Energy of an RBM with real-valued (unit-variance Gaussian) visible units and binary hidden units. Low energy indicates a likely configuration.
- **Role:** Models plausible pen trajectories for each digit class.

### 4.7 Free Energy (Eq. 4.7)

$$
F(\mathbf{v}) = \sum_i \frac{(v_i - b_i)^2}{2} - \sum_j \Big( P(h_j \mid \mathbf{v})\, t_j + P(h_j \mid \mathbf{v}) \log P(h_j \mid \mathbf{v}) + (1 - P(h_j \mid \mathbf{v})) \log(1 - P(h_j \mid \mathbf{v})) \Big)
$$

- $$t_j$$: total input to hidden unit $$j$$.
- **Meaning:** The "badness" of a visible vector after accounting for all hidden states: expected energy minus hidden-unit entropy.
- **Role:** One of three class-specific energy scores; unusual trajectories receive high free energy.

### 4.8 Additional Energy Scores

- **Patch-shifted squared error:** A greedy local search shifts $$5 \times 5$$ ink patches along the trajectory to minimize the sum of squared differences, reducing sensitivity to small misalignments (approximately 18% lower classification error than plain L2).
- **PCA residual energy:** For each class, PCA is fitted to residual images (image minus reconstruction). The score is the squared distance between a residual and its projection onto the PCA subspace.

### 4.9 Softmax Logistic Regression (Eq. 4.8)

$$
P(l_k = 1 \mid \mathbf{x}) = \frac{\exp\!\left(b_k + \sum_j W_{jk} x_j\right)}{\sum_{k'=1}^{K} \exp\!\left(b_{k'} + \sum_j W_{jk'} x_j\right)}
$$

- $$\mathbf{x}$$: the 30 energies (3 types times 10 classes).
- **Training:** Minimizes cross-entropy between predicted and true label distributions:

$$
\mathcal{L}_{CE} = -\sum_{n} \sum_{k} l^{(n)}_k \log P(l_k = 1 \mid \mathbf{x}^{(n)})
$$

- **Role:** A 310-parameter classifier that achieves 1.50% MNIST error.

### 4.10 Distance Metrics for kNN

$$
d_p(\mathbf{x}, \mathbf{y}) = \left( \sum_i |x_i - y_i|^p \right)^{1/p}, \qquad p \in \{2, 3\}
$$

- **Role:** L2 and L3 nearest-neighbour baselines; L3 performed slightly better, and both improved with motor-distortion data.

### 4.11 Kernel Selection Heuristics

- A $$5 \times 5$$ window with the largest L2 norm is cut from each learned feature.
- Kernels are accepted in descending norm order only if their cosine distance from all previously accepted kernels exceeds a threshold:

$$
d_{\cos}(\mathbf{a}, \mathbf{b}) = 1 - \frac{\mathbf{a} \cdot \mathbf{b}}{\lVert \mathbf{a} \rVert \, \lVert \mathbf{b} \rVert}
$$

- **Role:** Selects 25 diverse pre-trained kernels for the convolutional network.

## 5. Implicit Mixture of RBMs (Chapter 5)

### 5.1 Standard RBM Energy (Eq. 5.1)

$$
E(\mathbf{v}, \mathbf{h}) = -\sum_{i,j} W^{R}_{ij}\, v_i\, h_j
$$

An RBM with $$N$$ hidden units can be viewed as a mixture of $$2^N$$ Bernoulli models with shared parameters.

### 5.2 Third-Order Energy and Joint Distribution (Eqs. 5.2 to 5.4)

$$
E(\mathbf{v}, \mathbf{h}, \mathbf{l}) = -\sum_{i,j,k} W^{I}_{ijk}\, v_i\, h_j\, l_k, \qquad P(\mathbf{v}, \mathbf{h}, \mathbf{l}) = \frac{\exp(-E(\mathbf{v}, \mathbf{h}, \mathbf{l}))}{Z_I}
$$

$$
Z_I = \sum_{\mathbf{u}, \mathbf{g}, \mathbf{y}} \exp(-E(\mathbf{u}, \mathbf{g}, \mathbf{y}))
$$

- $$\mathbf{l}$$: a 1-of-K cluster indicator that selects one slice of the weight tensor.
- **Meaning:** Each slice $$W^{I}_{\cdot\cdot k}$$ is a separate RBM; all components share one global partition function $$Z_I$$.

### 5.3 Mixture Decomposition (Eq. 5.5)

$$
P(\mathbf{v}) = \sum_{\mathbf{h}, \mathbf{l}} P(\mathbf{v}, \mathbf{h}, \mathbf{l}) = \sum_{k} \sum_{\mathbf{h}} P(\mathbf{v}, \mathbf{h} \mid l_k = 1)\, P(l_k = 1)
$$

- **Meaning:** The model is a genuine mixture, but mixing proportions $$P(l_k = 1)$$ are implied by the energy function rather than stored as parameters.
- **Role:** Avoids computing each RBM's individual partition function, which makes the mixture tractable to train.

### 5.4 Log-Likelihood Gradient (Eq. 5.6)

$$
\frac{\partial L}{\partial W^{I}} = N \left\langle \frac{\partial E(\mathbf{v}, \mathbf{h}, \mathbf{l})}{\partial W^{I}} \right\rangle_{P(\mathbf{v}, \mathbf{h}, \mathbf{l})} - \sum_{n=1}^{N} \left\langle \frac{\partial E(\mathbf{v}^n, \mathbf{h}, \mathbf{l})}{\partial W^{I}} \right\rangle_{P(\mathbf{h}, \mathbf{l} \mid \mathbf{v}^n)}
$$

- **Meaning:** The gradient is the difference between expectations under the model (negative phase) and under the data (positive phase).
- **Difficulty:** The model expectation requires long Gibbs sampling, so it is approximated with Contrastive Divergence.

### 5.5 Conditional Sampling Distributions (Eqs. 5.7 and 5.8)

$$
P(v_i = 1 \mid \mathbf{h}, l_k = 1) = \sigma\!\left( \sum_j W^{I}_{ijk} h_j \right), \qquad P(h_j = 1 \mid \mathbf{v}, l_k = 1) = \sigma\!\left( \sum_i W^{I}_{ijk} v_i \right)
$$

- $$\sigma(x) = 1 / (1 + e^{-x})$$.
- **Meaning:** Given the selected component, units are conditionally independent Bernoulli variables (factorial distributions), which makes sampling fast.

### 5.6 Free Energy and Responsibilities (Eqs. 5.9 to 5.11)

$$
F(\mathbf{v}, l_k = 1) = -\sum_j \log\!\left( 1 + \exp\!\Big( \sum_i W^{I}_{ijk} v_i \Big) \right)
$$

$$
P(l_k = 1 \mid \mathbf{v}) = \frac{\exp(-F(\mathbf{v}, l_k = 1))}{\sum_{k'} \exp(-F(\mathbf{v}, l_{k'} = 1))}
$$

- **Meaning:** Hidden units are summed out analytically. The responsibility is a softmax over negative free energies, i.e., the posterior probability that component $$k$$ generated $$\mathbf{v}$$.
- **Role:** Responsibilities are used for component selection during training and as classification features.

### 5.7 Contrastive Divergence Gradient (Eq. 5.12)

$$
\frac{1}{N_b} \frac{\partial L}{\partial W^{I}_{k}} \approx \frac{1}{N_b} \left( \sum_{i=1}^{M} D^{+}_{ki} - \sum_{j=1}^{Q} D^{-}_{kj} \right), \qquad D^{+} = \mathbf{v}^{+} \mathbf{h}^{+T}, \; D^{-} = \mathbf{v}^{-} \mathbf{h}^{-T}
$$

- **Meaning:** Data-driven outer products minus one-step reconstruction outer products, accumulated per component. Positive and negative phases may select different components, so $$M$$ and $$Q$$ can differ.
- **Role:** A practical approximate maximum likelihood update.

### 5.8 Temperature-Scaled Responsibilities (Eq. 5.13)

$$
P(l_k = 1 \mid \mathbf{v}) = \frac{\exp(-F(\mathbf{v}, l_k = 1)/T)}{\sum_{k'} \exp(-F(\mathbf{v}, l_{k'} = 1)/T)}
$$

- **Meaning:** A temperature $$T > 1$$ flattens the softmax so that random scale differences in free energies do not cause one component to absorb all the data.
- **Role:** Prevents mixture collapse; a constant $$T$$ worked as well as annealing.

### 5.9 Comparison Models

- **Mixture of Factor Analyzers (MFA):** A directed, linear-Gaussian mixture trained with the EM algorithm.
- **Mixture of multivariate Bernoullis:** Equivalent to an implicit mixture with no hidden units; also trained with EM.
- **EM algorithm:** The E-step computes responsibilities; the M-step updates parameters and mixing proportions using responsibility-weighted data.

## 6. Deep Belief Nets with Hybrid Learning (Chapter 6)

### 6.1 Third-Order Top-Level Model (Eqs. 6.1 and 6.2)

$$
E(\mathbf{v}, \mathbf{h}, \mathbf{l}) = -\sum_{i,j,k} W_{ijk}\, v_i\, h_j\, l_k, \qquad P(\mathbf{v}, \mathbf{h}, \mathbf{l}) = \frac{\exp(-E(\mathbf{v}, \mathbf{h}, \mathbf{l}))}{Z}
$$

- Same form as Chapter 5, but $$\mathbf{l}$$ is now the observed class label.
- **Meaning:** The label multiplicatively selects which weights connect penultimate features $$\mathbf{v}$$ to top-level hidden units $$\mathbf{h}$$.

### 6.2 Exact Classification (Eqs. 6.3 and 6.4)

$$
F_k(\mathbf{v}) = -\sum_j \log\!\left( 1 + \exp\!\Big( \sum_i W_{ijk} v_i \Big) \right), \qquad P(l_k = 1 \mid \mathbf{v}) = \frac{\exp(-F_k(\mathbf{v}))}{\sum_{c=1}^{N_l} \exp(-F_c(\mathbf{v}))}
$$

- **Meaning:** Classification is exact and costs $$O(N_v N_h N_l)$$ operations.
- **Role:** Used at test time and to compute the exact discriminative gradient.

### 6.3 CD Update for the Generative Model (Eqs. 6.5 and 6.6)

$$
\Delta W_{\cdot,\cdot,k} = D^{+}_k - D^{-}_k, \qquad W_{\cdot,\cdot,k} \leftarrow W_{\cdot,\cdot,k} + \eta\, \Delta W_{\cdot,\cdot,k}
$$

- **Issue:** When learning the full joint $$P(\mathbf{v}, \mathbf{l})$$, the Markov chain rarely changes the label (poor mixing), effectively training independent class-conditional RBMs whose energies are not comparable.

### 6.4 Loss Functions (Eqs. 6.8 and 6.9)

$$
\mathcal{L}_{D}(W) = -\sum_{n=1}^{N} \log P(l_n \mid \mathbf{v}_n, W), \qquad \mathcal{L}_{CG}(W) = -\sum_{n=1}^{N} \log P(\mathbf{v}_n \mid l_n, W)
$$

- $$\mathcal{L}_{D}$$: discriminative negative log-likelihood, with an exact gradient.
- $$\mathcal{L}_{CG}$$: class-conditional generative negative log-likelihood, with a gradient approximated by CD.

### 6.5 Hybrid Objective and Update (Eqs. 6.7 and 6.10)

$$
\mathcal{L}_{H}(W) = \mathcal{L}_{CG}(W) + \lambda\, \mathcal{L}_{D}(W)
$$

$$
W_{\cdot,\cdot,c} \leftarrow W_{\cdot,\cdot,c} + \eta \left( \Delta W^{g}_{\cdot,\cdot,c} + \lambda\, \Delta W^{d}_{\cdot,\cdot,c} \right)
$$

- **Meaning:** A weighted combination of generative and discriminative learning signals. It resembles pseudo-likelihood, since it uses the two conditionals $$P(\mathbf{v} \mid l)$$ and $$P(l \mid \mathbf{v})$$ instead of the joint.
- **Theoretical support:** Bishop and Lasserre (2007) interpret such blending as learning a joint model that compensates for model mis-specification.
- **Hyperparameter:** $$\lambda \in [0.1, 20]$$ was tested; $$\lambda = 5$$ was best. The generative gradient's norm can be up to two orders of magnitude larger, so $$\lambda$$ does not directly indicate relative influence.

### 6.6 Semi-Supervised Extension

For an unlabeled image $$\mathbf{v}^u$$, a label is sampled as $$\hat{l} \sim P(l \mid \mathbf{v}^u)$$ and used for the generative update. The discriminative update is zero. This treats missing labels as latent variables, an imputation approach.

### 6.7 Sparsity Penalty

Cross-entropy between a target activation probability $$p$$ and the running average $$q$$:

$$
\mathcal{C} = -\left[ p \log q + (1 - p) \log(1 - q) \right], \qquad \frac{\partial \mathcal{C}}{\partial (\text{total input})} \propto p - q
$$

$$
q_{\text{new}} = 0.9\, q_{\text{old}} + 0.1\, q_{\text{current}}
$$

- $$p = 0.1$$; $$q_{\text{current}}$$ is computed per mini-batch of 100.
- **Meaning:** An exponential moving average tracks how often each hidden unit fires; the penalty nudges it toward $$p$$.
- **Role:** Produces rarely active, more discriminative features and revives inactive units.

### 6.8 Dimensionality Reduction by Foveal Averaging

The central $$64 \times 64$$ region is kept at full resolution, while outer rings are replaced by block averages ($$8 \times 8$$, $$4 \times 4$$, $$2 \times 2$$). The input dimension is reduced from 18,432 to 8,976.

## 8. Limitations Expressed Mathematically

- **Intractable partition function:**

$$
Z = \sum_{\mathbf{v}, \mathbf{h}} \exp(-E(\mathbf{v}, \mathbf{h}))
$$

  Summing over exponentially many states makes $$P(\mathbf{v}) = \frac{1}{Z} \sum_{\mathbf{h}} e^{-E(\mathbf{v}, \mathbf{h})}$$ impossible to compute exactly for realistic RBMs.

- **Consequence:** Generative models cannot be compared directly by held-out likelihood. Annealed Importance Sampling can estimate $$Z$$ but may be unreliable without warning, and upper bounds on $$Z$$ (which would give lower bounds on likelihood) are also intractable.
- **Misleading proxy:** Reconstruction error during CD training can improve simply because the Markov chain mixes poorly, so it is only an indirect quality indicator.

# Research Gaps, Limitations of Prior Work, and Proposed Solutions

**Paper:** Visual Object Recognition Using Generative Models of Images (Vinod Nair, PhD Thesis, University of Toronto, 2010)

| # | Problem / Research Gap | Limitation of Prior Work | Proposed Solution in the Thesis |
|---|---|---|---|
| 1 | Unclear value of generative models for recognition | Classification requires only $$P(l \mid v)$$; fitting $$P(v, l)$$ or $$P(v)$$ in high-dimensional image space is harder and often intractable, so discriminative models dominated recognition research | Empirical demonstration of four distinct uses of generative and reconstructive models (domain knowledge via synthesis inversion, latent features, hybrid objectives, synthetic data), showing they can match or outperform discriminative models on MNIST and NORB |
| 2 | Restricted domain knowledge in discriminative models | Prior knowledge enters only in narrow forms, such as convolutional weight sharing or tangent-based invariances (tangent distance, tangent propagation, invariant SVMs); rich knowledge encoded in graphics or physics programs cannot be used directly | Analysis-by-synthesis: domain knowledge is expressed as a synthesis model, and recognition is performed by learning its inverse |
| 3 | Inverting black-box synthesis models | Autoencoder-style learning needs the synthesizer's gradient, which complex graphics programs lack; finite differencing is inefficient; training images lack ground-truth codes; uniform code sampling wastes capacity because valid codes occupy a tiny region of code space | **Breeder learning:** starting from a hand-set prototype, the analysis network infers codes, perturbs them with Gaussian noise, synthesizes images, and regresses the perturbed codes, progressively expanding coverage of the relevant code space (lower test RMSE than random-code learning on eye and face models) |
| 4 | Purely feed-forward (open-loop) inference | One-step bottom-up inference cannot exploit top-down knowledge to correct errors, and gradient-based refinement is impossible without a differentiable synthesizer | A neural network emulator of the black box is trained on codes supplied by the analysis network, enabling gradient descent on pixel reconstruction error (5.41% lower test squared error on faces) |
| 5 | Lack of structured, physically meaningful representations for handwriting | Pixel-based classifiers and affine distortions do not capture how digits are actually produced, limiting the realism of augmentation and feature learning | A mass-spring physics model of pen movement yields "motor programs"; these support a 310-parameter energy-based classifier (1.50% error), realistic motor distortions that improve every classifier tested, and pre-trained features (0.53% MNIST error with a convolutional net) |
| 6 | Intractable mixtures of RBMs | Standard EM-based mixtures require normalized component densities, but RBM partition functions are intractable, so RBMs could not serve as mixture components | **Implicit mixture of RBMs:** a third-order Boltzmann machine with implicitly defined mixing proportions, trainable with Contrastive Divergence and exact responsibility computation; responsibilities outperform pixels and MFA features for classification |
| 7 | Mixture collapse during training | Unnormalized free energies differ in scale across components, causing one component to absorb all data | Temperature-scaled responsibilities that soften the softmax over free energies |
| 8 | Limited top-level models in Deep Belief Nets | Earlier DBNs used a bipartite RBM with an additive label unit, which models label-feature interactions weakly | A third-order RBM top level in which the label multiplicatively gates visible-hidden interactions (7.6% vs. 11.9% NORB error under CD) |
| 9 | Poor mixing of CD when learning $$P(v, l)$$ | The Markov chain rarely changes the label, effectively learning independent class-conditional models with incomparable energies; purely discriminative top levels overfit and cannot use unlabeled data | **Hybrid learning:** a weighted objective $$\mathcal{L}_{CG} + \lambda \mathcal{L}_{D}$$ combining CD for $$P(v \mid l)$$ with the exact gradient of $$P(l \mid v)$$ (6.5% NORB error, outperforming both pure CD and pure discriminative training) |
| 10 | Insufficient use of unlabeled data | Discriminative models such as convolutional networks and SVMs cannot learn from unlabeled images, and labeled sets like NORB are small | Semi-supervised DBN training: unlabeled data improves greedy pre-training and contributes generative updates with sampled labels (5.2% NORB error; 5.0% after discriminative top-level training initialized from the generative model, the best published result at the time) |
| 11 | Weak or uninterpretable hidden features | Densely active binary features are less discriminative, and some hidden units become inactive | A sparsity penalty based on cross-entropy between a target and a running-average activation probability |

| Remaining Gap | Effect | Suggested Direction |
|---|---|---|
| Generic generative models approximate the true image distribution poorly | Limited fidelity of learned image models | Build in domain knowledge (breeder learning) or scale up deep generative models (DBNs, Deep Boltzmann Machines) |
